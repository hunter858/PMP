# 第五章：项目范围管理 · 029

在创建WBS的过程中，需要使用以下哪个项目文件?()

- A. 范围管理计划
- B. 需求管理计划
- C. 项目范围说明书
- D. 需求跟踪矩阵

## 答案

**C**

## 解析

- **C 正确**：创建 WBS = 工作分解，但必须按范围边界拆；依据是**项目范围说明书**（项目文件）。
- **A 不选**：可能是过程输入，但属于**项目管理计划**，本题问「项目文件」，故不选。
- **B 错**：需求管理计划不指导如何分解 WBS。
- **D 错**：跟踪矩阵偏确认/控制范围，不是本题要的创建 WBS 项目文件。

**考点：** 创建 WBS 的项目文件输入；区分「项目文件」与「项目管理计划」。进基线后范围说明书进入范围基准；创建 WBS 当时按项目文件考。

材料常注《指南》约第 157 页。

## 对比要点

**题目问「项目文件」时，不要选管理计划类输入。**

| 选项 | 类别 | 与创建 WBS 的关系 |
| --- | --- | --- |
| 范围管理计划 | 项目管理计划 | 可能是过程输入，但不是「项目文件」 |
| 需求管理计划 | 项目管理计划 | 不指导如何分解 WBS |
| 项目范围说明书 | 项目文件 | 给出范围边界，创建 WBS 依此拆解 |
| 需求跟踪矩阵 | 项目文件 | 偏确认/控制范围，非本题要点 |
