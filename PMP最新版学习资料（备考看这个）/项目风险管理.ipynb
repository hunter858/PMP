# 第十一章：项目风险管理

每道题包含：题目、选项、正确答案、为什么选这个、为什么不选其他。
本节先收录已做过的题目，后续继续追加。


## 001. 关于风险应对措施，以下哪个说法是正确的？( )

**选项**
- A. 对每个风险都要采取最严厉的应对措施
- B. 对重要风险要指定不止一名风险责任人
- C. 风险应对措施需要全体相关方的同意
- D. 在规划风险管理过程中制定风险应对措施

**正确答案：B**

**为什么选这个**
重要风险影响大，常需明确多名相关责任角色（如风险责任人及具体应对行动负责人），确保有人盯、有人落实，责任覆盖更充分。

**为什么不选其他**
- 不选 A：应对应与风险重要性/优先级相匹配并考虑成本效益，不是每个都用最严厉措施。
- 不选 C：不必取得全体相关方同意，通常与关键相关方协商并按规定批准即可。
- 不选 D：制定风险应对措施属于**规划风险应对**，不是规划风险管理（后者是定义如何做风险管理）。

**出处**
PMBOK® Guide 第6版 11.5 规划风险应对：风险责任人与应对行动。
