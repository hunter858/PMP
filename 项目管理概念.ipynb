# 001. 项目生命周期

项目正式开始之前，先做前期准备。项目自己的通用生命周期有 4 个阶段：

1. 开始项目
2. 组织与准备
3. 执行项目工作
4. 结束项目

这 4 个是生命周期阶段，不是五大过程组。过程组见 004。

| 阶段 | 这时编制或产生的 | 对应过程组 |
| --- | --- | --- |
| 前期准备（项目还没开始） | 需要评估，然后是商业论证，再是效益管理计划 | 还没有项目 |
| 开始项目 | 项目章程。识别相关方，得到相关方登记册 | 启动 |
| 组织与准备 | 项目管理计划、需求文件。还会排出进度、成本等计划 | 规划 |
| 执行项目工作 | 可交付成果、工作绩效数据 | 执行，同时做监控 |
| 结束项目 | 最终产品、服务或成果，最终报告 | 收尾 |

需要评估、商业论证、效益管理计划是商业文件。发起人负责维护商业论证。项目经理不能直接改这些商业文件，只能提建议。三份文件编在前期准备，整个项目期间还要反复查看。结束时拿结果和商业论证里的目标对比，判断项目是否成功。效益常常在项目结束后才实现，效益管理计划会继续用到项目结束之后。

两个名字不要并成一件事：

- 需要评估（Needs Assessment）在前期准备里做。弄清业务目的、问题和机会，并提出处理建议。它排在商业论证之前，结果可以写进商业论证。
- 需求文件（Requirements Documentation）在组织与准备阶段编制，是规划过程组里「收集需求」的输出。它记录相关方的需要和需求，用来定义范围。

因此，应该在前期准备阶段编制的文件是需要评估。项目章程在开始项目时编制，项目管理计划和需求文件在组织与准备时编制。

# 003. 项目、项目集、项目组合与组织级项目管理

| | 管什么 |
| --- | --- |
| 项目管理 | 用正确的方式完成单个项目，实现项目目标 |
| 项目集管理 | 完成一系列相互关联的项目，获得单独管理得不到的更大收益 |
| 项目组合管理 | 选择最有利于实现战略目标的项目集和项目，也就是做正确的项目 |
| 组织级项目管理 | 为使项目符合组织的战略业务目标，对项目组合、项目集和项目做系统化管理 |

项目之间如果只共享顾主、供应商、技术或资源，没有必须的内在联系，当作项目组合来协调管理，不当作项目集。

# 005. 工作绩效数据、工作绩效信息、工作绩效报告

051. 工作绩效报告的主要用途是便于制定决策和采取行动。

三者是一条加工链：执行中记下原始数，控制过程里和计划对比、看出偏差，再汇编成报告交给相关方。

| | 区别 | 主要用途 | 从哪个过程出来 |
| --- | --- | --- | --- |
| 工作绩效数据 | 原始观察和测量，还没分析 | 实时显示项目的进展情况。如实际工时、实际成本、完成百分比、缺陷个数 | 指导与管理项目工作 |
| 工作绩效信息 | 把数据放进背景、和基准对比后得到的结论 | 显示已经发生的绩效偏差。如进度落后 2 天、成本超支 10% | 控制范围、控制进度、控制成本等监控过程 |
| 工作绩效报告 | 把信息汇编成的口头说明或书面文件 | 便于制定决策、提出问题、采取行动或引起关注。如状态报告、备忘录、仪表盘 | 监控项目工作 |

显示项目工作安排的是项目计划，不是这三者。

项目数据、项目信息、项目报告是上一层说法。工作绩效数据、工作绩效信息、工作绩效报告分别是它们在绩效上的具体形式。

# 007. 精益六西格玛价值流图

某组织实施一个运用精益六西格玛价值流图的项目，属于「创造、改进或修复产品、过程或服务」。

价值流图用来看清一个过程现在怎么跑，找出浪费，再改这个过程。所以它落在「改进过程」这一类，而不是合规要求，也不是在执行一项新的业务或技术战略。

| 词 | 指什么 |
| --- | --- |
| 精益 | 去掉不给客户增加价值的环节，让价值连续流下去。常见浪费：等待、库存、缺陷、过量生产、多余搬运、多余动作、过度加工 |
| 六西格玛 | 用数据把缺陷和波动降下来。步骤常记为 DMAIC：定义、测量、分析、改进、控制 |
| 价值流图 | 从客户提出需求一直画到交付。一张图上同时有物料怎么流、信息怎么流 |

图上要看的数：

- C/T：周期时间，做完一件要多久
- C/O：换型时间，换一种产品前的准备时间
- 可用率、一次合格率：设备能不能干、做出来的有多少不用返工
- 黄色三角：工序之间堆着的在制品，或人在等
- 时间线：绿色是增值（真正在加工），灰色是非增值（等待、库存）

# 009. PMO 的类型

003. 某公司的 PMO 虽然不直接管控项目，但有权要求项目采用公司特定的项目管理方法论。这是什么类型的 PMO？

答案：控制型。《指南》里只有三种 PMO，按对项目的控制程度从低到高排列。没有教练型。

| 类型 | 控制程度 | 特征 |
| --- | --- | --- |
| 支持型 | 低 | 当顾问。提供模板、最佳实践、培训、信息和其他项目的经验教训，相当于项目资源库。无权要求项目服从 |
| 控制型 | 中 | 既提供支持，也要求项目服从。可以要求采用特定的项目管理框架或方法论、使用特定模板、格式和工具，并遵从治理框架。不直接管理项目 |
| 指令型 | 高 | 直接管理和控制项目。项目经理由 PMO 指定，并向 PMO 汇报 |

这道题的关键是两句话：不直接管控项目，排掉指令型；有权要求项目采用特定方法论，不是只提供支持，所以是控制型。


# 010. 组织结构类型

005. 强矩阵组织和弱矩阵组织之间的主要差别在于项目经理对项目资源的控制程度不同。

《指南》表 2-1 按项目经理能拿到多少职权和资源，列了 10 种组织结构。从职能型到项目型，项目经理的职权和资源从几乎没有升到几乎全部。

| 类型 | 项目经理职权 | 资源可用性 | 谁管预算 | 项目经理的角色 | 项目管理行政人员 |
| --- | --- | --- | --- | --- | --- |
| 有机型或简单型 | 很少或没有 | 很少或没有 | 所有者或运营者 | 兼职，不一定有正式职位 | 很少或没有 |
| 职能型 | 很少或没有 | 很少或没有 | 职能经理 | 兼职，不一定有正式职位 | 兼职 |
| 多部门型 | 很少或没有 | 很少或没有 | 职能经理 | 兼职，不一定有正式职位 | 兼职 |
| 弱矩阵 | 低 | 低 | 职能经理 | 兼职，不是正式职位，常叫协调员 | 兼职 |
| 平衡矩阵 | 低到中 | 低到中 | 职能经理和项目经理混合 | 兼职，嵌在职能部门里 | 兼职 |
| 强矩阵 | 中到高 | 中到高 | 项目经理 | 全职，是正式职位 | 全职 |
| 项目型 | 高到几乎全部 | 高到几乎全部 | 项目经理 | 全职，是正式职位 | 全职 |
| 虚拟型 | 低到中 | 低到中 | 混合 | 全职或兼职 | 全职或兼职 |
| 混合型 | 混合 | 混合 | 混合 | 混合 | 混合 |
| PMO | 高到几乎全部 | 高到几乎全部 | 项目经理 | 全职，是正式职位 | 全职 |

强矩阵下，项目经理职权中到高，能控制较多资源，自己管预算，带的是全职团队。弱矩阵下，职权低，资源很少，预算在职能经理手里，全职团队成员很少，项目经理多半兼职。

职能型几乎不给项目经理职权，人和预算都在职能经理手里。项目型几乎把职权和资源都交给项目经理。平衡矩阵夹在弱矩阵和强矩阵中间，预算由双方分着管。


# 011. 风险、制约因素、事业环境因素

006. 经过实地考察，项目经理发现当地关税非常高，原材料进口很不划算。这个发现应该被记录为事业环境因素。

关税已经查明，是组织外部、项目团队控制不了的条件，属于事业环境因素里的财务考虑因素。

| | 是什么 | 怎么认 |
| --- | --- | --- |
| 事业环境因素 | 项目团队不能控制的条件，会影响、限制或指令项目 | 已经存在的环境。关税、汇率、利率、法律、市场、行业标准、组织文化、人员能力都算 |
| 制约因素 | 已经落到这个项目上的具体限制 | 比事业环境因素更具体。事先定下的预算、强制性里程碑、合同条款，以及范围、进度、成本、质量、资源、风险这些相互竞争的限制 |
| 风险 | 不确定的事件或条件，一旦发生，会对项目目标产生正面或负面影响 | 必须还没有定。已经查明的高关税没有不确定性，不记成风险 |

事业环境因素分组织内部和组织外部。

- 组织内部：组织文化、结构和治理，设施和资源的地理分布，基础设施，信息技术软件，资源可用性，员工能力。
- 组织外部：市场条件，社会和文化影响，法律限制，商业数据库，学术研究，政府或行业标准，财务考虑因素（汇率、利率、通货膨胀、关税、地理位置），物理环境。

组织过程资产是组织自己积累的流程、政策、程序和知识库。关税不是组织内部的资产。


# 012. 项目治理的领域和职能

044. 项目治理必须履行的四大职能是监督、控制、整合和决策。

项目治理是用来指导项目管理活动的框架、功能和过程，使项目能交出符合组织、战略和运营目标的产品、服务或成果。

治理先分四个领域，每个领域都要做同样的四件事。

| | 内容 |
| --- | --- |
| 四个领域 | 一致性、风险、绩效、沟通 |
| 四大职能 | 监督、控制、整合、决策 |

题目问的是职能，不是领域。管理、规划、总结都不在这四个职能里。
