# 第三章：项目经理角色

每道题包含：题目、选项、正确答案、为什么选这个、为什么不选其他。

### 001. 不同组织对项目经理的要求不一样，为了项目的成功，项目经理可能需要（）。

**选项：**

- A. 参与项目前期的评估和分析
- B. 组织专家对项目开展商业论证
- C. 在项目章程发布前不参与项目活动
- D. 负责跟进项目后的商业价值实现

**正确答案：A. 参与项目前期的评估和分析**

**为什么选这个：**

不同组织对项目经理介入时机要求不同。为了项目成功，项目经理常常需要在项目被正式批准、章程发布之前，就参与评估和分析（甚至协助起草章程），而不是等章程出来才上手。

**为什么不选其他：**

- 不选 B：商业论证通常由发起人、商业分析师或组织专家负责，不是项目经理去组织专家做论证。
- 不选 C：章程发布前不参与，与《指南》相反；很多组织恰恰要求项目经理早期介入。
- 不选 D：项目结束后的商业价值实现由发起人/效益责任人/运营跟进，不是项目经理长期盯效益兑现。

**记忆要点：** 项目经理可以、也常常应该在章程前介入评估分析；商业论证和项目后效益归发起人。

### 003. 带领敏捷团队时，服务型项目经理应该（）。

**选项：**

- A. 为团队分配工作内容
- B. 要求团队自主决策并解决问题
- C. 制定有效的奖励措施
- D. 为团队扫除妨碍工作进展的障碍

**正确答案：D. 为团队扫除妨碍工作进展的障碍**

**为什么选这个：**

服务型领导（Servant leadership）的核心是服务团队：清除障碍、提供资源、支持协作，让团队把活干好。敏捷里项目经理/Scrum Master 最典型动作就是扫除妨碍进展的障碍。

**为什么不选其他：**

- 不选 A：给团队分配工作是命令控制型，敏捷团队应自组织认领工作。
- 不选 B：自主决策是团队能力，服务型领导是赋能促进，不是「要求」团队必须自己决策；措辞偏指令。
- 不选 C：制定奖励措施属于传统激励管理，不是服务型领导的核心动作。

**记忆要点：** 服务型领导 = 清障、服务、赋能；不是派活、不是发号施令。

### 005. 杰克作为项目经理，带领大家按照计划实现了项目目标。但团队成员却对他颇有微词，不愿意再次跟他合作下一个项目。这说明杰克（）。

**选项：**

- A. 作为管理者是失败的
- B. 作为领导者是失败的
- C. 既没有当好管理者也没有当好领导者
- D. 没有管理好项目的负面相关方

**正确答案：B. 作为领导者是失败的**

**为什么选这个：**

管理偏做事：按计划达成目标；领导偏做人：影响、激励、凝聚人心。目标完成说明管理成功；成员不愿再合作说明领导失败。

**为什么不选其他：**

- 不选 A：按计划实现目标，说明管理者角色是成功的。
- 不选 C：管理成功、领导失败，不是两者都失败。
- 不选 D：题干说的是团队成员，不是负面相关方；核心考点是管理 vs 领导。

**记忆要点：** 事做成了但人不愿跟 → 领导失败；人愿意跟但事没做成 → 管理失败。

### 006. 项目经理应该如何处理与项目所在组织的关系？（）

**选项：**

- A. 要求所在组织采用项目型组织结构
- B. 与项目发起人合作处理会影响项目的政治和战略问题
- C. 时刻关注行业的最新发展趋势
- D. 防止组织中的其他项目经理干扰本项目

**正确答案：B. 与项目发起人合作处理会影响项目的政治和战略问题**

**为什么选这个：**

项目经理与执行组织的关系中，政治、权力、战略对齐等问题超出项目经理单独可控范围，应与发起人合作处理，让项目获得组织支持。

**为什么不选其他：**

- 不选 A：组织结构是事业环境因素，项目经理无权要求组织改成项目型。
- 不选 C：关注行业趋势属于与行业/专业领域的关系，不是与本组织关系的重点。
- 不选 D：应协作共享，不是防其他项目经理「干扰」。

**记忆要点：** 组织内政治战略 → 找发起人合作；改结构、防同事 → 都不对。

### 007. PMI人才三角指出了项目经理需要具备的三个关键技能，其中不包括（）。

**选项：**

- A. 项目管理专业技能
- B. 领导力
- C. 战略和商务管理
- D. 影响力

**正确答案：D. 影响力**

**为什么选这个：**

题干问「不包括」。PMI 人才三角三边是：技术项目管理、领导力、战略与商务管理。影响力是领导力技能的一部分，不是独立的一角。

**为什么不选其他：**

- 不选 A：技术项目管理（专业技能）是三角之一。
- 不选 B：领导力是三角之一。
- 不选 C：战略和商务管理是三角之一。

**记忆要点：** 人才三角 = 技术 + 领导 + 商务战略。影响力归领导力，不单独成角。

### 008. 在一个需要高度创新的科研项目上，项目经理应该主要使用哪种领导风格？（）

**选项：**

- A. 放任型
- B. 服务型
- C. 交易型
- D. 交互型

**正确答案：A. 放任型**

**为什么选这个：**

放任型（Laissez-faire）允许团队自己做决策、自己定目标，适合高度专业、需要创新的场景。科研创新项目专家能力强，项目经理应少指挥、多给自由空间。

**为什么不选其他：**

- 不选 B：服务型侧重清障赋能，敏捷常用，但「高度创新科研」对应的经典配对是放任型。
- 不选 C：交易型用奖惩换绩效，适合目标清晰、重复性强的工作，不利于创新。
- 不选 D：交互型是交易型+变革型+魅力型的综合，不是高度创新场景的首选标签。

**记忆要点：** 创新/专家团队 → 放任型；敏捷清障 → 服务型；奖惩换结果 → 交易型。

### 009. 项目经理在选择领导风格时需要考虑以下因素，除了（）。

**选项：**

- A. 项目经理自身的价值观
- B. 团队成员的价值观
- C. 项目团队的组织结构
- D. 项目规模的大小

**正确答案：D. 项目规模的大小**

**为什么选这个：**

题干问「除了」。影响领导风格的因素是：领导者特征、团队成员特征、组织特征、环境特征。项目规模不在其中。

**为什么不选其他：**

- 不选 A：自身价值观属于领导者特征，要考虑。
- 不选 B：成员价值观属于团队成员特征，要考虑。
- 不选 C：组织结构属于组织特征，要考虑。

**记忆要点：** 选领导风格看人、组织、环境；不看项目规模。

### 010. 大家都愿意跟随某个项目经理做项目，因为他赏罚分明，团队成员按时实现每个里程碑都有奖金，突出贡献者还有额外奖励。这位项目经理主要采用了哪种领导风格？（）

**选项：**

- A. 变革型
- B. 服务型
- C. 交易型
- D. 魅力型

**正确答案：C. 交易型**

**为什么选这个：**

交易型领导用明确的奖惩交换绩效。赏罚分明、里程碑奖金、额外奖励，正是交易型。

**为什么不选其他：**

- 不选 A：变革型靠愿景和鼓舞激发潜能，不是奖金条款。
- 不选 B：服务型是清障赋能，不是发奖金。
- 不选 D：魅力型靠个人魅力追随，题干强调制度性奖惩。

**记忆要点：** 奖惩换结果 → 交易型；愿景鼓舞 → 变革型；清障服务 → 服务型；个人魅力 → 魅力型。

### 011. 你想成为一个魅力型的领导，可以从以下几个方面改进，除了（）。

**选项：**

- A. 保持热情洋溢和精神饱满
- B. 充满自信，加强影响力
- C. 保持热情洋溢，加强说服力
- D. 赋予团队成员更多的自主权

**正确答案：D. 赋予团队成员更多的自主权**

**为什么选这个：**

题干问「除了」。魅力型靠热情、自信、影响力和说服力让人追随。给成员更多自主权是放任型/服务型做法，不是魅力型改进方向。

**为什么不选其他：**

- 不选 A：热情、精力充沛是魅力型典型特征。
- 不选 B：自信、加强影响力是魅力型核心。
- 不选 C：热情、加强说服力也是魅力型常用手段。

**记忆要点：** 魅力型 = 热情 + 自信 + 影响力/说服力；自主权 → 放任/服务。

### 018. 项目经理在选择管理项目的方法时，应该（）。

**选项：**

- A. 采用《指南》中的项目管理过程
- B. 采用适应型生命周期
- C. 考虑项目执行组织的文化
- D. 采用尽量敏捷的方法

**正确答案：C. 考虑项目执行组织的文化**

**为什么选这个：** 选方法要裁剪，须结合组织文化等环境；不能一律套指南或一律敏捷。

**为什么不选其他：**
- 不选 A：不必照搬全部过程。
- 不选 B：适应型不是默认必选。
- 不选 D：不是越敏捷越好。

**出处：** PMBOK 第六版裁剪与事业环境因素/组织文化。

### 021. 由于项目经理已经率先取得了专业资格证，团队成员都以他为榜样，努力学习考取证书。这是哪种权力的体现？（）

### 022. 在敏捷项目中，项目经理需要充分发挥影响力，确保客户频繁参与项目工作，因为（）。

**选项：**

- A. 客户满意度是项目的成功指标
- B. 需要客户及时提出需求和提供反馈
- C. 需要客户及时提供资金和制定决策
- D. 客户决定当前迭代期的成果能否通过验收并进入下个迭代

**正确答案：B. 需要客户及时提出需求和提供反馈**

**为什么选这个：** 敏捷靠客户频繁给需求与反馈。

**为什么不选其他：**
- 不选 A：满意度是结果指标，不是频繁参与的直接原因。
- 不选 C：资金决策偏发起人。
- 不选 D：验收通常是产品负责人，且不决定「能否进入下个迭代」。

**出处：** 敏捷实践——客户协作与频繁反馈。

### 060. 为了担任项目经理这个特定岗位而必须具备的知识、技能和态度以及相应的行为被称为（）。

**选项：**

- A. 能力
- B. 胜任力
- C. 权力
- D. 职责

**正确答案：B. 胜任力**

**为什么选这个：** 胜任力 = 担任特定岗位所需的知识、技能、态度及相应行为。

**为什么不选其他：**
- 不选 A：能力较笼统，不等同于岗位胜任力定义。
- 不选 C：权力是影响他人的能力，不是知识技能态度的集合。
- 不选 D：职责是该做什么，不是具备什么知识技能态度。

**出处：** PMBOK 第六版第 3 章，项目经理胜任力。
