# 001. 项目生命周期

项目正式开始之前，先做前期准备。项目自己的通用生命周期有 4 个阶段：

1. 开始项目
2. 组织与准备
3. 执行项目工作
4. 结束项目

这 4 个是生命周期阶段，不是五大过程组。过程组见 004。

| 阶段 | 这时编制或产生的 | 对应过程组 |
| --- | --- | --- |
| 前期准备（项目还没开始） | 需要评估，然后是商业论证，再是效益管理计划 | 还没有项目 |
| 开始项目 | 项目章程。识别相关方，得到相关方登记册 | 启动 |
| 组织与准备 | 项目管理计划、需求文件。还会排出进度、成本等计划 | 规划 |
| 执行项目工作 | 可交付成果、工作绩效数据 | 执行，同时做监控 |
| 结束项目 | 最终产品、服务或成果，最终报告 | 收尾 |

需要评估、商业论证、效益管理计划是商业文件。发起人负责维护商业论证。项目经理不能直接改这些商业文件，只能提建议。三份文件编在前期准备，整个项目期间还要反复查看。结束时拿结果和商业论证里的目标对比，判断项目是否成功。效益常常在项目结束后才实现，效益管理计划会继续用到项目结束之后。

两个名字不要并成一件事：

- 需要评估（Needs Assessment）在前期准备里做。弄清业务目的、问题和机会，并提出处理建议。它排在商业论证之前，结果可以写进商业论证。
- 需求文件（Requirements Documentation）在组织与准备阶段编制，是规划过程组里「收集需求」的输出。它记录相关方的需要和需求，用来定义范围。

因此，应该在前期准备阶段编制的文件是需要评估。项目章程在开始项目时编制，项目管理计划和需求文件在组织与准备时编制。

# 003. 项目、项目集、项目组合与组织级项目管理

| | 管什么 |
| --- | --- |
| 项目管理 | 用正确的方式完成单个项目，实现项目目标 |
| 项目集管理 | 完成一系列相互关联的项目，获得单独管理得不到的更大收益 |
| 项目组合管理 | 选择最有利于实现战略目标的项目集和项目，也就是做正确的项目 |
| 组织级项目管理 | 为使项目符合组织的战略业务目标，对项目组合、项目集和项目做系统化管理 |

项目之间如果只共享顾主、供应商、技术或资源，没有必须的内在联系，当作项目组合来协调管理，不当作项目集。

# 005. 工作绩效数据、工作绩效信息、工作绩效报告

051. 工作绩效报告的主要用途是便于制定决策和采取行动。

| | 区别 | 主要用途 | 从哪个过程出来 |
| --- | --- | --- | --- |
| 工作绩效数据 | 原始观察和测量，还没分析 | 实时显示项目的进展情况。如实际工时、实际成本、完成百分比、缺陷个数 | 指导与管理项目工作 |

| 工作绩效报告 | 把信息汇编成的口头说明或书面文件 | 便于制定决策、提出问题、采取行动或引起关注。如状态报告、备忘录、仪表盘 | 监控项目工作 |

显示项目工作安排的是项目计划，不是这三者。

项目数据、项目信息、项目报告是上一层说法。工作绩效数据、工作绩效信息、工作绩效报告分别是它们在绩效上的具体形式。

# 009. PMO 的类型

003. 某公司的 PMO 虽然不直接管控项目，但有权要求项目采用公司特定的项目管理方法论。这是什么类型的 PMO？

答案：控制型。《指南》里只有三种 PMO，按对项目的控制程度从低到高排列。没有教练型。

| 类型 | 控制程度 | 特征 |
| --- | --- | --- |
| 支持型 | 低 | 当顾问。提供模板、最佳实践、培训、信息和其他项目的经验教训，相当于项目资源库。无权要求项目服从 |
| 控制型 | 中 | 既提供支持，也要求项目服从。可以要求采用特定的项目管理框架或方法论、使用特定模板、格式和工具，并遵从治理框架。不直接管理项目 |
| 指令型 | 高 | 直接管理和控制项目。项目经理由 PMO 指定，并向 PMO 汇报 |

这道题的关键是两句话：不直接管控项目，排掉指令型；有权要求项目采用特定方法论，不是只提供支持，所以是控制型。


# 010. 组织结构类型

005. 强矩阵组织和弱矩阵组织之间的主要差别在于项目经理对项目资源的控制程度不同。

《指南》表 2-1 按项目经理能拿到多少职权和资源，列了 10 种组织结构。从职能型到项目型，项目经理的职权和资源从几乎没有升到几乎全部。

| 类型 | 项目经理职权 | 资源可用性 | 谁管预算 | 项目经理的角色 | 项目管理行政人员 |

| 有机型或简单型 | 很少或没有 | 很少或没有 | 所有者或运营者 | 兼职，不一定有正式职位 | 很少或没有 |
| 职能型 | 很少或没有 | 很少或没有 | 职能经理 | 兼职，不一定有正式职位 | 兼职 |
| 多部门型 | 很少或没有 | 很少或没有 | 职能经理 | 兼职，不一定有正式职位 | 兼职 |

# 011. 风险、制约因素、事业环境因素

006. 经过实地考察，项目经理发现当地关税非常高，原材料进口很不划算。这个发现应该被记录为事业环境因素。

关税已经查明，是组织外部、项目团队控制不了的条件，属于事业环境因素里的财务考虑因素。

| | 是什么 | 怎么认 |
| --- | --- | --- |
| 事业环境因素 | 项目团队不能控制的条件，会影响、限制或指令项目 | 已经存在的环境。关税、汇率、利率、法律、市场、行业标准、组织文化、人员能力都算 |
| 制约因素 | 已经落到这个项目上的具体限制 | 比事业环境因素更具体。事先定下的预算、强制性里程碑、合同条款，以及范围、进度、成本、质量、资源、风险这些相互竞争的限制 |
| 风险 | 不确定的事件或条件，一旦发生，会对项目目标产生正面或负面影响 | 必须还没有定。已经查明的高关税没有不确定性，不记成风险 |

事业环境因素分组织内部和组织外部。

- 组织内部：组织文化、结构和治理，设施和资源的地理分布，基础设施，信息技术软件，资源可用性，员工能力。
- 组织外部：市场条件，社会和文化影响，法律限制，商业数据库，学术研究，政府或行业标准，财务考虑因素（汇率、利率、通货膨胀、关税、地理位置），物理环境。

组织过程资产是组织自己积累的流程、政策、程序和知识库。关税不是组织内部的资产。

# 013. 项目经理要关注的三个整合层面

017. 整合是项目经理的一项关键技能，需要关注的整合层面不包括哪一个？

| 层面 | 整合什么 |

客户层面不是《指南》里的整合层面。

| **A. 领导（正确答案）** | 是。建立 + 传达 + 带领一群人去实现 | 项目经理先画出「半年后用户打开 App 就能远程看家」的图景，在启动会和每周例会上反复讲给开发、测试、产品听，并带着这群人一起把远程查看做成。 |
| **B. 管理** | 不是。管理是要求一群人按既定规则行动，不负责画愿景 | 同样这群人，项目经理按进度计划、缺陷阈值、每日站会规则，要求大家按规定流程推进远程查看功能。没有建立愿景，也没有传达愿景。 |
| **C. 命令** | 不是。题干里没有「命令」的成分 | 项目经理对这群人说：「周五前必须交远程查看接口，不交按绩效扣。」只有指令和服从，没有建立愿景，也没有传达愿景。 |
| **D. 激励** | 只覆盖一部分。激励是领导里的一项，不如 A 全面 | 项目经理给这群人发上线奖金、公开表扬先做出演示的人，让大家更有干劲。有「带动一群人」的味道，但没有建立愿景，也没有传达愿景。 |

## 怎么记

- 领导：愿景从哪来、怎么讲出去、怎么带着人走过去。
- 管理：规则、计划、控制，让人把既定的事做对。
- 命令：只下指令。
- 激励：鼓劲、奖赏，是领导的一块，不是领导的全部。

题干三件事都点到了，选最全的 **领导**。


# 019. 人才三角：战略和商业管理技能

057. PMI 发布的人才三角中提出了战略和商业管理技能，这项技能要求项目经理（）。

答案：C. 了解和应用项目所在行业的相关知识

## 题干

问的是人才三角里「战略和商业管理」这一角，项目经理要会什么。

## 《指南》原文在说什么（第 6 版约第 56–58 页，3.4 人才三角）

PMI 人才三角是三块技能，项目经理都要具备：

| 一角 | 《指南》怎么界定 | 对应本题干扰项 |
| --- | --- | --- |
| 项目管理专业技能（Technical Project Management） | 与项目、项目集、项目组合管理特定领域有关的知识、技能和行为，也就是把本职的技术工作做出来 | A 进度计划编制和执行；D 针对每个项目裁剪传统和敏捷工具 |
| 领导力（Leadership） | 引导、激励和带领团队，帮助组织实现业务目标所需的知识、技能和行为 | B 能够激励和带领团队实现目标 |
| **战略和商业管理（Strategic and Business Management）** | **对行业和组织的了解与专长**，用来提升绩效、更好地交付业务成果 | **C 了解和应用项目所在行业的相关知识** |

战略和商业管理这一角，《指南》还写了这些要点（意译，不逐字照抄全书）：

1. 能从组织高层看全局，谈成并落实那些**对齐战略、支持创新**的决策和行动。
2. 这类本领也叫**领域知识 / 行业知识（domain knowledge）**：了解并应用项目所在行业、组织的相关知识。
3. 项目经理要有足够的业务知识，才能向别人讲清项目在业务上为什么必要，包括向**高管、运营经理、职能经理**说明项目的必要性。
4. 要能和发起人、团队、主题专家一起制定合适的交付策略，并按能**最大化项目商业价值**的方式去落实。
5. 决策时要听取运营经理的意见：他们懂组织日常工作，也懂项目计划会怎样影响这些工作。
6. 项目经理应能向他人解释：战略、使命、目标、产品与服务、运营方式、市场与竞争等。
7. 还可能包括财务、营销、运营等其他职能的工作知识。

本题解析里那句，就是上面第 2、3 点的压缩：战略和商业管理技能包括**了解和应用相关行业知识**，以便向高管、运营经理和职能经理**宣传项目的必要性**。

## 为什么是 C，不是 A/B/D

- **C：** 原文把「行业 / 组织 / 领域知识」写在战略和商业管理这一角。
- **A、D：** 进度计划、裁剪传统/敏捷工具，属于项目管理专业技能。
- **B：** 激励、带领团队，属于领导力。

## 怎么记

- 专业技能：把项目做对（计划、工具、裁剪）。
- 领导力：把人带起来。
- 战略和商业：懂行业、懂公司为什么做这个项目，能跟高管把必要性讲清楚。


# 020. 实施整体变更控制的工具与技术

009. 以下哪项不是实施整体变更控制过程的工具与技术？

答案：D. 项目变更评审

## 题干

问的是《指南》**列出的工具与技术名单**里有没有这一项，不是问「做变更时会不会评审」。评审会开会做，但书上没把「项目变更评审」写成独立工具。

## 本过程《指南》列出的工具与技术（约第 118–120 页）

| 工具与技术 | 干什么 |
| --- | --- |
| **专家判断** | 审批变更请求时，请懂范围、进度、成本、风险、合同的人判断该不该批 |
| **变更控制工具** | 手工或自动，用来管变更请求和审批结果；常和配置管理一起用 |
| **数据分析** | 备选方案分析、成本效益分析，比较批或不批、选哪条改法 |
| **决策** | 投票、独裁型决策、多标准决策分析 |
| **会议** | 变更控制委员会（CCB）开会讨论、批准、推迟或否决变更 |

所以 A 专家判断、B 会议、C 变更控制工具都在名单里。D 项目变更评审是会里发生的事，**不是单独列出的工具**。

## 变更控制工具具体有哪些

《指南》不报软件名，只说按相关方需要、组织环境来选。工具要能管住**变更请求**和**做出的决定**。通常覆盖两条线：

**配置控制（管「基线上的东西是什么」）**

1. 识别并记录配置项
2. 记录并报告配置项状态
3. 对配置项做核实和审计

**变更控制（管「谁要改、批没批」）**

1. 识别并记录变更请求
2. 审查并批准或否决变更请求
3. 跟踪并沟通变更请求状态
4. 记录并报告变更请求的处理结果

实务里常见形态：变更申请单 / 变更日志、配置项清单与版本库、电子审批流、和 CCB 会议纪要连在一起的系统。这些都算变更控制工具，不是「项目变更评审」这个选项名。

## 怎么记

- 名单里有：专家判断、会议、变更控制工具（以及数据分析、决策）。
- 名单里没有：项目变更评审。评审靠会议 + 工具完成，本身不是工具名。


# 020. 实施整体变更控制的工具与技术

009. 以下哪项不是实施整体变更控制过程的工具与技术？

答案：D. 项目变更评审

## 题干

问的是《指南》**实施整体变更控制**（4.6）这一过程，书上点名列出的工具与技术有哪些。变更评审这件事过程里会做，但**没有单独列成一项工具**，所以选 D。

这个过程属于**监控过程组**、整合知识领域，从项目开始到结束都可以做：谁提出变更，都先变成变更请求，再在这里审批，批准后才改基准。

## 《指南》列出的工具（约第 118–120 页）

| 工具 | 书上有没有 | 本题 |
| --- | --- | --- |
| 专家判断 | 有 | A |
| 会议 | 有 | B |
| 变更控制工具 | 有 | C |
| 数据分析（备选方案分析、成本效益分析） | 有 | 本题没出现 |
| 决策（投票、独裁型决策、多标准决策分析） | 有 | 本题没出现 |
| **项目变更评审** | **未单列** | **D，应排除** |

## A. 专家判断：哪个阶段用、干什么

**过程：** 实施整体变更控制（监控过程组）。审批每一条变更请求时都可能用。

**谁当专家：** 变更控制委员会（CCB）、主题专家、顾问、有行业经验的人、发起人指定的人。

**例子：** 产品要加「远程看家」，要改范围基准。项目经理拉硬件专家、法务、安全顾问判断：能不能做、对进度成本风险有什么影响、该批还是驳。专家不代替 CCB 拍板，是给审批提供依据。

专家判断在识别风险、估算、规划等很多过程也会用；本题只问它是不是**本过程**的工具——是。

## B. 会议：哪个阶段用、干什么

**过程：** 同样是实施整体变更控制。最典型的是 **CCB 会议**：讨论变更请求，做出批准、否决、推迟或补充信息后再议。

**例子：** 每周五 CCB 开会，逐条看变更请求清单。远程看家这条：开发说要加 3 周，发起人说战略需要，CCB 批准并指定更新范围基准和进度基准。会后会议纪要写进变更日志。

日常站会、开工会不是这个过程的「会议」工具；这里专指**为变更评审和决策开的会**。

## C. 变更控制工具：哪个阶段用、具体是什么

**过程：** 实施整体变更控制全程用，用来管变更请求和配置项，通常是人工台账或软件系统。

《指南》把这类工具看成两块能力叠在一起：

| 一块 | 管什么 | 典型功能 |
| --- | --- | --- |
| **变更管理** | 变更请求本身 | 识别、记录、批准或否决对文件、可交付成果、基准的修改；跟踪并沟通每条请求的状态 |
| **配置管理** | 产品/文件的版本和组成 | 识别配置项；记录并报告配置项状态；核实配置项是否齐全、是否与需求一致 |

**具体例子（工具长什么样）：**

1. **Jira / Azure DevOps 的变更单工作流：** 提交 → 评估 → CCB 审批 → 实施 → 关闭。状态谁都能看见。
2. **Git + 配置管理：** 需求规格、图纸、源代码当配置项，改基准必须走分支/标签，批完才能合进发布基线。
3. **Excel 变更日志：** 小项目没有系统时，用表格记录编号、提出人、影响、CCB 结论、实施日期。这也是变更控制工具。
4. **文档库版本（SharePoint / Confluence）：** 项目管理计划、WBS 只认「已批准版本」，防止有人私下改基准。

**和会议、专家判断怎么串起来：**

提出变更 → 写入变更控制工具（建单、编号）→ 专家判断（评估影响）→ 会议 / CCB（批准或否决）→ 结论回写到变更控制工具（状态、版本、基准）

## 一条变更走一遍（远程看家）

1. 产品经理在 Jira 提变更请求（**变更控制工具**）。
2. 架构师、安全顾问评估工期和合规（**专家判断**）。
3. CCB 开会决定批准，要求同步改范围和进度基准（**会议**）。
4. 项目经理在系统里把状态改成「已批准」，升一版范围基准（**变更控制工具**）。

「项目变更评审」是第 3 步在干的事，但书上不把它单独列成工具，评审靠的是会议 + 专家判断 + 决策。

## 怎么记

- 问「是不是工具」：只认书上点名的。
- 专家判断 = 谁来评估该不该批。
- 会议 = CCB 在哪拍板。
- 变更控制工具 = 用什么系统/台账把请求、状态、配置项版本管住。



# 021. 管理项目知识：信息管理与知识管理

020. 以下哪个工具技术可以在人和信息之间建立联系，有利于分享显性知识？

答案：A. 信息管理

## 题干

《指南》**管理项目知识**（4.4，约第 103 页）里，哪种工具是「把人和信息连起来」，用来分享**已经写下来的知识**。

先分清两种知识：

| | 是什么 | 怎么传 | 本题对应 |
| --- | --- | --- | --- |
| **显性知识** | 能写进文件、数据库、图纸、代码、检查单的 | 人和**信息**之间建立联系 | **信息管理** |
| **隐性知识** | 在人脑子里：手感、经验、判断、谁熟谁 | 人和**人**之间建立联系 | **知识管理** |

## 适用于显性分享：信息管理

《指南》的说法：信息管理是促进**显性知识**分享的各种具体方法的统称。目标是让人能找到、用上已经编码的信息。

常见做法：

1. **经验教训登记册**（写下来、可检索）
2. **文档库 / 图书馆服务**（章程、计划、规格、纪要）
3. **项目管理信息系统（PMIS）**：进度软件、配置库、门户。它是信息管理的一种具体方法，不如「信息管理」这个统称全面，所以本题不选 C。
4. 信息收集，把分散材料归到可搜索的地方

**例子：** 远程看家接口的时序图画进 Confluence，新人打开页面就能做，不必找原作者坐旁边讲。人和文档连上了，这是显性分享。

## 适用于隐性分享：知识管理

《指南》的说法：知识管理有利于分享**隐性知识**，也是把已有知识凑在一起、并创造出新知识的各种方法的统称。目标是让人跟人互动，把说不清的经验带出来。

常见做法：

1. 人际网络、实践社区
2. 工作会议、研讨会、知识集市
3. 工作见习、影子学习（跟着熟手看一天）
4. 辅导 / 师徒
5. 讲故事（把踩坑经过讲给别人听）
6. 讨论区、论坛（有互动，不只是贴文档）

**例子：** 蓝牙配网只有老员工「听声音就知道哪步失败了」，说不全。新人跟他联调半天，或开复盘会让他讲故事。人和人连上了，这是隐性分享。

## 本题四个选项怎么摆

| 选项 | 偏向 | 为什么本题选或不选 |
| --- | --- | --- |
| **A. 信息管理** | 显性 | 题干原话：在人和信息之间建立联系，利于分享显性知识 |
| **B. 知识管理** | 隐性 | 对，但是人和人，不是人和信息 |
| **C. PMIS** | 显性里的一种工具 | 覆盖面小于 A |
| **D. 数据收集** | 收集数据的统称（头脑风暴、核对单等） | 不是「人—信息」这条分享通道，也不如 A 对准题干 |

## 怎么记

- 能写进 Wiki / 登记册 / PMIS → 显性 → **信息管理**。
- 必须跟着人学、开会聊、带徒弟 → 隐性 → **知识管理**。
- 题干出现「人和信息」「显性」→ 选信息管理。
- 题干出现「人和人」「隐性」「创造新知识」→ 选知识管理。



# 022. 项目管理计划由哪些部分组成

021. 以下哪项不是项目管理计划的组成部分？

答案：D. 项目进度计划

## 题干

问哪一项**不是**项目管理计划（PMP）里的东西。坑在「进度管理计划 / 进度基准」在计划里，**详细的项目进度计划**是项目文件，不在计划里。

出处：《指南》第 6 版约第 86–88 页，制定项目管理计划的输出。

## 项目管理计划三块

### 1. 分项管理计划（怎么管）

| 分项计划 | 管什么 |
| --- | --- | --- |
| 范围管理计划 | 怎么定义、制定、监督、控制和确认范围 |
| 需求管理计划 | 怎么分析、记录、管理需求 |
| **进度管理计划** | 怎么编制、监督、控制**进度**（注意：不是进度表本身） |
| 分项计划 | 管什么 |
| --- | --- |
| 资源管理计划 | 怎么估算、获取、管理和使用团队及实物资源 |
| 沟通管理计划 | 怎么规划、执行、监督沟通 |
| 风险管理计划 | 怎么安排风险管理活动 |
| 采购管理计划 | 怎么从组织外获取货物和服务 |
| 相关方参与计划 | 怎么促进相关方有效参与 |
| 变更管理计划 | 怎么管理变更控制过程 |
| 配置管理计划 | 怎么识别、记录、控制产品/文件的配置项 |

本题 **A 范围管理计划** 就属于这一块。

### 2. 基准（用来对照绩效的尺子）

| 基准 | 里面有什么 |
| --- | --- |
| **范围基准** | 项目范围说明书、WBS、WBS 词典 |
| **进度基准** | 经批准的进度模型（作为比较基准的那一版） |
| **成本基准** | 经批准的、按时间分段的预算 |

三个基准合在一起常叫**绩效测量基准**。本题 **B 范围基准** 属于这一块。

### 3. 其他组成部分

| 内容 | 作用 |
| --- | --- |
| **项目生命周期** | 项目分哪几个阶段、各阶段做什么 |
| 开发方法 | 预测型、适应型、混合型等 |
| 管理审查 | 谁在什么节点审查项目 |

本题 **C 项目生命周期** 属于这一块。

## 容易混：计划里的 vs 项目文件

| 在项目管理计划里 | 是项目文件，不在计划「组成部分」里 |
| --- | --- |
| 进度**管理**计划 | **项目进度计划**（活动清单、甘特图、网络图等详细进度） |
| 进度**基准** | 项目日历、进度数据 |
| 成本管理计划、成本基准 | 成本估算、项目资金需求 |
| 范围管理计划、范围基准 | 需求文件、需求跟踪矩阵 |
| 风险管理计划 | 风险登记册、风险报告 |
| 变更管理计划 | 变更日志 |

所以 **D 项目进度计划** 不是项目管理计划的组成部分。书上把它归为项目文件：执行时用来排活，变了可以按变更流程更新；计划里只规定「怎么做进度、拿哪一版当基准」。

## 怎么记

- 计划 = **怎么管**（分项计划）+ **拿什么当尺子**（基准）+ **生命周期/开发方法**。
- 带「管理计划」「基准」「生命周期」→ 多半在 PMP 里。
- 带详细表、登记册、估算、甘特图 → 多半是项目文件。
- 进度管理计划 ≠ 项目进度计划。



# 023. 项目管理计划组成（精简）

对，**D 的解释成立**：项目进度计划是**项目文件**，不是项目管理计划的组成部分。

对照：

- **进度管理计划**：在项目管理计划里（规定怎么编进度、怎么控制）。
- **进度基准**：在项目管理计划里（经批准、用来比绩效的那一版）。
- **项目进度计划**：项目文件（甘特图、网络图、活动清单等干活用的详细表）。

## 项目管理计划就三块

**1. 分项管理计划**  
范围、需求、进度、成本、质量、资源、沟通、风险、采购、相关方、变更、配置。

**2. 基准**  
范围基准、进度基准、成本基准（合称绩效测量基准）。

**3. 其他**  
项目生命周期、开发方法、管理审查。

带「管理计划 / 基准 / 生命周期」→ 在计划里。带详细进度表、登记册、估算 → 项目文件。



# 024. 知识管理的两种关键活动

036. 知识管理的两种关键活动是（）。

答案：D. 知识分享和知识集成

## 题干

问知识管理（不是信息管理）的**两种关键活动**是什么。《指南》约第 100 页。

## 答案

**分享**：让已有知识被别人用上（人传人、也能配合文件）。  
**集成**：把分散的知识凑在一起，并形成新知识。

| 选项 | 为什么不选 |
| --- | --- |
| A 显性知识和隐性知识 | 这是管理**对象**，不是活动 |
| B 工作跟随和跟随指导 | 只是隐性分享的两种具体方法 |
| C 文献搜索和图书馆服务 | 只是显性分享（信息管理）的方法 |
| D 知识分享和知识集成 | 书上说的两种关键活动 |

和 021 题一起记：显性靠信息管理，隐性靠知识管理；知识管理这件事本身 = 分享 + 集成。



# 025. 配置控制关注技术规范

043. 下列关于配置控制的说法正确的是什么？

答案：配置控制重点关注项目的技术规范。

配置控制看的是产品和过程的技术参数，例如规格、图纸、技术标准。普通的变更控制看的是绩效测量基准，也就是范围、进度和成本基准。


# 026. 外部项目既要合同也要项目章程

048. 项目团队为外部客户开展项目，合同和项目章程是否都要准备？

答案：合同和项目章程都需要准备。签订合同确保在法律上合法。制定项目章程则有利于建立组织内部的合作关系，以确保正确交付合同内容。两者缺一不可。


# 027. 项目整合管理没有「规划整合管理」

059. 项目整合管理的过程包括以下所有，除了哪一个？

答案：除了规划整合管理。没有这个过程。整合管理在规划阶段的过程叫制订项目管理计划，不叫规划整合管理。其他知识领域常有「规划某管理」，整合管理没有这个名称。

项目整合管理的七个过程是：制定项目章程，制订项目管理计划，指导与管理项目工作，管理项目知识，监控项目工作，实施整体变更控制，结束项目或阶段。


# 028. 协议是制定项目章程的输入

051. 协议是以下哪个项目管理过程的输入？

答案：制定项目章程。

有两种主要情况：一是两个或多个发起组织之间签署合作协议，作为制定项目章程的依据；二是买卖方的协议（合同），作为卖方制定项目章程的依据。


# 029. 整合管理各过程都用专家判断

060. 在项目整合管理中要使用哪个工具与技术来实施各过程？

答案：专家判断。它是项目整合管理七个过程共同采用的工具与技术。

其余工具只出现在部分过程里：引导技术只用于制定项目章程和制订项目管理计划；分析技术只用于监控项目工作和结束项目或阶段；项目管理信息系统只用于指导与管理项目工作和监控项目工作。


# 030. 产品测量指标属于需求管理计划

036. 下列哪项不属于范围管理计划的内容？

答案：产品测量指标及使用这些指标的理由。

它属于需求管理计划。范围管理计划写的是怎么制定范围说明书、怎么创建工作分解结构、怎么审批和维护范围基准、怎么正式验收已完成的可交付成果。


# 031. 确认范围的工作绩效信息

038. 确认范围过程会得到工作绩效信息，其中应该记录什么？

答案：哪些可交付成果未通过验收及原因。也会写明哪些已经被验收。

确认范围是对可交付成果的正式验收。控制范围也会输出工作绩效信息，记的是对照范围基准的实施情况、识别的范围偏差和原因、对将来范围绩效的预测。
