# 挣值指标：成本偏差

题目：项目工作绩效报告中 PV=110，EV=100，AC=120，总预算为 150。成本偏差是多少？

**成本偏差 CV = EV − AC = 100 − 120 = −20。** 负数表示超支：完成的工作只值 100，实际却花了 120。

## 缩写

| 缩写 | 英文 | 这道题里的含义 |
| --- | --- | --- |
| PV | Planned Value，计划价值 | 按计划到今天应该完成的工作值 110 |
| EV | Earned Value，挣值 | 实际已经完成的工作，按预算算值 100 |
| AC | Actual Cost，实际成本 | 到今天已经花掉 120 |
| CV | Cost Variance，成本偏差 | EV − AC，衡量省钱还是超支 |
| SV | Schedule Variance，进度偏差 | EV − PV，衡量进度超前还是落后 |
| BAC | Budget at Completion，完工预算 | 题目里的总预算 150，计算成本偏差用不到 |

进度偏差是另一列：SV = 100 − 110 = −10，表示进度落后。总预算 150 是整个项目的完工预算，不参与 CV 的计算。

```python
PV = 110
EV = 100
AC = 120
BAC = 150

CV = EV - AC
SV = EV - PV

print(f"CV = EV - AC = {EV} - {AC} = {CV}")
print(f"SV = EV - PV = {EV} - {PV} = {SV}")
print(f"BAC = {BAC}，不参与 CV 的计算")
```
