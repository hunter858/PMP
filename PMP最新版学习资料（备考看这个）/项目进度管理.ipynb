# 第六章：项目进度管理

### 001. 项目进度管理的6个过程，正确的先后顺序是（）。

**选项：**

- A. 规划进度管理—定义活动—估算活动持续时间—排列活动顺序—制订进度计划—控制进度
- B. 规划进度管理—定义活动—排列活动顺序—估算活动持续时间—制订进度计划—控制进度
- C. 规划进度管理—定义活动—估算活动持续时间—制订进度计划—排列活动顺序。控制进度
- D. 规划进度管理—定义活动—估算活动资源—估算活动持续时间—制订进度计划—控制进度

**正确答案：B. 规划进度管理—定义活动—排列活动顺序—估算活动持续时间—制订进度计划—控制进度**

**为什么选这个：** PMBOK 第六版进度管理六过程顺序为：规划进度管理 → 定义活动 → 排列活动顺序 → 估算活动持续时间 → 制订进度计划 → 控制进度。

**为什么不选其他：**
- 不选 A：把估算持续时间放在排列顺序之前，顺序错了。
- 不选 C：顺序混乱，且把排列活动顺序放到制订进度计划之后。
- 不选 D：估算活动资源在第6版属资源管理，不是进度管理这六个过程之一。

**出处：** PMBOK 第六版第 6 章—项目进度管理过程。

### 004. 某活动的工期最乐观估算6天，最悲观估算18天，最可能估算为9天，假设活动工期呈三角分布。则活动的期望持续时间为多少天？在9至13天内完成该活动的概率是多少？（）

**选项：**

- A. 10天, 68.26%
- B. 11天, 68.26%
- C. 10天, 99.73%
- D. 11天, 97.73%

**正确答案：B. 11天, 68.26%**

**为什么选这个：** 三角分布期望 te=(O+M+P)/3=(6+9+18)/3=11；σ=(P-O)/6=2，9～13天为 11±1σ，概率约 68.26%。

**为什么不选其他：**
- 不选 A/C：10 天是贝塔分布 (O+4M+P)/6，题干是三角分布。
- 不选 D：±1σ 对应 68.26%，不是 97.73%。

**出处：** PMBOK 第六版第 6 章，6.4—三点估算（三角分布）。

### 009. 向某个已经延误的活动增加新的资源，将导致的结果是（）。

**选项：**

- A. 活动可能延误更长时间
- B. 延误的时间被赶回
- C. 活动效率提高
- D. 活动生产率提高

**正确答案：A. 活动可能延误更长时间**

- 不选 B：不一定能赶回延误，往往适得其反。
- 不选 C/D：加人初期效率/生产率通常下降，而非提高。

**出处：** 进度压缩/资源相关—布鲁克斯定律。

### 011. 对于一个低风险，而且活动之间主要是软逻辑关系的项目，最好采用什么方法压缩进度？（）

**选项：**

- A. 快速跟进
- B. 赶工
- C. 资源平衡
- D. 资源优化技术

**正确答案：A. 快速跟进**

**为什么选这个：** 软逻辑可改为并行，低风险项目更能承受快速跟进带来的返工风险。

**出处：** PMBOK 第六版第 6 章，6.5.2.6—进度压缩（快速跟进）。

### 015. 以下关于关键路径的说法哪项是错误的？（）

**选项：**

- A. 关键路径是项目路径中最长的那条路径
- B. 关键路径决定了项目的最短工期
- C. 关键路径可能没有，可能有一条，也可能有多条

**正确答案：C. 关键路径可能没有，可能有一条，也可能有多条**

- 不选 D：正确——关键路径总浮动一般为 0，延误会推迟完工。

### 020. 在以下哪种情况下，关键路径上的活动可能出现负的浮动时间？（）

## 032. 参数估算的准确度取决于（）。

**选项**
- A. 活动的相似度和估算人员的专业程度及经验
- B. 参数模型的成熟度和基础数据的可靠性
- C. 单个活动或工作包的规模和复杂程度
- D. 实际分布和假定分布的契合度

**正确答案：B**

**为什么选B**
参数估算用历史数据与统计关系（参数模型）推算工期/成本。PMBOK指出：准确度取决于所用模型的成熟度，以及模型中基础数据的可靠性。模型越成熟、数据越可靠，估算越准。

**为什么不选其他**
- A：活动相似度 + 估算人员经验，主要影响**类比估算**的准确度，不是参数估算的核心条件。
- C：活动/工作包规模与复杂程度会影响估算难度，但不是参数估算准确度的标准表述。
- D：实际分布与假定分布的契合度，更贴近**三点估算/PERT**对概率分布假设的讨论，不是参数估算准确度的依据。

**出处**
PMBOK® Guide 第6版 6.4.2.2 / 6.5.2.3 参数估算：准确度取决于参数模型的成熟度以及基础数据的可靠性。


## 033. 机器组装和测试分别由团队中的两个小组负责。只有机器组装完毕，才能对其进行测试，这是采用的以下哪一种关系？（）

**选项**
- A. 外部强制性依赖
- B. 外部选择性依赖
- C. 内部强制性依赖
- D. 内部选择性依赖

**正确答案：C**

**为什么选C**
依赖可从两个维度判断：
1. **内部/外部**：双方都是项目团队内的小组 → **内部依赖**。
2. **强制/选择**：物理或合同上必须满足——组装未完成就无法测试 → **强制性依赖（硬逻辑）**。
合起来为**内部强制性依赖**。

**为什么不选其他**
- A：外部强制性——外部指项目团队之外（如供应商、法规、政府审批），本题双方都在团队内。
- B：外部选择性——既非外部，也非“优选/软逻辑”。
- D：内部选择性——选择性依赖是团队基于经验或最佳实践“希望”如此安排，可调整；本题是“只能如此”，不是偏好。

**出处**
PMBOK® Guide 第6版 6.3.2.2 确定依赖关系：强制性/选择性、内部/外部依赖。

## 035. 项目进度计划中所安排的应急储备，通常应该随着项目的进展而逐渐（）。

**选项**
- A. 增加
- B. 减少
- C. 保持不变
- D. 被隐藏

**正确答案：B**

**为什么选B**
应急储备用于应对已知—未知风险（已识别风险的应急）。随着项目推进，相关活动完成、风险发生并消耗储备，或风险未发生而释放储备，剩余应急储备通常**逐渐减少**。

**为什么不选其他**
- A：一般不会随进展“自然增加”；除非重新评估后增加储备（变更/重估），不是“通常”趋势。
- C：保持不变不符合储备随风险状态变化而消耗或释放的管理逻辑。
- D：应急储备应透明管理，不应“被隐藏”（隐藏的是垫支/管理储备相关错误做法，不是本题正确做法）。

**出处**

- B. 可以开展进度活动的时间段
- C. 固定的项目总工期
- D. 比较短而且固定长度的时间段

## 037. 不遵守下列哪种逻辑关系，会给项目带来最大的不利影响？（）

**选项**
- A. 外部依赖关系
- B. 选择性依赖关系

## 038. 关键路径法的假设前提是（）。

**选项**
- A. 项目必须在某个强制日前完工
- B. 项目必须在既定的成本目标下完工
- C. 项目的资源是充分的
- D. 项目相关方都是支持项目的

**正确答案：C**

**为什么选C**
关键路径法（CPM）在进度网络上按活动历时与逻辑关系做正推/逆推，找出最长路径。其基本假设是**资源充足（不受资源约束）**；若资源不足，需再做资源平衡/资源受限进度安排，关键路径可能改变。

**为什么不选其他**
- A：强制完工日期是约束条件，可能导致负浮动，但不是 CPM 的假设前提。
- B：成本目标属于成本管理，不是关键路径法的假设。
- D：相关方支持属于相关方/沟通管理范畴，与 CPM 计算假设无关。

**出处**
PMBOK® Guide 第6版 6.5.2.2 关键路径法；常见考点：CPM 假设资源无限/充分。


## 039. 在使用节点法编制网络图时，很少用到的一种逻辑关系是（）。

## 041. 对于电缆铺设项目，将电缆的长度乘以铺设每米电缆所需的工时来估算整个项目的工期。这是在使用以下哪种工具进行估算？（）

PMBOK® Guide 第6版 6.4.2.2 / 6.5.2.3 参数估算。

## 044. 在项目进度计划中，你可以发现多种时差。在某特定活动中被使用且不影响任一紧后活动的时差被称为（）。

**选项**

- B. 自由浮动时间
- C. 总浮动时间
- D. 项目浮动时间

**正确答案：B**

**为什么选B**

**出处**

## 045. 一个资源在同一时段内被分配给两个活动，你应该采取什么措施？（）

**选项**
- A. 资源优化
- B. 资源平衡
- C. 资源平滑
- D. 进度压缩

**正确答案：B**

**为什么选B**
同一资源同时分给两个活动，属于**资源过度分配/资源冲突**。应采用**资源平衡（Resource Leveling）**：按资源约束调整活动开始/完成日期，使需求与供给匹配；可能延长工期、改变关键路径。

**为什么不选其他**
- A：资源优化是总称，包含资源平衡与资源平滑；本题要解决的是冲突，更具体的措施是资源平衡。
- C：资源平滑只在浮动时间内微调，使资源使用更均匀，且不改变关键路径；不能用来解决超出可用量的冲突。
- D：进度压缩（赶工、快速跟进）是为缩短工期，不是解决资源冲突。

**出处**
PMBOK® Guide 第6版 6.5.2.3 资源优化：资源平衡与资源平滑。

## 047. 在敏捷项目中，可以使用拳五（从拳头到五指）技术进行决策。哪种情况表示所有团队成员对估算的活动时间达成了共识？（）

**选项**
- A. 所有人都举拳头
- B. 所有人都伸出三根以上的手指
- C. 大多数人伸出三根以上的手指
- D. 所有人伸出的手指数都不低于两根

**正确答案：B**

**为什么选B**
拳五（Fist of Five）：成员举 0～5 根手指表达支持程度。拳头＝强烈反对；五指＝全力支持。通常约定：**所有人都伸出三根及以上手指**才算达成共识；若有人少于三根，需讨论顾虑后再投票。

**为什么不选其他**
- A：全员举拳头表示都不支持，是反对而非共识。
- C：“大多数”是多数决，不是全员共识。
- D：标准阈值一般是三根及以上，不是两根。

**出处**
《敏捷实践指南》拳五投票 / 共识决策相关说明。

## 049. A活动的紧后活动是B活动。B活动最早开始日期是第4天早上上班，最晚开始时间是第5天早上上班。A活动的最早完成时间是第1天下班。请问A活动的自由浮动时间是（）。

**选项**
- A. 1天
- B. 2天
- C. 3天
- D. 无法计算

**正确答案：B**

**为什么选B**
自由浮动：活动可推迟且不影响紧后活动**最早开始**的时间量。
- A 最早完成：第1天下班
- B 最早开始：第4天早上
中间空出第2、第3整天，故自由浮动 = **2天**。
（按日历日公式：FF = ES_B − EF_A − 1 = 4 − 1 − 1 = 2）

**为什么不选其他**
- A/C：天数算错；注意从“第1天下班”到“第4天早上”之间是2个完整工作日。
- D：已知 EF_A 与 ES_B 即可算自由浮动；B 的最晚开始（LS）用于总浮动，本题不算自由浮动所需。

**出处**
PMBOK® Guide 第6版 6.5.2.2：自由浮动 = 不影响紧后最早开始的可推迟时间。


## 050. 关于项目进度网络图，以下哪个说法是错误的？（）

**选项**
- A. 带有多个紧前活动的活动代表路径汇聚
- B. 带有多个紧后活动的活动代表路径分支
- C. 位于路径汇聚和分支处的活动存在较小风险
- D. 项目进度网络图是表示项目进度活动之间的逻辑关系的图形

**正确答案：C**

**为什么选C**

**为什么不选其他**
- A：正确——多个紧前 → 路径汇聚（path convergence）。
- B：正确——多个紧后 → 路径分支（path divergence）。
- D：正确——进度网络图展示活动间逻辑关系。

**出处**
PMBOK® Guide 第6版 6.3.3.1 项目进度网络图：路径汇聚与路径分支；汇聚/分支点需特别关注风险。


## 051. 用快速跟进的方法来压缩进度，是指（）。

**选项**
- A. 通过加班工作，使项目提前完工
- B. 以强有力的奖励手段，使项目提前完工
- C. 把工作并行实施，使项目提前完工
- D. 加快关键路径上的活动的进度，使项目提前完工

**正确答案：C**

**为什么选C**
快速跟进（Fast Tracking）：把通常按顺序进行的活动改为**并行或部分重叠**开展，从而压缩工期；常通过调整软逻辑实现，但会增加返工与风险。

**为什么不选其他**
- A：加班属于赶工时可能采取的加资源/加时间投入手段，不是快速跟进的定义。
- B：奖励手段不属于进度压缩的标准技术。
- D：加快关键路径活动进度更接近**赶工（Crashing）**——增加资源以缩短历时。

**出处**
PMBOK® Guide 第6版 6.5.2.6 进度压缩：赶工与快速跟进。
