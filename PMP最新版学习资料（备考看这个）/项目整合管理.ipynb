# 第四章：项目整合管理

每道题包含：题目、选项、正确答案、为什么选这个、为什么不选其他。

### 001. 项目整合管理是指（）。

**选项：**

- A. 识别和选择项目的资源配置方案
- B. 平衡相互竞争的项目目标，如范围、时间、成本和质量
- C. 协调项目管理各知识领域之间的相互影响
- D. 以上都是

**正确答案：D. 以上都是**

**为什么选这个：** 整合管理包括资源配置抉择、平衡竞争目标、协调各知识领域相互影响，A/B/C 都是。

**为什么不选其他：**
- 不选 A：只是整合的一部分，不完整。
- 不选 B：只是整合的一部分，不完整。
- 不选 C：只是整合的一部分，不完整。

**出处：** PMBOK 第六版第 4 章，项目整合管理概述。

### 002. 总结项目上的全部采购工作的经验教训，并把经验教训写入最终的经验教训登记册。这是哪个管理过程的活动？（）

**选项：**

- A. 控制采购
- B. 结束项目或阶段
- C. 结束采购
- D. 监控项目工作

**正确答案：B. 结束项目或阶段**

**为什么选这个：** 「全部」采购经验写入「最终」经验教训登记册，属结束项目或阶段。

**为什么不选其他：**
- 不选 A：控制采购可记录采购经验，但不形成项目最终经验教训登记册。
- 不选 C：第六版已无「结束采购」独立过程。
- 不选 D：监控项目工作是跟踪报告偏差，非整项目收尾汇总。

**出处：** PMBOK 第六版第 4 章，结束项目或阶段。

### 003. 关于项目整合管理的责任描述，正确的是（）。

**选项：**

- A. 项目发起人对整合管理承担最终责任
- B. 项目经理亲自执行项目工作并承担最终责任
- C. 被授权的团队成员对整合管理承担最终责任
- D. 项目经理必须亲自承担项目整合管理的最终责任

**正确答案：D. 项目经理必须亲自承担项目整合管理的最终责任**

**为什么选这个：** 整合管理最终责任在项目经理，且不可授权。

**为什么不选其他：**
- 不选 A：发起人提供授权与资源，不承担整合最终责任。
- 不选 B：项目经理不亲自执行全部项目工作。
- 不选 C：整合不能授权给团队成员承担最终责任。

**出处：** PMBOK 第六版第 4 章，整合由项目经理负责。

### 004. 在敏捷环境下，由谁决定工作的开展方式和整合方式？（）

**选项：**

- A. 项目经理
- B. 产品负责人
- C. 团队成员
- D. 相关方

**正确答案：C. 团队成员**

**为什么选这个：** 敏捷中由团队成员（领域专家）决定工作开展与整合方式。

**为什么不选其他：**
- 不选 A：项目经理侧重营造协作决策环境，而非规定怎么做、怎么整合。
- 不选 B：产品负责人决定做什么（优先级），不决定怎么做。
- 不选 D：相关方不决定工作开展与整合方式。

**出处：** PMBOK 第六版第 4 章，敏捷/适应型环境中的整合。

### 005. 关于项目知识管理，下列说法正确的是（）。

**选项：**

- A. 知识管理主要是将知识记录下来用于分享
- B. 本项目生成的知识可用来支持组织运营或未来的项目
- C. 知识管理主要是管理现有知识
- D. 显性知识易于分享，是知识管理的重点

**正确答案：B. 本项目生成的知识可用来支持组织运营或未来的项目**

**为什么选这个：** 项目知识既服务当前项目，也可沉淀为组织过程资产，支持运营与未来项目。

**为什么不选其他：**
- 不选 A：仅靠记录不够，还要营造分享氛围（尤其隐性知识）。
- 不选 C：不只管理现有知识，也包括创造与运用新知识。
- 不选 D：显性易记录，但知识管理更难也更关键的是隐性知识分享。

**出处：** PMBOK 第六版第 4 章，管理项目知识。

### 006. 以下哪个文件授权项目经理为开展项目活动而动用组织资源？（）

**选项：**

- A. 项目章程
- B. 项目管理计划
- C. 协议
- D. 项目文件

**正确答案：A. 项目章程**

- 不选 C：协议是买卖双方约定，不授权项目经理动用组织资源。
- 不选 D：项目文件是各类执行文件总称，非授权文件。

**出处：** PMBOK 第六版第 4 章，制定项目章程。

### 007. 以下哪项是结束项目或阶段过程的输入？（）

### 008. 拟派的项目经理拿到2年前完成的商业论证报告，着手项目的启动工作，他首先应该（）。

**选项：**

- A. 据此编写项目章程
- B. 要求组织对他正式任命
- C. 审查商业论证报告的内容和结论
- D. 审查并修改商业论证报告

**正确答案：C. 审查商业论证报告的内容和结论**

**为什么选这个：** 商业论证已过时（2年前），启动前先审查内容与结论是否仍有效。

**为什么不选其他：**
- 不选 A：未先确认仍有效，不能直接据此写章程。
- 不选 B：正式任命随章程授权，不是针对旧商业论证的第一步。
- 不选 D：商业论证通常由发起人等负责，项目经理先审查而非自行修改。

**出处：** PMBOK 第六版第 4 章，制定项目章程与商业论证。

### 010. 对于在完工前就提前终止的项目，应该在哪个过程中调查和记录提前终止的原因？（）

**选项：**

- A. 结束项目或阶段
- B. 指导与管理项目工作
- C. 制订项目管理计划
- D. 适用于项目所处阶段的具体过程

**正确答案：A. 结束项目或阶段**

**为什么不选其他：**
- 不选 B：指导与管理是执行工作，不专门调查终止原因。
- 不选 C：制订计划属规划，非收尾调查。
- 不选 D：提前终止原因统一在结束项目或阶段处理。

**出处：** PMBOK 第六版第 4 章，结束项目或阶段。

### 011. 项目开工会议的召开，意味着（）。

- A. 启动阶段结束，规划阶段开始

- C. 执行阶段结束，监控阶段开始

### 012. 关于项目整合管理说法正确的是（）。

**选项：**

- A. 整合管理只需要在项目内部开展

- 不选 D：再简单的项目也需要整合。

**出处：** PMBOK 第六版第 4 章，项目整合管理概述。

### 015. 以下哪项最准确地描述了开展项目变更控制的目的？（）

- A. 确保所有变更都经过审批，确保只有经批准的变更才能纳入计划中

### 016. 执行以下哪项会导致基准的变更？（）

**选项：**

- A. 采取措施预防不利绩效偏差的出现

**出处：** PMBOK 第六版第 4 章，变更请求与基准。
