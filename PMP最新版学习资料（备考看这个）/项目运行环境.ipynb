# 第二章：项目运行环境

每道题包含：题目、选项、正确答案、为什么选这个、为什么不选其他。

### 001. 项目应该在何时更新组织过程资产？（）

**选项：**

- A. 完成一个阶段时
- B. 完成一个里程碑时
- C. 项目全部完成时
- D. 项目全过程中

**正确答案：D. 项目全过程中**

**为什么选这个：**

组织过程资产（OPA）包括流程、模板、经验教训、历史信息等。项目要在**全过程**持续更新它们（尤其是经验教训登记册），让当前和后续项目都能用上，而不是攒到某个时点再一次性补。

**为什么不选其他：**

- 不选 A：阶段结束时可以更新，但不是唯一时机。
- 不选 B：里程碑时也可以更新，同样不是唯一时机。
- 不选 C：全部完成时才更新就太晚，过程中的教训无法及时复用，也容易遗漏。

**记忆要点：** OPA / 经验教训 = 全过程更新，不要等到收尾再补。

### 002. 总结第一阶段工作时，项目专家指出，来自组织过程资产的工作程序存在严重问题，作为项目经理的你应该如何处理？（）

**选项：**

- A. 决定在第二阶段优化该工作程序
- B. 向PMO上报此问题
- C. 说明这是组织规定的，不能更改
- D. 请专家制定新的工作程序

**正确答案：B. 向PMO上报此问题**

**为什么选这个：**

来自组织过程资产的工作程序是组织级资产，通常由 PMO 维护。项目经理发现严重问题应**上报 PMO**，由组织决定是否修订，不能自己改组织流程。

**为什么不选其他：**

- 不选 A：项目经理无权单方面决定优化组织级工作程序。
- 不选 C：组织规定不是绝对不能改；有严重问题应升级处理，而不是直接拒绝。
- 不选 D：请专家另写一套，等于绕过组织资产所有者，新程序也无法自动成为组织过程资产。

**记忆要点：** 组织级流程有问题 → 报 PMO；项目经理不私自改 OPA。

### 003. 某公司的PMO虽然不直接管控项目，但有权要求项目采用公司特定的项目管理方法论。这是什么类型的PMO？（）

**选项：**

- A. 支持型
- B. 控制型
- C. 指令型
- D. 教练型

**正确答案：B. 控制型**

**为什么选这个：**

控制型 PMO 提供支持，并要求项目遵守指定的方法论、模板或治理框架，控制程度中等。题干「不直接管控项目，但有权要求采用公司方法论」正是控制型。

**为什么不选其他：**

- 不选 A：支持型只当顾问，提供模板、培训、经验教训，控制程度低，不能强制采用方法论。
- 不选 C：指令型直接管理和控制项目，项目经理由 PMO 委派，控制程度高。本题明确不直接管控。
- 不选 D：教练型不是 PMI 三种标准 PMO 类型。

**记忆要点：** 支持型=顾问（低）；控制型=强制合规（中）；指令型=直接管项目（高）。

### 004. 在整个项目期间不断更新问题与缺陷管理数据库，写入问题与缺陷的最新状态、解决方案以及相关行动的结果。这项工作应该由谁完成？（）

**选项：**

- A. 项目团队
- B. 职能经理
- C. PMO
- D. 项目经理

**正确答案：A. 项目团队**

**为什么选这个：**

问题与缺陷管理数据库是组织过程资产。谁在一线处理问题和缺陷，谁就写入最新状态、解决方案和行动结果，所以由**项目团队**在整个项目期间持续更新。

**为什么不选其他：**

- 不选 B：职能经理管本部门资源和专业指导，不负责维护项目问题/缺陷库。
- 不选 C：PMO 维护组织级方法论和资产框架，不代替团队登记每个问题和缺陷。
- 不选 D：项目经理确保更新发生、把关质量，但「写入状态和结果」是团队的执行工作。

**记忆要点：** 问题/缺陷库、经验教训的具体填写 → 项目团队；全过程更新（接 001 题）。

### 005. 强矩阵组织和弱矩阵组织之间的主要差别在于（）。

**选项：**

- A. 组织文化对项目的影响不同
- B. 项目团队成员之间以及他们与项目经理之间的亲近度不同
- C. 项目经理对项目资源的控制程度不同
- D. 项目团队成员之间的凝聚力不同

**正确答案：C. 项目经理对项目资源的控制程度不同**

**为什么选这个：**

矩阵组织按项目经理权力大小区分：弱矩阵接近职能型，项目经理更像协调员；强矩阵接近项目型，项目经理对资源控制更强。主要差别就是**项目经理对项目资源的控制程度**。

**为什么不选其他：**

- 不选 A：组织文化会影响项目，但不是强/弱矩阵的划分标准。
- 不选 B：亲近度不是组织结构分类依据。
- 不选 D：凝聚力也不是强矩阵和弱矩阵的本质区别。

**记忆要点：** 弱矩阵 ≈ 职能经理权大；平衡矩阵 ≈ 分权；强矩阵 ≈ 项目经理权大。

### 006. 经过实地考察，项目经理发现当地的关税非常高，原材料进口很不划算。这个发现应该被记录为（）。

**选项：**

- A. 组织过程资产
- B. 风险
- C. 制约因素
- D. 事业环境因素

**正确答案：D. 事业环境因素**

**为什么选这个：**

关税、法规、市场条件是组织外部已经存在的环境，属于**事业环境因素（EEF）**。题目是实地考察后确认的既成事实，不是尚未发生的不确定事件。

**为什么不选其他：**

- 不选 A：组织过程资产是组织内部的流程、模板、经验教训，不是外部关税。
- 不选 B：风险是不确定的未来事件。关税已经很高、已经发现，不是「可能发生」。
- 不选 C：制约因素是对项目的明确限制（如必须某日完工、预算上限）。高关税是环境事实，可能衍生出制约，但发现本身应记为事业环境因素。

**记忆要点：** 已存在的外部条件（法规、关税、市场）→ EEF；不确定的未来事件 → 风险；组织内部流程模板 → OPA。

### 007. 项目管理办公室(PMO)是（）。

**选项：**

- A. 组织中对所辖各项目进行集中协调管理的一个临时性部门
- B. 组织中对所辖各项目进行集中协调管理的一个永久性部门
- C. 组织中对相互关联的各项目进行集中协调管理的一个职能部门
- D. 组织中对相互关联的各项目进行集中协调管理的一个项目部门

**正确答案：B. 组织中对所辖各项目进行集中协调管理的一个永久性部门**

**为什么选这个：**

PMO 是组织里标准化治理、共享资源和方法论的**常设**结构，对**所辖**项目做集中协调。所辖项目不必相互配套。

**为什么不选其他：**

- 不选 A：PMO 不是临时部门。项目临时，PMO 常设。
- 不选 C、D：「相互关联的各项目」是项目集的特征，不是 PMO 的定义。PMO 管的是管辖范围内的项目，彼此可以无关。

**记忆要点：** PMO = 永久 + 所辖项目；项目集 = 相互关联 + 更大收益。

### 009. 项目的技术工作已经全部完成。产品也通过了最终验收，接着应该开展以下哪一项工作？（）

**选项：**

- A. 写项目总结
- B. 更新问题日志
- C. 遣散团队成员
- D. 举办庆功宴

**正确答案：A. 写项目总结**

**为什么选这个：**

技术完成且产品终验通过后，进入收尾：编写最终报告/项目总结、归档、把经验教训转入组织过程资产。这些还需要团队完成，所以下一步是写项目总结。

**为什么不选其他：**

- 不选 B：问题日志应在全过程更新，不是终验之后的关键下一步。
- 不选 C：遣散团队是收尾的**最后**一步。先总结归档，再放人；过早解散就没人做收尾。
- 不选 D：庆功宴可以有，但不是项目管理规定的下一步工作。

**记忆要点：** 验收 → 总结/归档/更新 OPA → 最后才解散团队。

### 011. 很多组织都会在不同的组织层级、不同的时间，针对不同的项目，采用不同的组织结构。这种组织通常被称为（）。

**选项：**

- A. 灵活型组织
- B. 适应型组织
- C. 混合型组织
- D. 项目型组织

**正确答案：C. 混合型组织**

**为什么选这个：**

组织在不同层级、不同时间、对不同项目采用不同结构（有的职能、有的矩阵、有的项目型），这就是**混合型/复合型组织**。教材原句基本就是这道题。

**为什么不选其他：**

- 不选 A：灵活型不是 PMI 标准组织结构名称。
- 不选 B：适应型多用来形容开发生命周期或敏捷组织氛围，不是这种「多种结构并存」的叫法。
- 不选 D：项目型是其中一种结构（项目经理权力最大），不是「同时用多种结构」。

**记忆要点：** 多种结构混用 → 混合型；全是项目经理说了算 → 项目型。

### 012. 输入的变更并不会产生可预测的输出，这是系统的哪个原则的体现？（）

**选项：**

- A. 系统可以优化
- B. 系统及其组件不能同时优化
- C. 系统是动态的
- D. 系统呈现非线性响应

**正确答案：D. 系统呈现非线性响应**

**为什么选这个：**

系统原则里有一条：系统对变化是**非线性响应**——输入变了，输出不会成比例、可预测地变。题目几乎是教材原句。

**为什么不选其他：**

- 不选 A：系统可以优化，说的是整体可以调优，不解释输入输出不可预测。
- 不选 B：系统和组件不能同时最优，讲的是局部最优 ≠ 整体最优。
- 不选 C：系统是动态的，指系统会随时间变化，不是「输入变了输出不可预测」。

**记忆要点：** 输入变了但输出不可预测 → 非线性响应。

### 013. 在以下哪种组织结构下，项目执行组织中的各职能部门参与项目的程度最低？（）

**选项：**

- A. 职能型组织
- B. 平衡矩阵型组织
- C. 强矩阵组织
- D. 项目型组织

**正确答案：D. 项目型组织**

**为什么选这个：**

项目型组织里，成员全职向项目经理汇报，项目自带班子。职能部门对项目的参与程度最低。

**为什么不选其他：**

- 不选 A：职能型里项目在部门内完成，职能部门参与程度最高。
- 不选 B：平衡矩阵里职能经理和项目经理分权，职能部门仍深度参与。
- 不选 C：强矩阵里项目经理权更大，但成员往往仍有职能经理，部门还有参与。

**记忆要点：** 职能部门参与：职能型最高 → 弱/平衡/强矩阵递减 → 项目型最低。和项目经理权力正好相反。

### 014. PMO最基本的职责是（）。

**选项：**

- A. 直接管理和控制某些特定项目
- B. 指令项目管理团队使用特定的模板
- C. 为项目提供咨询和指导
- D. 开展特定项目的变更控制

**正确答案：C. 为项目提供咨询和指导**

**为什么选这个：**

PMO 职责可以从支持到直接管项目。**最基本**的是提供咨询、指导、模板和培训，也就是支持型都会做的事。

**为什么不选其他：**

- 不选 A：直接管控项目是指令型 PMO，控制程度最高，不是最基本职责。
- 不选 B：强制使用特定模板是控制型 PMO 的特点。
- 不选 D：具体项目的变更控制是项目经理/CCB 的事，不是 PMO 的基本职责。

**记忆要点：** 最基本 = 支持咨询；强制模板 = 控制型；直接管项目 = 指令型。接 003 题。

### 015. 在以下哪种项目组织中，项目经理能对项目资源进行最有力的控制？（）

**选项：**

- A. 项目指挥部组织
- B. 项目型组织
- C. 平衡式矩阵组织
- D. 强矩阵组织

**正确答案：B. 项目型组织**

**为什么选这个：**

项目经理对资源的控制程度：职能型最低，弱矩阵、平衡矩阵、强矩阵递增，**项目型最高**（成员全职向项目经理汇报）。

**为什么不选其他：**

- 不选 A：项目指挥部不是 PMI 标准组织结构，不能作为「权力最大」的标准答案。
- 不选 C：平衡矩阵是职能经理和项目经理分权，项目经理不是最有力。
- 不选 D：强矩阵里项目经理权已经很大，但成员通常仍有职能经理，不如项目型彻底。

**记忆要点：** 资源控制权最大 = 项目型。接 005、013 题。

### 016. 某大型集团公司，对每一个产品或服务类别分别设立自主性很大的部门来集中管理。该集团公司采用的是以下哪种组织结构？（）

**选项：**

- A. 多部门型
- B. 项目型
- C. 职能型
- D. 说不清

**正确答案：A. 多部门型**

**为什么选这个：**

多部门型（事业部型）按产品、服务类别、客户或地域划分，各部门自主性高，内部通常自带一套职能。题干「按产品或服务类别设立自主性很大的部门」就是它。

**为什么不选其他：**

- 不选 B：项目型是为项目成立临时班子，项目结束即解散，不是按产品类别设常设部门。
- 不选 C：职能型按专业职能划分（生产、销售、财务），不是按产品/服务类别高度分权。
- 不选 D：特征已经够清楚，可以判断。

**记忆要点：** 按产品线/地域分权、部门很自主 → 多部门型；按专业工种划分 → 职能型。

### 017. 以下哪项是项目管理办公室（PMO）的重要特性？（）

**选项：**

- A. 用特定的资源去实现具体项目的目标
- B. 编制项目管理的规章制度
- C. 对在项目与职能部门兼职的人员进行管理
- D. 随项目的完成而解散

**正确答案：B. 编制项目管理的规章制度**

**为什么选这个：**

PMO 的重要特性是标准化项目治理：制定方法论、规章制度、模板和流程，并在所辖项目间共享资源与经验。

**为什么不选其他：**

- 不选 A：用资源实现某个具体项目目标，是项目经理和项目团队的职责。
- 不选 C：矩阵里兼职人员由项目经理和职能经理双重管理，不是 PMO 的典型特性。
- 不选 D：PMO 是永久性部门，不随单个项目结束而解散（见 007 题）。

**记忆要点：** PMO 定制度、给支持；项目经理拿资源把项目做完。

### 018. 在项目的执行和监控过程组，项目管理团队最有可能利用以下哪种组织过程资产？（）

**选项：**

- A. 项目生命周期指南
- B. 项目收尾指南
- C. 相关的行业标准
- D. 变更控制程序

**正确答案：D. 变更控制程序**

**为什么选这个：**

执行和监控过程组要实施已批准变更、做整体变更控制，最需要的组织过程资产是**变更控制程序**。

**为什么不选其他：**

- 不选 A：项目生命周期指南更多用于启动和规划，用来划分阶段。
- 不选 B：项目收尾指南用于收尾过程组。
- 不选 C：行业标准属于事业环境因素（EEF），不是组织过程资产。

**记忆要点：** 执行/监控 → 变更程序（OPA）；行业标准 → EEF；收尾指南 → 收尾。

### 019. 在敏捷项目中，对工作进行频繁的优先级排序，目的是（）。

**选项：**

- A. 满足客户需求，提升满意度
- B. 提升团队应对变化的能力
- C. 优先完成最具价值的工作
- D. 避免项目失败

**正确答案：C. 优先完成最具价值的工作**

**为什么选这个：**

敏捷里频繁排优先级（待办列表排序），直接目的就是**先做价值最高的工作**，即使中途停也能先交付最大价值。

**为什么不选其他：**

- 不选 A：客户满意是结果，不是「频繁排序」这个动作的直接目的。
- 不选 B：应对变化是敏捷整体能力，排序是手段之一，但本题问的是排序本身为了什么。
- 不选 D：排序不能保证避免失败，说法过满。

**记忆要点：** 频繁排优先级 = 先做最有价值的。接第一章 023 题。

### 020. 项目团队刚实施过一个和本项目非常类似的项目，但你发现某团队成员又出现了和上次一样的失误。你应该要求该团队成员（）。

**选项：**

- A. 查阅他过去的工作总结
- B. 询问其他团队成员
- C. 查阅组织过程资产
- D. 接受培训

**正确答案：C. 查阅组织过程资产**

**为什么选这个：**

类似项目的经验教训、历史信息应已进入**组织过程资产**。同一失误再出现，说明没用这些资产。应要求该成员查阅 OPA，避免重复犯错。

**为什么不选其他：**

- 不选 A：个人工作总结不完整，也不等于组织级经验教训库。
- 不选 B：问同事是非正式途径，不如查已文档化的组织过程资产。
- 不选 D：培训可能有用，但知识已经在组织里，应先用现成的历史信息，而不是一上来就培训。

**记忆要点：** 重复犯错 → 去查 OPA/经验教训。接 001 题：OPA 要全过程更新，就是为了这种情况。

### 021. 关于系统与组件的关系，以下说法正确的是（）。

**选项：**

- A. 系统的功能就是各组件的功能之和
- B. 系统和组件必须同时优化
- C. 组件不可以优化，但系统可以优化
- D. 系统可以实现单个组件无法实现的成果

**正确答案：D. 系统可以实现单个组件无法实现的成果**

**为什么选这个：**

系统有涌现性：组合在一起能实现单个组件做不到的成果，整体大于部分之和。

**为什么不选其他：**

- 不选 A：系统功能不是组件功能的简单相加。
- 不选 B：系统和组件**不能**同时优化。局部最优往往损害整体最优。
- 不选 C：组件可以优化，只是不能和系统同时最优。

**记忆要点：** 系统 > 组件之和；系统和零件不能同时最优。接 012 题非线性。

### 023. 组织过程资产可以分为哪两大类？（）

**选项：**

- A. 过程、政策与程序，组织知识库
- B. 标准与规范、项目管理信息系统
- C. 模板与指南、基础设施

1. **过程、政策与程序**（流程、模板、变更程序、收尾指南等）

### 024. 你的公司刚刚成立，只有9名员工，每个人没有明确的分工，一人要身兼多职。有问题大家商量着办，最终由你来决策。这是采用的哪种组织结构形式？（）

**选项：**

- A. 简约型
- B. 职能型
- C. 弱矩阵型
- D. 紧密型

**正确答案：A. 简约型**

**为什么选这个：**

简约型（有机型/简单型）常见于创业小公司：人少、几乎没有正式组织结构图、一人多职、灵活协作，所有者/经营者做决策。题干全部对上。

**为什么不选其他：**

- 不选 B：职能型有明确专业分工和部门，本题「没有明确分工」。
- 不选 C：弱矩阵要有项目经理和职能经理两条线，小公司尚未形成。
- 不选 D：紧密型不是 PMI 标准组织结构。

**记忆要点：** 创业小团队、一人多职、老板拍板 → 简约型。

### 025. 关于事业环境因素，以下哪个说法是错误的？（）

**选项：**

- A. 事业环境因素可能对项目产生积极或消极的影响
- B. 事业环境因素包括组织治理和组织结构
- C. 事业环境因素可以来自组织内部或外部
- D. 项目管理过程不会导致对事业环境因素的更新

**正确答案：D. 项目管理过程不会导致对事业环境因素的更新**

**为什么选这个：**

题干问「错误」。少数过程会更新事业环境因素，主要是资源管理：获取资源、建设团队、管理团队、控制资源（例如资源可用性、人员档案）。所以「不会导致更新」这句话本身是错的。

**为什么不选其他：**

- 不选 A：事业环境因素可促进也可制约项目，对结果可好可坏。
- 不选 B：内部事业环境因素包含组织文化、结构和治理。
- 不选 C：事业环境因素分组织内部和外部两类。

**记忆要点：** 组织过程资产全过程常更新；事业环境因素只在少数过程更新，不是「从不更新」。

### 026. 组织治理框架会直接影响以下哪个方面？（）

**选项：**

- A. 项目沟通管理的方式
- B. 项目风险管理的方式
- C. 项目目标的设定和实现方式
- D. 组织目标的设定和实现方式

**正确答案：D. 组织目标的设定和实现方式**

**为什么选这个：**

治理是组织内行使权力的框架（规则、政策、程序、规范、关系、系统和过程）。它直接影响：组织目标如何设定与实现、风险如何监控与评估、绩效如何优化。选项 D 是原文对应。

**为什么不选其他：**

- 不选 A：沟通管理方式由沟通管理计划和过程决定，治理只是约束环境。
- 不选 B：风险管理方式属于项目管理知识领域，治理影响的是组织层面的风险监控评估。
- 不选 C：主语是「组织治理」，对应「组织目标」，不是「项目目标」。项目治理才更贴近项目目标如何达成。

**记忆要点：** 组织治理 → 组织目标；项目治理 → 项目怎么管。看到「治理框架」先对层级。

### 027. PMO提供了一份组织预先批准的供应商清单，项目经理可以参考它以缩短供应商的筛选时间。这份清单对于项目经理是（）。

**选项：**

- A. 组织内部的事业环境因素
- B. 组织外部的事业环境因素
- C. 组织过程资产中的流程与程序
- D. 组织过程资产中的组织知识库

**正确答案：C. 组织过程资产中的流程与程序**

**为什么选这个：**

预先批准的供应商清单属于组织过程资产里的「流程、政策和程序」：模板、标准合同、预审供应商名单等，用来按规定开展工作。PMO 提供、组织已批准、可直接拿来缩短筛选，正是这类资产。

**为什么不选其他：**

- 不选 A：内部事业环境因素是组织文化、结构、基础设施、资源可用性等约束条件，不是组织做好的清单模板。
- 不选 B：外部事业环境因素才是市场状况、市场上有哪些供应商；本题是组织自己预先批准的名单。
- 不选 D：组织知识库是经验教训、历史信息、缺陷库、财务数据库等「过去留下的数据」，不是现行流程用的批准清单。

**记忆要点：** 预先批准供应商清单、合同模板 → 过程资产的流程与程序；市场上有谁能供货 → 外部事业环境因素。

### 028. 你在一个组织中做着项目管理的工作，但是没有项目经理的正式头衔。你所在部门的领导对你有100%的绩效打分权。目前你正处于哪种组织结构中？（）

**选项：**

- A. 简约型
- B. 矩阵型
- C. 职能型
- D. 虚拟型

**正确答案：C. 职能型**

**为什么选这个：**

职能型里项目工作由职能经理主导，做项目管理的人常是项目联络员或协调员，没有正式项目经理头衔，绩效完全由部门领导评定。题干两点都对上。

**为什么不选其他：**

- 不选 A：简约型是人少、一人多职、老板拍板的创业结构，不是「部门领导打绩效」。
- 不选 B：矩阵型至少有项目经理和职能经理两条汇报线；弱矩阵也有项目经理头衔，只是权力小。本题明确没有正式头衔。
- 不选 D：虚拟型强调地理位置分散、靠通信技术协作，与头衔和绩效权无关。

**记忆要点：** 没项目经理头衔 + 职能经理100%绩效 → 职能型；有头衔但权力很小 → 弱矩阵。

### 029. 项目治理是指（）。

**选项：**

- A. 把权力集中在项目经理手中
- B. 加强公司CEO对项目的控制力度
- C. 在不划分项目阶段的前提下，把项目作为一个整体加以控制
- D. 对项目的高层次的指导、支持和监控

**正确答案：D. 对项目的高层次的指导、支持和监控**

**为什么选这个：**

项目治理是与组织治理对齐的监督职能，贯穿项目生命周期，给项目经理和团队提供结构、过程、决策模型和工具，从高层次指导、支持和监控项目，使其达成组织战略与运营目标。

**为什么不选其他：**

- 不选 A：治理是组织/发起人/治理委员会的监督框架，不是把权集中给项目经理。项目经理权力大小由组织结构决定。
- 不选 B：治理不等于 CEO 亲自加码管控；通常由治理委员会、发起人、PMO 等按框架履职。
- 不选 C：治理常借助阶段和阶段关口做决策，不是「不划分阶段、当一个整体控制」。

**记忆要点：** 治理 = 高层指导、支持、监控；管理 = 项目经理把工作做出来。

### 030. 公司本部只有三名员工，其他人都通过互联网互相配合完成项目工作。其中一部分人是全职的，一部分人是兼职的。这家公司采用的是哪种组织结构形式？（）

**选项：**

- A. 职能型
- B. 简约型
- C. 虚拟型
- D. 混合型

**正确答案：C. 虚拟型**

**为什么选这个：**

虚拟型（网络型）组织核心人员很少，其余通过通信技术远程协作，成员可以全职或兼职、地理分散。本部三人、互联网配合、全职加兼职，全部对上。

**为什么不选其他：**

- 不选 A：职能型按专业设部门、同地办公为主，不是靠互联网把大部分人连起来。
- 不选 B：简约型是小公司一人多职、当面商量、老板拍板（如 024 题 9 人无明确分工），不强调远程网络协作。
- 不选 D：混合型是同时采用两种以上结构（如有的部门职能型、有的矩阵型），题干只描述一种远程协作形态。

**记忆要点：** 核心班子很小 + 互联网远程配合 → 虚拟型；人少但面对面一人多职 → 简约型。

### 031. 以下关于PMO的说法，正确的是（）。

**选项：**

- A. PMO的职责范围总是一样的
- B. PMO只管理隶属于同一个项目集的各个项目
- C. PMO通常就是大型项目的项目部
- D. PMO所管理的项目不一定彼此关联

**正确答案：D. PMO所管理的项目不一定彼此关联**

**为什么选这个：**

PMO 是标准化治理并共享资源、方法、工具和技术的组织实体，所辖项目可以只是被放在一起管理，彼此不必有配套关系。这正是它和项目集的差别。

**为什么不选其他：**

- 不选 A：PMO 有支持型、控制型、指令型，职责和支持力度不同。
- 不选 B：同一项目集里互相关联的项目由项目集经理管；PMO 不限于某一个项目集。
- 不选 C：大型项目的项目部是该项目团队的一部分，临时随项目结束；PMO 通常是永久性组织部门。

**记忆要点：** 项目集 = 项目必须相互关联；PMO = 所管项目可以互不相关。

### 032. 公司既有职能部门也有项目部，但职能经理对每个人的绩效考核输入占80%，项目经理只有20%的绩效考核权，这很有可能是哪种组织结构形式的公司？（）

**选项：**

- A. 职能型
- B. 弱矩阵型
- C. 强矩阵型
- D. 虚拟型

**正确答案：B. 弱矩阵型**

**为什么选这个：**

同时存在职能部门和项目部，是矩阵结构（双线汇报）。绩效权 80% 在职能经理、20% 在项目经理，项目经理权力很小，对应弱矩阵。

**为什么不选其他：**

- 不选 A：职能型通常没有真正的项目部，部门领导几乎 100% 打绩效（见 028 题）。本题已有项目部，只是项目经理权很小。
- 不选 C：强矩阵是项目经理权力大于职能经理，绩效权应明显偏向项目经理。
- 不选 D：虚拟型看的是远程网络协作，与绩效权比例无关。

**记忆要点：** 有两条线就是矩阵；谁打分比重大，权力就偏向谁。职能经理远大于项目经理 → 弱矩阵。

### 033. 系统中，各组件的状态、组件之间的关系以及整个系统的状态都是不断变化的，这是对系统的哪个特性的描述？（）

**选项：**

- A. 动态性
- B. 开放性
- C. 可优化性
- D. 非线性响应性

**正确答案：A. 动态性**

**为什么选这个：**

动态性指系统一直在变：组件状态、组件之间的关系、系统整体状态都不断变化。题干几乎是定义原文。

**为什么不选其他：**

- 不选 B：开放性指系统与外部环境交换信息、资源，题干没提环境交互。
- 不选 C：可优化性指系统和组件都能优化，但不能同时优化。
- 不选 D：非线性响应性指输入变化不会带来可预测的输出变化（见 012 题）。

**记忆要点：** 一直在变 → 动态；输入输出对不上 → 非线性；不能同时调到最优 → 可优化性的限制。

### 034. 在以下哪些组织结构中，项目经理有权管理项目预算？（）

**选项：**

- A. 强矩阵型、有机型、PMO型
- B. 强矩阵型、项目型、事业部型
- C. 平衡矩阵型、项目型、PMO型
- D. 强矩阵型、项目型、PMO型

**正确答案：D. 强矩阵型、项目型、PMO型**

**为什么选这个：**

PMBOK 组织结构对照表里，「谁管理项目预算」明确写项目经理的只有：强矩阵、项目型、PMO。这三类项目经理权力都是中高到几乎全权。

**为什么不选其他：**

- 不选 A：有机型（简约型）预算由所有者/经营者管，不是项目经理。
- 不选 B：事业部型（多部门型）预算由职能经理管。
- 不选 C：平衡矩阵预算是「混合」（职能经理和项目经理分担），不能算项目经理有权独管。

**记忆要点：** 预算在项目经理手里 = 强矩阵 / 项目型 / PMO。弱矩阵和职能型在职能经理手里；平衡矩阵是混合。

### 035. 项目进展非常不顺利，新上任的项目经理非常关注所在组织内部的资源的可用性，并要求查看所在组织与外部供应商的相关资源采购协议，他关注的焦点是什么？（）

**选项：**

- A. 组织过程资产
- B. 事业环境因素
- C. 流程和程序
- D. 经验教训知识库

**正确答案：B. 事业环境因素**

**为什么选这个：**

内部资源可用性是内部事业环境因素；已与外部供应商签订的资源采购协议是项目必须接受的约束，属于事业环境因素/协议类输入，项目团队通常无法随意改。新经理盯的是「手头能用什么、外面已经锁死了什么」。

**为什么不选其他：**

- 不选 A：组织过程资产是流程、模板、知识库等组织自己积累、项目可以更新的东西。资源在不在、已签协议是什么，不是过程资产。
- 不选 C：流程和程序是组织过程资产的一类，题干没有提做事的标准流程。
- 不选 D：经验教训知识库也是过程资产，题干没有提历史项目经验。

**记忆要点：** 资源可用性、现有合同约束 → 事业环境因素；模板、流程、经验教训 → 组织过程资产。

### 036. 为了确保项目质量合规，项目团队收集了公司的质量政策和质量标准。这些内容将被列入（）。

**选项：**

- A. 组织内部的事业环境因素
- B. 组织过程资产
- C. 经验教训知识库
- D. 以往项目的档案

**正确答案：B. 组织过程资产**

**为什么选这个：**

公司自己的质量政策和质量标准属于组织过程资产中的「过程、政策与程序」（质量方针、质量体系、组织内部标准）。团队拿来保证合规，正是用组织已有的政策标准。

**为什么不选其他：**

- 不选 A：政府法规、行业强制标准才是事业环境因素。本题是「公司的」政策和标准，组织可以制定和更新。
- 不选 C：经验教训知识库是过程资产的一小类，装的是历史经验，不是现行质量政策。
- 不选 D：以往项目档案同样是知识库内容，不是质量政策本身。

**记忆要点：** 公司自己的政策/标准/模板 → 组织过程资产；政府/行业强制标准 → 事业环境因素。

### 037. 公司的PMO是核心战略部门和职能部门，所有项目经理都要向其汇报。这种情况会在哪种组织结构形式中出现？（）

**选项：**

- A. 项目型
- B. 强矩阵型
- C. 混合型
- D. PMO型

**正确答案：D. PMO型**

**为什么选这个：**

PMBOK 把 PMO 也列为一种组织结构。PMO 作为核心战略部门兼职能部门，所有项目经理向它汇报，项目经理权力高到几乎全权，正是 PMO 型组织。

**为什么不选其他：**

- 不选 A：项目型是按项目组建独立团队，项目经理向高层/发起人汇报，不一定存在作为职能部门的 PMO。
- 不选 B：强矩阵仍有职能经理和项目经理双线，项目经理权力大，但不是「所有项目经理都向 PMO 汇报」。
- 不选 C：混合型是同时采用多种结构，题干只描述 PMO 作为核心职能部门这一种形态。

**记忆要点：** 所有项目经理向 PMO 汇报、PMO 既是战略部门又是职能部门 → PMO 型。


### 038. PMO否决了你的项目提出的某个重大变更，并要求你进行方案优化后再向其汇报。这是哪种类型的PMO？（）

**选项：**

- A. 支持型
- B. 控制型
- C. 指令型
- D. 混合型

**正确答案：C. 指令型**

**为什么选这个：**

指令型 PMO 直接管理和控制项目，项目经理向 PMO 汇报，控制程度高。能否决重大变更、要求优化后再汇报，说明决策权在 PMO，属于指令型。

**为什么不选其他：**

- 不选 A：支持型只提供模板、培训、经验教训等咨询，控制程度低，不能否决变更。
- 不选 B：控制型要求遵守方法论/模板，中等控制；「不直接管项目但强制用方法论」才是控制型（见 003 题）。本题是直接否决项目变更。
- 不选 D：混合型不是 PMO 的三种标准类型。

**记忆要点：** 支持型给建议；控制型强制合规；指令型直接管项目、可拍板否决。


### 039. 以下所有都是组织过程资产的例子，除了（）。

**选项：**

- A. 标准化的工作指南
- B. 过程测量数据库
- C. 工作授权系统
- D. 配置管理知识库

**正确答案：C. 工作授权系统**

**为什么选这个：**

题干问「除了」。工作授权系统规定何时、由谁、按什么顺序开展工作，属于项目管理信息系统，是事业环境因素，不是组织过程资产。

**为什么不选其他：**

- 不选 A：标准化工作指南属于过程资产中的流程、政策与程序。
- 不选 B：过程测量数据库属于过程资产中的组织知识库。
- 不选 D：配置管理知识库也属于组织知识库。

**记忆要点：** 指南/模板/各类知识库 → OPA；工作授权系统、PMIS → EEF。


### 040. PMO会对它所辖多个项目的哪些资源进行管理？（）

**选项：**

- A. 全部资源
- B. 共享资源
- C. 各项目的稀缺人力资源
- D. 实物资源

**正确答案：B. 共享资源**

**为什么选这个：**

PMO 的一项主要职能是：在其所辖全部项目之间管理共享资源，并促进方法论、工具和技术的共享。管的是跨项目共用的那部分，不是每个项目自己的全部资源。

**为什么不选其他：**

- 不选 A：各项目专属资源仍由项目经理/职能经理管，PMO 不管「全部资源」。
- 不选 C：稀缺人力可能是共享资源的一种，但范围太窄；PMO 管的是各类共享资源，不限于稀缺人力。
- 不选 D：实物资源同样只是共享资源的一种可能，不是定义本身。

**记忆要点：** PMO 管跨项目的共享资源；项目经理管本项目资源。


### 041. 在以下哪种组织结构中，项目团队成员最需要在特定时间临时集中办公？（）

**选项：**

- A. 项目型
- B. 多部门型
- C. 虚拟型
- D. PMO型

**正确答案：C. 虚拟型**

**为什么选这个：**

虚拟型组织地理分散、靠通信技术协作。平时不在一起，最需要在开工、关键节点等特定时间临时集中办公（或开面对面会）来补沟通。

**为什么不选其他：**

- 不选 A：项目型团队通常专职、常驻项目，本身就在一起工作，不是「临时」集中。
- 不选 B：多部门型按产品/服务设高度自主的事业部，成员主要在本部门办公。
- 不选 D：PMO 型强调汇报关系和治理，与是否集中办公无直接对应。

**记忆要点：** 虚拟型平时分散 → 最需要临时集中办公；项目型本来就集中。


### 042. 当项目经理遇到困难时，可以向PMO求助。因为PMO的主要职能是（）。

**选项：**

- A. 制定和管理项目共享文件
- B. 对项目进行监督
- C. 向项目经理提供支持
- D. 提供资金

**正确答案：C. 向项目经理提供支持**

**为什么选这个：**

PMO 的一项主要职能就是以多种方式支持项目经理（辅导、培训、方法论、共享资源等）。题干「遇到困难向 PMO 求助」对应的就是提供支持。与 014 题「最基本职责是咨询指导」同一意思。

**为什么不选其他：**

- 不选 A：管理共享文件只是支持手段之一，不是「求助」的原因。
- 不选 B：监督是控制型/指令型的管控色彩，不是项目经理遇到困难去求助的理由。
- 不选 D：资金由发起人/组织提供，不是 PMO 的主要职能。

**记忆要点：** PMO 最基本/主要 = 支持项目经理；监督、否决是控制型/指令型加出来的权力。


### 043. 在下列哪一种组织结构中，项目成员在收尾阶段最感到忧心忡忡？（）

**选项：**

- A. 职能型
- B. 矩阵型
- C. 项目型
- D. 弱矩阵型

**正确答案：C. 项目型**

**为什么选这个：**

项目型组织里成员全职归项目，项目结束团队解散。没有稳定的职能部门可回，收尾阶段最担心下一步去哪、会不会失业。

**为什么不选其他：**

- 不选 A：职能型成员编制在部门，项目结束回原岗位，最不慌。
- 不选 B：矩阵型有职能经理这条「老家」，项目结束后回职能部门。
- 不选 D：弱矩阵更接近职能型，成员主要归属部门，收尾更不慌。

**记忆要点：** 项目型权力最大，但收尾最焦虑（没有部门可回）。这是项目型的典型缺点。


### 044. 项目治理必须履行的四大职能是（）。

**选项：**

- A. 监督、控制、管理和决策
- B. 规划、控制、整合和决策
- C. 监督、控制、整合和总结
- D. 监督、控制、整合和决策

**正确答案：D. 监督、控制、整合和决策**

**为什么选这个：**

项目治理的四大职能是：监督（oversight）、控制（control）、整合（integration）、决策（decision making）。对应高层次指导、支持和监控，而不是替项目经理做日常管理。

**为什么不选其他：**

- 不选 A：「管理」是项目经理的工作，治理不代替管理。
- 不选 B：「规划」属于项目管理过程组，不是治理职能。
- 不选 C：「总结」不是四大职能；缺的是决策。

**记忆要点：** 治理四职能 = 监督、控制、整合、决策。治理定框架和拍板，管理把活干完。


### 045. 项目经理有很大职权，使用全职的项目员工。但跨项目的沟通常常让他觉得力不从心，有时必须请高层出面协调。他所在的组织很可能（）。

**选项：**

- A. 没有专门的项目部
- B. 没有分管项目的高层领导
- C. 未设立PMO
- D. 才成立不久

**正确答案：C. 未设立PMO**

**为什么选这个：**

职权大、全职项目员工，说明是项目型或强矩阵，项目内部没问题。缺的是跨项目协调——这正是 PMO 的职能（共享资源、跨项目沟通、标准化）。没有 PMO，只能请高层临时出面。

**为什么不选其他：**

- 不选 A：有全职项目员工且项目经理职权很大，通常已有项目部/项目团队，不是缺项目部。
- 不选 B：题干已经能请高层出面协调，说明有高层，只是没有常态化的跨项目协调机构。
- 不选 D：成立不久与「跨项目沟通无力」没有必然对应。

**记忆要点：** 项目经理强管本项目；跨项目协调靠 PMO。本项目很强、跨项目很弱 → 缺 PMO。


### 046. 你从一个简约型组织结构的公司跳槽到一家强矩阵型结构的组织当项目经理，你将面临的新情况是（）。

**选项：**

- A. 决策都由高层制定
- B. 项目经理说了算
- C. 必须与职能经理合作
- D. 基本没有书面规章制度

**正确答案：C. 必须与职能经理合作**

**为什么选这个：**

简约型是老板拍板、几乎没有部门墙。强矩阵虽项目经理权力大，但仍是职能经理+项目经理双线汇报，资源在职能部门，必须与职能经理合作。这是从简约型过来会遇到的新情况。

**为什么不选其他：**

- 不选 A：决策都由高层定，更像简约型（所有者决策）或职能型，不是强矩阵的新特点。
- 不选 B：项目经理说了算是项目型；强矩阵权力大，但不是一个人说了算。
- 不选 D：基本没有书面制度是简约型的旧环境；强矩阵通常更正规。

**记忆要点：** 矩阵的本质 = 必须和职能经理合作。强矩阵只是项目经理这边权重大一些。


### 047. 为了应对外界变化，公司计划重组以提升组织的敏捷性。此次变革的最大阻碍可能来自于（）。

**选项：**

- A. 集权的管理模式
- B. 独裁的领导风格
- C. 现有的组织文化
- D. 员工的技能不足

**正确答案：C. 现有的组织文化**

**为什么选这个：**

组织变革、提升敏捷性时，最大阻力通常是现有组织文化（价值观、习惯、潜规则）。文化是事业环境因素，项目团队很难快速改变，也最难「培训掉」。

**为什么不选其他：**

- 不选 A：集权管理模式会妨碍敏捷，但它往往是文化的一种表现，不是最大、最底层的阻碍。
- 不选 B：独裁领导风格同理，属于文化/治理的一部分。
- 不选 D：技能不足可以通过培训、辅导弥补，相对好改。

**记忆要点：** 变革/推敏捷，最大阻碍 = 组织文化。技能能教，文化难改。


### 048. 作为控制型PMO管理下的一个项目，项目经理必须做到以下几点，除了（）。

**选项：**

- A. 采用PMO规定的项目管理框架和方法
- B. 服从项目治理架构
- C. 接受PMO对项目的直接管理和控制
- D. 采用PMO规定的模板编制项目文件

**正确答案：C. 接受PMO对项目的直接管理和控制**

**为什么选这个：**

题干问「除了」。控制型 PMO 中等控制：支持之外，强制遵守框架、方法论、模板和治理，但不直接接管项目。直接管理和控制是指令型 PMO 的特征。

**为什么不选其他：**

- 不选 A：采用规定的框架和方法，正是控制型的合规要求（见 003 题）。
- 不选 B：服从项目治理架构，属于控制型要求遵守的治理。
- 不选 D：按规定模板编制文件，也是控制型的典型要求。

**记忆要点：** 控制型 = 必须合规，但不直接管项目；指令型 = 直接管、项目经理向 PMO 汇报。


### 049. 关于组织过程资产的说法，以下哪项是错误的？（）

**选项：**

- A. 执行组织所特有并使用的计划、过程、政策、程序和知识库
- B. 可以来自任何项目执行组织
- C. 包括来自组织以往项目的经验教训和历史信息
- D. 既可以来自项目执行组织的内部，也可来自项目执行组织的外部

**正确答案：D. 既可以来自项目执行组织的内部，也可来自项目执行组织的外部**

**为什么选这个：**

题干问「错误」。组织过程资产是执行组织特有的内部资产。项目若有多个执行组织，可以来自其中任何一个（所以 B 对），但不能来自执行组织之外。内外都有的是事业环境因素。

**为什么不选其他：**

- 不选 A：这是组织过程资产的定义原文。
- 不选 B：涉及多个执行组织时，过程资产可以来自其中任何一方。
- 不选 C：经验教训和历史信息属于组织知识库，是过程资产的第二大类。

**记忆要点：** OPA = 组织内部特有；EEF = 可内可外。看到「内部或外部」先想事业环境因素。


### 050. 以下哪一项包括在组织过程资产中？（）

**选项：**

- A. 历史资料和经验教训
- B. 组织文化和组织结构
- C. 项目管理信息系统
- D. 组织的人事管理制度和人力资源现状

**正确答案：A. 历史资料和经验教训**

**为什么选这个：**

历史资料和经验教训属于组织过程资产中的组织知识库，项目全过程都会更新、供后续项目使用。

**为什么不选其他：**

- 不选 B：组织文化、组织结构是内部事业环境因素，项目团队通常无法改变。
- 不选 C：项目管理信息系统（含工作授权系统）是事业环境因素。
- 不选 D：人事管理制度像政策，似乎像过程资产，但「人力资源现状」是资源可用性，属于事业环境因素；整项不能算过程资产。

**记忆要点：** 经验教训/历史资料/模板 → OPA；文化、结构、PMIS、资源现状 → EEF。


### 051. 《指南》中提及的十六个组织管理原则，其中对于沟通的要求是（）。

**选项：**

- A. 按层级进行沟通
- B. 畅通的沟通渠道
- C. 不受限制的沟通
- D. 优先的沟通渠道

**正确答案：B. 畅通的沟通渠道**

**为什么选这个：**

PMBOK 把一般管理的关键职能/原则列了约 16 条，其中与沟通对应的是畅通（清晰）的沟通渠道（Clear communication channels）。

**为什么不选其他：**

- 不选 A：按层级沟通是传统等级链，不是《指南》这条原则的表述。
- 不选 C：不受限制的沟通不是这 16 条里的原文，夸大了。
- 不选 D：优先的沟通渠道《指南》未这样写。

**记忆要点：** 16 条管理原则里，沟通 = 畅通的沟通渠道。


### 052. 矩阵式组织结构最突出的优点是（）。

**选项：**

- A. 有利于充分利用资源
- B. 需要向两个老板汇报
- C. 可兼顾项目工作与职能工作
- D. 项目经理权力大，对资源有控制权

**正确答案：A. 有利于充分利用资源**

**为什么选这个：**

矩阵把专业人员同时投入项目和职能部门，稀缺资源可跨项目共享，这是它相对项目型（人被锁死在一个项目）最突出的优点。

**为什么不选其他：**

- 不选 B：向两个老板汇报是矩阵最突出的缺点，不是优点。
- 不选 C：兼顾项目与职能工作是矩阵的运作方式，也是优点之一，但不如「充分利用资源」那么核心；它同时带来双头领导的冲突。
- 不选 D：项目经理权力大、控资源是强矩阵/项目型的特点，弱矩阵并不如此，不能概括整个矩阵。

**记忆要点：** 矩阵优点 = 资源共享、利用充分；矩阵缺点 = 两个老板、权责冲突。


### 053. 按照组织应该遵守的基本管理原则，以下哪一项是正确的？（）

**选项：**

- A. 由一个特定的部门开展规划和实施
- B. 允许专家级员工参与规划和实施
- C. 允许任何员工参与规划和实施
- D. 允许高层员工参与规划和实施

**正确答案：C. 允许任何员工参与规划和实施**

**为什么选这个：**

与 051 题同一组「约 16 条组织管理原则」。其中有一条是：对将要完成的工作的规划和实施做公开贡献（Open contribution to planning and execution）。也就是任何员工都可以参与规划和实施，不是少数人的专利。

**为什么不选其他：**

- 不选 A：只由特定部门做，与「公开贡献」相反。
- 不选 B：仅专家参与，范围太窄。
- 不选 D：仅高层参与，同样不是公开贡献。

**记忆要点：** 管理原则里，沟通要畅通；规划实施要开放，任何员工都可参与。


### 054. 项目治理框架应该涉及哪四个治理领域？（）

**选项：**

- A. 一致性、风险、绩效、沟通
- B. 一致性、风险、质量、沟通
- C. 一致性、资源、绩效、沟通
- D. 一致性、风险、绩效、整合

**正确答案：A. 一致性、风险、绩效、沟通**

**为什么选这个：**

治理有「领域」和「职能」两套四条，容易混。四个治理领域是：一致性（Alignment，与战略对齐）、风险、绩效、沟通。

**为什么不选其他：**

- 不选 B：质量不是四个治理领域之一，绩效才是。
- 不选 C：资源不是四个治理领域之一，风险才是。
- 不选 D：整合是治理四大职能之一（见 044 题），不是领域。

**记忆要点：** 治理领域 = 一致性、风险、绩效、沟通；治理职能 = 监督、控制、整合、决策。


### 055. 下列哪种组织结构中，项目最不受关注？（）

**选项：**

- A. 平衡式矩阵
- B. 紧密式矩阵
- C. 职能型
- D. 项目型

**正确答案：C. 职能型**

**为什么选这个：**

职能型按专业部门运作，重心是部门日常工作。项目经理权力很小甚至没有正式头衔，项目只是部门工作的附属，最不受关注。

**为什么不选其他：**

- 不选 A：平衡矩阵项目与职能大致平分关注，项目已被当成正式工作。
- 不选 B：紧密式矩阵通常指强矩阵或集中办公，项目关注度更高。
- 不选 D：项目型整个组织围着项目转，项目最受关注。

**记忆要点：** 项目受关注程度：职能型最低 → 弱矩阵 → 平衡矩阵 → 强矩阵 → 项目型最高。


### 056. 选择组织结构时应该考虑以下因素，除了（）。

**选项：**

- A. 集中办公还是虚拟远程办公
- B. 成本
- C. 职责分配
- D. 设计的稳定性

**正确答案：D. 设计的稳定性**

**为什么选这个：**

题干问「除了」。《指南》选择组织结构的考虑因素包括：职责分配、成本、物理位置（集中/区域/虚拟办公）、设计的适应性、设计的简单性等。写的是「适应性」不是「稳定性」。

**为什么不选其他：**

- 不选 A：物理位置（集中办公还是虚拟）是明确要考虑的因素。
- 不选 B：成本是明确要考虑的因素。
- 不选 C：职责分配、问责安排都是表中的因素。

**记忆要点：** 选结构看适应性和简单性，不看稳定性。位置、成本、职责都要算进去。


### 057. 为实现某个特定目标而开展的一系列活动，只能由一个人来领导。这是以下哪个基本管理原则的要求（）

**选项：**

- A. 统一命令
- B. 统一方向
- C. 行动纪律
- D. 工作职责

**正确答案：B. 统一方向**

**为什么选这个：**

统一方向（Unity of direction）：同一目标的一组活动，只有一个计划、一个领导。题干「为实现某个特定目标而开展的一系列活动，只能由一个人来领导」就是这句话。

**为什么不选其他：**

- 不选 A：统一命令（Unity of command）是「一个人只从一个上级接受命令」，管的是下属对上司，不是一组活动对一个领导。
- 不选 C：行动纪律指尊重权威、规则和他人，与「谁领导这组活动」无关。
- 不选 D：工作职责指按职权恰当分配责任，不是「一组活动一个领导」。

**记忆要点：** 统一命令 = 一个人一个上司；统一方向 = 一个目标一个领导一个计划。


### 058. 在项目型组织中，全职项目团队成员的比例通常是（）。

**选项：**

- A. 很少
- B. 几乎100%
- C. 大约1/3
- D. 大约2/3

**正确答案：B. 几乎100%**

**为什么选这个：**

项目型组织成员专职为项目工作，全职比例通常约 85%–100%，也就是几乎全部全职。这也是收尾时成员最焦虑的原因（见 043 题）。

**为什么不选其他：**

- 不选 A：很少全职是职能型或弱矩阵。
- 不选 C：大约 1/3 更接近弱矩阵到平衡矩阵的区间。
- 不选 D：大约 2/3 更接近强矩阵（约 50%–95%）。

**记忆要点：** 全职比例：职能型很少 → 弱矩阵 0–25% → 平衡 15–60% → 强矩阵 50–95% → 项目型 85–100%。


### 059. 在管理具体项目时，项目管理团队应该根据具体需要裁剪（）。

**选项：**

- A. 组织过程资产
- B. 事业环境因素
- C. 组织结构
- D. 组织文化

**正确答案：A. 组织过程资产**

**为什么选这个：**

组织提供标准过程、政策和程序，项目管理团队要按具体项目裁剪后再用。过程资产是组织的、也可以为项目调整；这正是裁剪的对象。

**为什么不选其他：**

- 不选 B：事业环境因素项目团队通常无法控制，更谈不上裁剪。
- 不选 C：组织结构是内部事业环境因素，由组织决定，不是项目团队裁剪的。
- 不选 D：组织文化同样是事业环境因素，是变革最大阻力（见 047 题），项目团队裁不了。

**记忆要点：** 能裁剪的是过程资产和项目管理过程；裁不了的是事业环境因素（结构、文化、市场、法规）。


### 060. 某人有机会接手管理一个新项目。他决定先全面地调查一下项目执行组织中员工的共享价值观和信念。这些价值观和信念属于（）。

**选项：**

- A. 市场条件
- B. 组织治理
- C. 组织结构
- D. 组织文化

**正确答案：D. 组织文化**

**为什么选这个：**

组织文化就是组织成员共享的价值观、信念、规范和工作习惯。新项目经理先摸清这些，是在识别内部事业环境因素。文化也是变革最大阻力（见 047 题）。

**为什么不选其他：**

- 不选 A：市场条件是外部事业环境因素，与员工信念无关。
- 不选 B：组织治理是行使权力的框架（规则、政策、程序），不是价值观本身。
- 不选 C：组织结构是汇报关系、部门划分，不是共享信念。

**记忆要点：** 共享价值观/信念/习惯 → 组织文化；汇报线 → 结构；权力框架 → 治理；市场行情 → 外部事业环境因素。
