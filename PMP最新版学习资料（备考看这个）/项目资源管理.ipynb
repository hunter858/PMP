# 第九章：项目资源管理

每道题包含：题目、选项、正确答案、为什么选这个、为什么不选其他。
本节先收录已做过的题目，后续继续追加。


## 001. 资源管理中，以下哪项管理过程不会造成事业环境因素的更新？（）

**选项**
- A. 获取资源
- B. 建设团队
- C. 管理团队
- D. 控制资源

**正确答案：D**

**为什么选这个**
获取资源、建设团队、管理团队的输出都包含**事业环境因素更新**（如人员可用性、人事管理信息、员工绩效评价等）。
**控制资源**的输出有工作绩效信息、变更请求、项目管理计划更新、项目文件更新、**组织过程资产更新**，**不包含**事业环境因素更新。

**为什么不选其他**
- 不选 A：获取资源会更新事业环境因素（如资源可用性、组织资源日历相关信息等）。
- 不选 B：建设团队会更新事业环境因素（如员工发展记录、技能评估等）。
- 不选 C：管理团队会更新事业环境因素（如人事绩效评价输入、组织开销费率等）。

**出处**
PMBOK® Guide 第6版 9.3–9.6：获取资源 / 建设团队 / 管理团队 / 控制资源的输出对比。


## 002. 可以使用以下哪种结构来把组织中的部门与项目中的工作包联系起来？（）

**选项**
- A. 工作分解结构
- B. 资源分解结构
- C. 风险分解结构
- D. 组织分解结构

**正确答案：D**

**为什么选这个**
**组织分解结构（OBS）**按照组织现有部门、单元或团队排列，并与工作包对应，用来把组织中的部门与项目工作包联系起来，明确责任归属。

**为什么不选其他**
- 不选 A：WBS 按可交付成果/工作分解项目范围，不直接按组织部门关联。
- 不选 B：资源分解结构按资源类别和类型分层（人力、材料、设备等）。
- 不选 C：风险分解结构按风险来源/类别分层识别风险。

**出处**
PMBOK® Guide 第6版：组织分解结构（OBS）与责任分配矩阵相关概念。


## 003. 以下哪项不是团队章程的内容？（）

**选项**
- A. 团队价值观
- B. 沟通指南
- C. 会议指南
- D. 团队建设

**正确答案：D**

**为什么选这个**
团队章程记录团队共同认可的价值观、沟通与会议指南、决策标准、冲突处理流程等约定。
**团队建设**是建设团队过程中的活动/工具，不是章程文档中的内容条目。

**为什么不选其他**
- 不选 A/B/C：这些都属于团队章程的常见内容。

**出处**
PMBOK® Guide 第6版 9.1.3.2 / 9.3：团队章程；9.4 建设团队—团队建设。


## 004. 责任分配矩阵具有以下作用，除了（）。

**选项**
- A. 反映与每个人有关的所有活动
- B. 反映与每个活动有关的所有人
- C. 为每个工作指定唯一责任点
- D. 使每个人都只负责一项工作

**正确答案：D**

**为什么选这个**
责任分配矩阵（如 RACI）展示人员与活动的对应关系：可从人看相关活动、从活动看相关人员，并为每项工作指定唯一责任点（通常一个 Accountable）。
它并不限制“每个人只能负责一项工作”，同一人可对应多项活动。

**为什么不选其他**
- 不选 A/B：这正是矩阵纵横双向查看的作用。
- 不选 C：RACI 中每项工作通常只有一个 A，即唯一责任点。

**出处**
PMBOK® Guide 第6版：责任分配矩阵（RAM）/ RACI。


## 005. 在敏捷项目中，采用最大限度的集中和协作的团队结构有利于（）。

**选项**
- A. 减少变更的数量
- B. 减少所需的人力资源
- C. 适应快速变化的环境
- D. 降低决策的难度

**正确答案：C**

**为什么选这个**
敏捷强调集中办公与紧密协作，以便快速沟通、即时反馈，从而更好地**适应快速变化的环境**。

**为什么不选其他**
- 不选 A：敏捷不追求减少变更数量，而是更好地应对变更。
- 不选 B：集中协作不等于减少人力。
- 不选 D：决策难度降低不是该结构的主要目的表述。

**出处**
PMBOK® Guide / 敏捷实践：协作型、集中式团队结构。


## 006. 高效的项目团队应该（）。

**选项**
- A. 以领导为导向
- B. 以任务和结果为导向
- C. 集中办公
- D. 通过电子网络联系

**正确答案：B**

**为什么选这个**
高效项目团队的核心特征是以**任务和结果为导向**，聚焦可交付成果与目标达成。

**为什么不选其他**
- 不选 A：高效团队更强调目标与成果，而非单纯以领导为导向。
- 不选 C/D：集中办公、电子网络是协作方式，不是高效团队的本质定义。

**出处**
PMBOK® Guide 第6版 9.4 建设团队—高效团队特征。


## 007. 通过审查项目绩效评估文件，以及与团队成员进行交谈，项目经理发现需要开展一项计划外的技能培训，接下来他应该（）。

**选项**
- A. 什么也不做，只能按计划开展培训
- B. 动用管理储备开展计划外培训
- C. 向组织申请计划外的培训费用
- D. 压缩其他培训费用

**正确答案：B**

**为什么选这个**
计划外技能培训属于事先未纳入计划的需要，更接近用**管理储备**应对（未知—未知）；使用时通常需经管理层批准/变更控制，把资金纳入基准后再执行。

**为什么不选其他**
- 不选 A：发现能力缺口却什么也不做，不利于团队绩效。
- 不选 C：表述不如“动用管理储备”贴合教材对应关系；管理储备本就由组织/管理层掌控。
- 不选 D：压缩其他培训不是处理计划外需求的标准做法。

**出处**
PMBOK® Guide：管理储备 vs 应急储备；建设团队中的培训。


## 008. 项目已经启动，刚刚进入计划编制阶段。在计划编制阶段的早期，项目经理通常应该采用什么领导风格？（）

**选项**
- A. 命令式
- B. 教练型
- C. 支持型
- D. 授权型

**正确答案：A**

**为什么选这个**
计划编制早期，目标、角色与工作方式尚不清晰，项目经理通常采用**命令式**领导，明确指出做什么、怎么做。随团队成熟度提高，再逐步转向教练型、支持型、授权型。

**为什么不选其他**
- 不选 B：教练型适合团队有一定能力但仍需指导时。
- 不选 C：支持型适合能力较强、需要激励与参与时。
- 不选 D：授权型适合高度成熟、可自我管理的团队，偏项目后期。

**出处**
情境领导 / PMBOK 领导风格随项目生命周期变化。


## 009. 项目的早期，应该让哪些人参与项目的规划和决策？（）

**选项**
- A. 少数技术专家
- B. 项目经理
- C. 核心团队成员
- D. 所有项目团队成员

**正确答案：C**

**为什么选这个**
项目早期应让**核心团队成员**参与规划与决策，有利于尽早达成共识、提高计划可执行性，并为后续团队建设打基础。

**为什么不选其他**
- 不选 A：仅少数技术专家视角偏窄，缺少整体规划参与。
- 不选 B：不应仅由项目经理单独规划和决策。
- 不选 D：早期往往尚未组建全部团队，全体参与不现实。

**出处**
PMBOK® Guide 第6版：项目早期核心团队参与规划。


## 010. 关于认可与奖励，下列说法错误的是（）。

**选项**
- A. 在决定认可与奖励时，应考虑文化差异
- B. 认可与奖励应该在项目结束之后做
- C. 只有满足被奖励者的某个重要需求的奖励，才是有效的奖励
- D. 除了金钱奖励，还存在各种有效的无形奖励

**正确答案：B**

**为什么选这个**
认可与奖励应在项目全过程中**及时**开展，才能持续激励；拖到项目结束后再做，激励作用会明显减弱。因此 B 是错误说法。

**为什么不选其他**
- 不选 A：不同文化对奖励方式敏感度不同，应考虑文化差异。
- 不选 C：有效奖励需对接被奖励者的重要需求。
- 不选 D：除金钱外，表扬、机会、荣誉等无形奖励也很有效。

**出处**
PMBOK® Guide 第6版 9.4 建设团队—认可与奖励。


## 011. 在敏捷型项目中，高效的自组织团队是（）。

**选项**
- A. 由通用的专才构成
- B. 由主题专家构成
- C. 由团队成员自发形成
- D. 由分布在不同地方的人构成

**正确答案：A**

**为什么选这个**
敏捷高效自组织团队通常由**通用型专才**（一专多能、T 型）构成，既能深耕某一领域，又能跨职能协作。

**为什么不选其他**
- 不选 B：仅主题专家（窄专才）不利于跨职能协作与负载互补。
- 不选 C：自组织指工作方式上自主决策与协作，不等于成员自发凑成团队。
- 不选 D：分布式不是高效自组织团队的必要条件。

**出处**
Agile Practice Guide / PMBOK：自组织团队与通用型专才。


## 012. 项目即将结束，为了减轻团队成员的焦虑情绪，让团队成员向新项目平滑过渡，项目经理应该（）。

**选项**
- A. 按资源管理计划执行
- B. 提前遣散团队成员
- C. 做好人力资源风险控制
- D. 对团队成员进行培训

**正确答案：A**

**为什么选这个**
资源管理计划通常包含人员释放/遣散安排。临近收尾时按计划执行，让成员清楚何时、如何离开并过渡到新工作，可减轻焦虑并实现平滑过渡。

**为什么不选其他**
- 不选 B：提前遣散可能增加不确定与焦虑，也不一定符合计划。
- 不选 C：风险控制表述过宽，不是本题针对过渡焦虑的直接做法。
- 不选 D：培训不是解决收尾过渡焦虑的主要手段。

**出处**
PMBOK® Guide 第6版：资源管理计划—人员遣散计划。


## 013. 人员的预分派不适用于（）。

**选项**
- A. 在投标文件中所指定的人员
- B. 具有特定的知识和技能的人员，项目因他们而存在
- C. 项目章程中指定的项目经理
- D. 根据雇佣合同就位的优秀专业人员

**正确答案：D**

**为什么选这个**
预分派指资源在获取前已确定。常见情形：竞争性提案/投标已承诺人员、项目依赖特定专家、章程等文件已指定人员（含项目经理）。
“因雇佣合同就位的优秀专业人员”只说明其在组织任职，并不等于已预分派到本项目。

**为什么不选其他**
- 不选 A/B/C：这些都是教材列出的预分派适用情形。

**出处**
PMBOK® Guide 第6版 9.3.2.1 预分派。


## 014. 为了获得项目所需的人力资源，项目经理经常要与以下各方谈判，除了（）。

**选项**
- A. 高级管理层
- B. 职能部门经理
- C. 其他项目经理
- D. 外部资源供应商

**正确答案：D**

**为什么选这个**
获取项目人力资源时，常见谈判对象是高级管理层、职能经理、其他项目的项目经理（争夺稀缺人员）。
与**外部资源供应商**的谈判通常属于**采购管理**，不是本题所说的内部人力资源获取谈判。

**为什么不选其他**
- 不选 A/B/C：这些都是获取人力资源时常见的谈判对象。

**出处**
PMBOK® Guide 第6版 9.3 获取资源—谈判；与采购管理区分。


## 015. 采取虚拟团队的方式开展项目工作，其优势是（）。

**选项**
- A. 团队成员之间更容易分享知识和经验
- B. 更容易跟进项目进度
- C. 可有效利用各个地区专家的力量
- D. 在家办公的团队成员容易产生孤立感

**正确答案：C**

**为什么选这个**
虚拟团队可突破地域限制，把各地专家组织到同一项目中，这是其主要优势之一。

**为什么不选其他**
- 不选 A/B：知识分享与进度跟进通常比集中办公更难，不是优势。
- 不选 D：孤立感是虚拟团队的**劣势**，不是优势。

**出处**
PMBOK® Guide 第6版 9.3.2.4 虚拟团队。


## 016. 完成一个大型复杂项目后，项目经理向组织争取到原来的团队成员继续开展下一个新项目。在新项目上，项目团队建设将（）。

**选项**
- A. 按五个阶段依次进行
- B. 停滞在规范阶段
- C. 跳过震荡阶段
- D. 退回到解散阶段

**正确答案：A**

**为什么选这个**
即便原班人马进入新项目，团队建设仍应按塔克曼五阶段依次进行：形成→震荡→规范→成熟→解散。可能进展更快，但不能省略阶段。

**为什么不选其他**
- 不选 B：不会停滞在规范阶段。
- 不选 C：不能跳过震荡阶段；新目标、新角色仍可能产生冲突。
- 不选 D：解散是上一项目收尾阶段，新项目不是退回解散。

**出处**
PMBOK® Guide 第6版 9.4：塔克曼团队发展模型。


## 017. 在项目环境中，冲突不可避免，冲突的来源有很多。最常见的冲突来源从最常见到最不常见的排序是（）。

**选项**
- A. 成员个性、资源稀缺、进度优先级排序
- B. 进度优先级排序、资源稀缺、成员个性
- C. 资源稀缺、进度优先级排序、成员个性
- D. 资源稀缺、进度优先级排序、个人工作风格

**正确答案：B**

**为什么选这个**
项目中最常见的冲突来源通常按频率排序为：**进度优先级 → 资源稀缺 → 成员个性**。个性冲突常被误以为最多，实际相对较少。

**为什么不选其他**
- 不选 A：把成员个性放在最前，顺序反了。
- 不选 C：资源稀缺不是最常见的第一位。
- 不选 D：虽然“个人工作风格”是教材用语，但把资源稀缺放第一不符合“从最常见到最不常见”的经典排序。

**出处**
PMBOK® Guide 第6版 9.5 管理团队—冲突管理；常见冲突来源排序。


## 018. 可以用以下哪种格式来记录团队成员的详细职责？（）

**选项**
- A. 组织分解结构
- B. 责任分配矩阵
- C. RACI矩阵
- D. 文本型

**正确答案：D**

**为什么选这个**
记录**详细职责**应使用**文本型**格式（如岗位说明书、角色描述），可详细写明职责、职权、能力与资格要求。

**为什么不选其他**
- 不选 A：OBS 把组织单元与工作包对应，不写详细职责。
- 不选 B/C：RAM/RACI 展示“谁与哪项工作相关”，是概要对应关系，不是详细职责文本。

**出处**
PMBOK® Guide 第6版 9.1.2.2：数据表现—文本型格式。


## 019. 在弱矩阵中，项目经理对团队成员只有很小的命令职权，使用哪种技能对项目成功非常关键？（）

**选项**
- A. 领导力
- B. 决策能力
- C. 冲突管理能力
- D. 影响力

**正确答案：D**

**为什么选这个**
弱矩阵中项目经理正式职权很小，团队成员主要听命于职能经理。此时更关键的是用**影响力**进行协商、说服与资源争取，推动工作完成。

**为什么不选其他**
- 不选 A/B/C：这些能力也重要，但针对“几乎没有命令职权”这一情景，教材强调的关键技能是影响力。

**出处**
PMBOK® Guide：矩阵型组织中的项目经理职权与影响力。


## 020. 为了成功激励一个由高学历、高素质人才组成的科技团队，项目管理者可以用下面哪些激励理论？（）

**选项**
- A. 期望理论和X理论
- B. X理论和马斯洛需求层次理论
- C. Y理论和赫兹伯格的激励因素理论
- D. 赫兹伯格的保健因素理论，以及期望理论

**正确答案：C**

**为什么选这个**
对高学历、高素质科技团队，适合：
- **Y 理论**：相信成员能自我激励、愿承担责任；
- **赫兹伯格激励因素**：成就、认可、工作本身、责任、成长等，真正激发积极性。

**为什么不选其他**
- 不选 A/B：含 **X 理论**（认为人厌恶工作、需严管），不适合高素质团队。
- 不选 D：保健因素（薪酬、环境等）只能消除不满，难以真正激励；不是本题最佳组合。

**出处**
麦格雷戈 X/Y 理论；赫兹伯格双因素理论。


## 021. 在矩阵式管理之下，项目经理应该如何取得项目所需的人力资源？（）

## 055. 项目经理通过授权可以()。

**选项**
- A. 把最终责任转移给下级，减轻自己的责任
- B. 把整合管理的工作授权出去
- C. 把执行责任转移给下级
- D. 把自己不想做的工作授权给下级

**正确答案：C**

**为什么选这个**
授权可以把**执行责任**交给下级去做；项目经理仍对结果负责，最终责任不能转嫁。

**为什么不选其他**
- 不选 A：最终责任仍在项目经理，不能通过授权减轻自己的最终责任。
- 不选 B：整合管理是项目经理核心职责，不能授权出去。
- 不选 D：授权应基于能力、发展与工作需要，不是甩自己不想做的事。

**出处**
项目管理授权原则：可授权执行，不可推卸最终责任；整合管理不可授权。


## 056. 为了真正解决冲突，冲突当事人决定把问题摆到桌面上，把不同的方案综合起来，最终得到双方都愿意接受的方案。这是采用了哪种冲突解决方法?()

- B. 妥协
- C. 强制
- D. 撤退 / 回避

- 不选 C：强制是推行一方立场，他方受损。
- 不选 D：撤退/回避是推迟或不处理冲突。
