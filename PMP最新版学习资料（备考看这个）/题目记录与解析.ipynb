# PMP 题目记录与解析

每道题包含：**题目、选项、正确答案、为什么选这个**。后续题目按顺序追加。

## 第一章：引论

### 001.《项目管理知识体系指南》（《指南》）作为一部公认的项目管理标准，规定了在大多数时间适用于大多数项目的知识和做法。这些知识和做法？（）

**选项：**

- A. 可供各种项目裁剪使用
- B. 是具有强制性的项目管理规范
- C. 可以不加修改地用于任何行业的任何项目
- D. 是对项目管理知识体系的完整描述

**正确答案：A. 可供各种项目裁剪使用**

**为什么选这个：**

《PMBOK®指南》归纳的是「大多数时候适用于大多数项目」的良好做法。每个项目的环境、规模、行业都不同，这些知识和做法必须按项目情况**裁剪（tailoring）**后使用。

- 不选 B：指南不是强制性规范，只是良好实践的参考标准。
- 不选 C：不能「不加修改」地用于任何行业、任何项目，必须裁剪。
- 不选 D：指南只是项目管理知识体系中「公认良好做法」的一部分，不是完整描述。

**记忆要点：** 指南 = 公认良好实践 + 需要裁剪。

### 002.《指南》是关于以下哪个的标准？（）

**选项：**

- A. 单个项目的管理
- B. 项目集管理
- C. 项目组合管理
- D. 项目管理工作者的专业行为规范

**正确答案：A. 单个项目的管理**

**为什么选这个：**

《PMBOK®指南》的适用范围是**单个项目**的管理，不是项目集、项目组合，也不是职业道德规范。

- 不选 B：项目集管理另有《项目集管理标准》。
- 不选 C：项目组合管理另有《项目组合管理标准》。
- 不选 D：专业行为规范对应 PMI《道德与专业行为准则》。

**记忆要点：** 指南管「一个项目」；项目集、项目组合、职业道德各有各自的标准/准则。

### 003. 以下哪个是项目的特点？（）

**选项：**

- A. 必须为组织实现利润
- B. 通常会产出相同的产品
- C. 推动组织从当前状态转变到将来状态
- D. 项目是需要持续开展的重复性工作

**正确答案：C. 推动组织从当前状态转变到将来状态**

**为什么选这个：**

项目是为创造独特产品、服务或成果而进行的临时性努力。它的本质是**推动变革**：把组织从当前状态带到希望的将来状态。

- 不选 A：项目不一定要盈利。政府、非营利、合规类项目也可以创造价值。
- 不选 B：项目产出的是独特产品；重复产出相同产品的是运营。
- 不选 D：项目是临时的，不是持续开展的重复性工作；那是运营的特点。

**记忆要点：** 项目 = 临时 + 独特 + 推动从当前状态转到将来状态；运营 = 持续 + 重复。

### 004. 通用项目生命周期包括以下哪些阶段？（）

**选项：**

- A. 开始项目、组织与准备、执行项目工作、结束项目
- B. 开始项目、规划项目、执行项目、监控项目、结束项目
- C. 开始项目、执行项目、监控项目、结束项目
- D. 开始项目、组织与准备、执行项目工作、关闭采购合同

**正确答案：A. 开始项目、组织与准备、执行项目工作、结束项目**

**为什么选这个：**

通用项目生命周期只有四个阶段：开始项目 → 组织与准备 → 执行项目工作 → 结束项目。

- 不选 B：这是五大**过程组**（启动、规划、执行、监控、收尾），不是生命周期。
- 不选 C：缺少「组织与准备」，且混进了过程组里的监控。
- 不选 D：「关闭采购合同」是采购知识领域里的具体过程，不是生命周期阶段。

**记忆要点：** 生命周期 4 阶段 ≠ 过程组 5 个。

### 005. 以下哪一项最能体现项目的特征？（）

**选项：**

- A. 需要制订进度计划
- B. 需要整合范围、进度和成本
- C. 有确定的期限
- D. 需要由专门的团队来实施

**正确答案：C. 有确定的期限**

**为什么选这个：**

项目的两大本质特征是**临时性**和**独特性**。本题里最能体现项目特征的是「有确定的期限」，也就是临时性：项目有明确的开始和结束。

- 不选 A：运营也需要制订进度计划。
- 不选 B：日常运营同样要整合范围、进度和成本。
- 不选 D：运营也可以由专门团队实施。

**记忆要点：** 项目 = 临时（有期限）+ 独特；进度计划、整合、专班都不是项目独有的。

### 006. 在生命周期的早期阶段就可以确定项目的范围、时间和成本，适合采用以下哪种模式的开发生命周期？（）

**选项：**

- A. 预测型生命周期
- B. 迭代型生命周期
- C. 增量型生命周期
- D. 适应型生命周期

**正确答案：A. 预测型生命周期**

**为什么选这个：**

预测型（瀑布/计划驱动）的前提是：项目早期就能把范围、进度、成本基本确定下来，然后按计划推进。本题说「早期就可以确定范围、时间和成本」，正好对应预测型。

- 不选 B：迭代型适合需求会反复细化、通过多轮循环逐步完善产品。
- 不选 C：增量型是分批交付可用功能，范围通常不会在一开始就全部锁死。
- 不选 D：适应型（敏捷）适合需求多变、早期无法确定详细范围/时间/成本。

**记忆要点：** 早期能定死范围、时间、成本 → 预测型；需求不清或多变 → 适应型/迭代/增量。

### 007. 以下哪项是启动项目的最可能的理由？（）

**选项：**

- A. 接到客户对标准化产品的大订单
- B. 需要不停地生产出相同的产品
- C. 现有生产线达不到新环保法律的要求
- D. 在标准化的生产线上开展生产

**正确答案：C. 现有生产线达不到新环保法律的要求**

**为什么选这个：**

项目常因法规/合规、市场需求、技术升级等原因而启动。现有生产线达不到新环保法律，必须做一次性改造才能合规，这是典型的项目启动理由。

- 不选 A：标准化产品的大订单通常由现有运营产能消化，不是新项目。
- 不选 B：不停生产相同产品是运营。
- 不选 D：标准化生产线上的生产也是运营。

**记忆要点：** 启动项目的理由往往是「要改变现状」（合规、新产品、新能力）；重复生产标准化产品是运营。

### 008. 敏捷或适应型项目管理方法适用于以下哪类项目？（）

**选项：**

- A. 需求明确，且产品可以整体交付的项目
- B. 需求不明确，且产品可一部分一部分交付的项目
- C. 涉及组织战略的项目
- D. IT企业的项目

**正确答案：B. 需求不明确，且产品可一部分一部分交付的项目**

**为什么选这个：**

敏捷/适应型方法的适用条件是：需求不确定或会变化，同时产品可以增量交付、尽早获得反馈。

- 不选 A：需求明确且适合整体交付，用预测型更合适。
- 不选 C：是否涉及组织战略，不决定用敏捷还是瀑布。
- 不选 D：行业不是选择生命周期的依据，IT 项目也可以用预测型。

**记忆要点：** 需求不清 + 可分批交付 → 敏捷；需求清楚 + 一次交付 → 预测型。选方法看项目特征，不看是不是 IT、是不是战略。

### 009. 以下哪项是对项目管理过程组和项目生命周期之间的关系的正确描述？（）

**选项：**

- A. 每个项目管理过程组都可以在项目生命周期的任一阶段开展
- B. 各项目管理过程组与项目生命周期的各阶段是一对一的对应关系
- C. 启动过程组只能在“开始项目”阶段开展
- D. 收尾过程组只能在“结束项目”阶段开展

**正确答案：A. 每个项目管理过程组都可以在项目生命周期的任一阶段开展**

**为什么选这个：**

过程组和生命周期阶段不是一回事。生命周期按阶段划分项目；五大过程组是管理工作的逻辑分组。每个阶段内部都可以走一遍启动、规划、执行、监控、收尾。

- 不选 B：过程组与阶段不是一对一。常见错觉是把启动对应开始、规划对应准备、执行对应实施、收尾对应结束。
- 不选 C：每个阶段开始时都可以做启动（例如阶段章程）。
- 不选 D：每个阶段结束时都可以做收尾，不是只有整个项目结束时才收尾。

**记忆要点：** 过程组 ≠ 阶段；每个阶段都可以跑五个过程组。

### 010. 关于项目组合、项目集和项目，以下哪个说法是正确的？（）

**选项：**

- A. 项目集管理的重点在于以“正确”的方式开展项目集
- B. 项目管理的重点在于以“正确”的方式开展项目
- C. 项目组合管理的重点在于开展“正确”的项目集和项目
- D. 以上都正确

**正确答案：D. 以上都正确**

**为什么选这个：**

三层的侧重点不同，但选项表述都对：

- **项目组合：** 做正确的事。按战略挑选、排序资源，开展“正确”的项目集和项目。
- **项目集 / 项目：** 把事情做正确。重点是以“正确”的方式开展工作，交付成果和收益。

因此 A、B、C 都成立，选 D。

**记忆要点：** 组合 = 做正确的事；项目集/项目 = 把事做正确。

### 011. 在适应型生命周期的项目中，何时可以确定每个迭代期的任务？（）

**选项：**

- A. 项目开始之前
- B. 当产品负责人提出新需求后
- C. 回顾性审查完成之后
- D. 每个迭代开始前

**正确答案：D. 每个迭代开始前**

**为什么选这个：**

适应型生命周期里，详细范围和本轮任务是在**每个迭代开始时**（如迭代规划）确定并批准的，而不是项目一开始就全部排好。

- 不选 A：项目开始前就定死各迭代任务，这是预测型做法。
- 不选 B：产品负责人可以随时提新需求，但需求进待办列表不等于本轮任务已确定；要等到迭代规划才锁定本轮工作。
- 不选 C：回顾性审查（retrospective）看的是过程改进，不是确定下一轮任务。

**记忆要点：** 适应型 = 每轮开始时再定本轮任务；预测型 = 项目开始前尽量定全。

### 012. 以下哪个说法是错误的？（）

**选项：**

- A. 在项目开始时，项目资源被转移到运营中；而随着项目趋于结束，运营资源被转移到项目中
- B. 项目和运营都服务于既定的组织战略
- C. 项目与运营会在产品生命周期的不同时点交叉
- D. 在每个交叉点，可交付成果及知识在项目与运营之间转移，以完成工作交接

**正确答案：A**

**为什么选这个：**

本题要找**错误**表述。A 把资源和成果的转移方向说反了。正确关系是：

- 项目**开始**时：运营侧抽调人员/资源进入项目。
- 项目**结束**时：项目把可交付成果、知识和资源交回运营。

B、C、D 都对：项目和运营都服务组织战略；两者会在产品生命周期的多个时点交叉；交叉点上要做成果与知识移交。

**记忆要点：** 开始 = 运营支援项目；结束 = 项目移交运营。

### 013. 在组织战略发生变化之后，以下哪项也要发生变化？（）

**选项：**

- A. 项目范围
- B. 项目集范围
- C. 项目组合范围
- D. 子项目范围

**正确答案：C. 项目组合范围**

**为什么选这个：**

项目组合的职责就是按组织战略挑选、排序和调整项目集/项目。战略变了，组合必须跟着变：哪些做、哪些停、资源怎么重新分配。

单个项目、项目集、子项目的范围不一定都要改——有的继续，有的取消，有的才调整范围。

- 不选 A、B、D：它们不与战略一一对应，战略变化不会强制每一个都改范围。

**记忆要点：** 战略变 → 先变组合；项目/项目集是否改范围要再评估。

### 014. 关于项目阶段的说法，以下哪项是正确的？（）

**选项：**

- A. 阶段审查是项目阶段的一个关键组成部分
- B. 项目阶段是项目管理过程组的另一个名称
- C. 多阶段管理项目的方式会使项目管理工作陷于混乱
- D. 各种项目的项目阶段划分通常是一样的

**正确答案：A. 阶段审查是项目阶段的一个关键组成部分**

**为什么选这个：**

每个阶段结束通常要做**阶段审查**（阶段关口、杀点），检查成果、决定继续、调整还是终止。这是阶段管理的关键环节。

- 不选 B：项目阶段 ≠ 过程组，两者不是同一概念的两个名字。
- 不选 C：分阶段管理是常见做法，用来控制风险、便于治理，不是制造混乱。
- 不选 D：不同行业、不同项目的阶段划分可以不同，没有统一模板。

**记忆要点：** 阶段要有阶段审查；阶段不是过程组；阶段划分因项目而异。

### 015. 每个组织都要为实现目标而开展工作。组织的工作可分为项目和运营两大类。项目与运营会在产品生命周期的以下时间相互交叉，除了（）。

**选项：**

- A. 在项目执行阶段
- B. 在项目收尾阶段
- C. 在产品更新改造时
- D. 在提高运营水平时

**正确答案：A. 在项目执行阶段**

**为什么选这个：**

本题是「除了」，要找**不是**交叉点的选项。项目与运营的典型交叉时点是：

- 项目**收尾**时，把成果和知识移交运营；
- **产品更新改造**时，运营提出需求、项目来实施；
- **提高运营水平**时，用项目去改进运营过程。

项目**执行阶段**主要是项目团队在做项目工作，不是教材列出的交叉点。

**记忆要点：** 交叉点 ≈ 收尾移交、改产品、改运营；不是执行当中。

### 016. 在管理一个项目时，需要平衡哪些相互竞争的制约因素？（）

**选项：**

- A. 质量和范围
- B. 资源和风险
- C. 进度和预算
- D. 以上都是

**正确答案：D. 以上都是**

**为什么选这个：**

项目管理要平衡一组相互竞争的制约因素，常见包括：**范围、质量、进度、预算、资源、风险**。改其中一个，通常会影响其他几个。A、B、C 都是其中的组合，所以选 D。

**记忆要点：** 制约因素不只有铁三角（范围/进度/成本），质量和资源、风险也要一起平衡。

### 017. 什么是混合型项目生命周期？（）

**选项：**

- A. 预测型生命周期和适应型生命周期的组合
- B. 迭代型生命周期和增量型生命周期的组合
- C. 产品生命周期和项目管理过程组的组合
- D. 项目生命周期和产品生命周期的组合

**正确答案：A. 预测型生命周期和适应型生命周期的组合**

**为什么选这个：**

混合型生命周期 = **预测型 + 适应型**。常见做法是：需求清楚、变更少的部分用预测型；需求会变、适合分批交付的部分用敏捷/适应型。

- 不选 B：迭代和增量是适应型内部的做法，两者组合不等于混合型的定义。
- 不选 C：产品生命周期和过程组不是开发方法的组合。
- 不选 D：项目生命周期套在产品生命周期里，这是包含关系，不是混合型。

**记忆要点：** 混合型 = 预测型 + 适应型（瀑布 + 敏捷）。

### 018. 为了获得更大的效益，而对一系列相互配套的项目进行协调管理，这属于哪一种管理？（）

**选项：**

- A. 项目组合管理
- B. 多项目管理
- C. 项目集管理
- D. 项目管理办公室应该开展的管理

**正确答案：C. 项目集管理**

**为什么选这个：**

项目集管理的定义就是：对一组**相互关联/配套**的项目进行协调管理，以获得单独管理这些项目时无法获得的收益。

- 不选 A：项目组合按战略把项目/项目集放在一起选优先级，它们不必相互配套。
- 不选 B：「多项目管理」不是 PMI 标准层级，只是同时管多个项目。
- 不选 D：PMO 提供支持、标准和治理，本身不等于这种协调管理。

**记忆要点：** 相互配套 + 更大收益 → 项目集；战略取舍、不一定相关 → 项目组合。

### 019. 下列哪项不能成为项目结束的理由？（）

**选项：**

- A. 项目目标按计划达成
- B. 项目不能达到目标
- C. 某个关键相关方不支持项目
- D. 项目需求不复存在

**正确答案：C. 某个关键相关方不支持项目**

**为什么选这个：**

本题要找**不能**作为结束理由的选项。项目结束的常见理由包括：目标已达成、目标无法达成、需求不再存在、资金耗尽等。

某个关键相关方不支持，属于相关方管理问题，项目经理应沟通、影响、升级，而不是据此直接结项。

- A 对：目标达成是正常收尾。
- B 对：确认无法达成目标可以提前终止。
- D 对：需求没了，项目没有继续的必要。

**记忆要点：** 结束看目标/需求还在不在；相关方不支持 → 去管理，不是直接关门。

### 020. 某企业对生产线的日常维护工作通常被看作是（）。

- A. 项目
- B. 项目集
- C. 项目组合
- D. 运营

**正确答案：D. 运营**

**为什么选这个：**

日常维护是持续开展、重复进行的工作，目的是维持现有生产线运转，符合**运营**的特点。

- 不选 A：项目是临时、独特的。若是一次性大修或为达标而改造生产线，才更像项目。
- 不选 B、C：项目集、项目组合是对一组项目的管理层级，不是这类日常工作。

### 022. 为了确保项目得到执行组织最大程度的支持，项目经理应该（）。

- C. 把项目需求与高级管理者的个人需求联系在一起
- D. 在项目计划中规定将如何进行团队建设

**为什么选这个：**

执行组织按战略分配资源和注意力。项目需求与组织战略对齐，才最容易获得最大支持。

- 不选 A：沟通计划有用，但不是获得组织级支持的根本原因。
- 不选 C：把项目绑到高管个人需求上，既不专业也不符合组织利益。

- B. 项目开始前就做详细周密的计划
- C. 用于管理的时间比项目执行的时间更多
- D. 强调不断对任务进行优先级排序

**正确答案：D. 强调不断对任务进行优先级排序**

**为什么选这个：**

适应型项目需求会变，必须持续按价值对待办任务重新排序（如产品待办列表），先做最重要的。

**为什么不选其他：**

- 不选 A：「只允许」太绝对。适应型可以持续调整；迭代内工作相对稳定，但不是只能在迭代结束后才改计划。
- 不选 B：开始前就做详细周密计划，这是预测型。适应型是滚动/渐进明细。
- 不选 C：适应型强调把时间花在交付可用产品上，不是管理时间多于执行时间。

**记忆要点：** 适应型 = 持续排优先级 + 滚动计划；详细计划前置 = 预测型。

- 不选 B：满足客户/市场需求、能产生收益，运营也能做到，没有临时性和独特性。

### 026. 以下哪个关于项目组合的说法是正确的？（）

**选项：**

- A. 只要你愿意，可以把任何项目放在一个项目组合中
- B. 任何一个项目都必须至少隶属于一个项目组合
- C. 各项目因服务于同一个战略目标而被放入同一个项目组合
- D. 既然项目是临时的，项目组合就必然是临时的

**正确答案：C. 各项目因服务于同一个战略目标而被放入同一个项目组合**

**为什么选这个：**

项目组合是为了实现组织战略目标，把项目、项目集、子组合和相关运营放在一起管理。放进同一个组合的原因是它们服务于同一（类）战略目标，不是因为项目之间必须相互配套。

**为什么不选其他：**

- 不选 A：不能凭个人意愿随便塞。组合成员要按战略来选，不是想放就放。
- 不选 B：并非每个项目都必须属于某个组合，组织里可以有未纳入组合的项目。
- 不选 D：项目是临时的，组合通常随组织战略持续存在，项目进进出出，组合不必是临时的。

**记忆要点：** 组合按战略归堆；项目集才要求相互配套。组合持续，项目临时。

### 027. 以下哪个关于开发生命周期的说法是正确的？（）

**选项：**

- A. 开发生命周期等同于项目生命周期
- B. 在一个项目生命周期中可以有多个开发生命周期
- C. 开发生命周期等同于产品生命周期
- D. 开发生命周期通常长于项目生命周期

**正确答案：B. 在一个项目生命周期中可以有多个开发生命周期**

**为什么选这个：**

开发生命周期指的是项目中与产品、服务或成果**开发**相关的一个或多个阶段（预测、迭代、增量、适应、混合）。一个项目生命周期里可以包含多个开发生命周期，例如先用预测型做硬件，再用适应型做软件。

**为什么不选其他：**

- 不选 A：项目生命周期覆盖从开始到结束的全部阶段；开发生命周期只是其中与开发有关的部分，两者不等同。
- 不选 C：产品生命周期从概念到退市，比单个项目长得多，也不等同于开发生命周期。
- 不选 D：开发生命周期通常嵌在项目之内，不会比项目生命周期更长。更长的是产品生命周期。

**记忆要点：** 产品生命周期 ⊃ 项目生命周期 ⊃ 开发生命周期（可以有多个）。

### 028. 如果项目之间的联系仅限于共享顾主、供应商、技术或资源，则这些项目应该（）。

- A. 由项目管理办公室统一管理
- B. 被作为项目组合来协调管理
- C. 被作为项目集来协调管理
- D. 被归入项目作战室统一管理

**正确答案：B. 被作为项目组合来协调管理**

**为什么选这个：**

项目之间如果只是共用客户、供应商、技术或资源，并没有为实现共同收益而相互配套，就应按**项目组合**来协调（统一排优先级、分资源），而不是项目集。

### 029. 以下哪种方法能使组织资源得到最大利用以支持组织战略的实现？（）

**选项：**

- A. PMO统一管理
- B. 项目组合管理
- C. 项目集管理
- D. 项目管理

**正确答案：B. 项目组合管理**

**为什么选这个：**

项目组合管理按组织战略选择、排序和调整项目/项目集，并把有限资源优先配给最能实现战略的工作。这正是「资源最大利用、支持战略」的方法。

**为什么不选其他：**

- 不选 A：PMO 提供支持、标准和治理，不负责在全组织层面按战略优化全部资源投向。
- 不选 C：项目集是协调配套项目拿更大收益，范围小于整个组织的战略资源配置。
- 不选 D：项目管理关注把单个项目做正确，不负责组织级资源最优。

**记忆要点：** 战略 + 资源最大利用 → 项目组合。

### 030. 以下关于项目管理过程的哪个说法是正确的？（）

**选项：**

- A. 运用工具与技术把输入转化成输出
- B. 由特定的人把输入转化成输出
- C. 在特定的时间把输入转化成输出
- D. 按特定的顺序把输入转化成输出

**正确答案：A. 运用工具与技术把输入转化成输出**

**为什么选这个：**

项目管理过程的基本结构是 **ITTO**：输入（Input）经工具与技术（Tools & Techniques）转化为输出（Output）。这就是过程的定义。

**为什么不选其他：**

- 不选 B：谁来做可以裁剪，过程不绑定特定的人。
- 不选 C：开展时点因过程而异，不是固定在某个特定时间。
- 不选 D：过程可以重叠、反复、按项目裁剪，不是必须按某个固定顺序。

**记忆要点：** 过程 = 输入 + 工具技术 → 输出。

### 031. 项目管理的五大过程组是（）。

**选项：**

- A. 开始、组织、准备、执行、结束
- B. 前期准备、规划、执行、监控、收尾
- C. 启动、规划、执行、监控、收尾
- D. 商业论证、启动、规划、执行、收尾

**正确答案：C. 启动、规划、执行、监控、收尾**

**为什么选这个：**

五大过程组是：启动、规划、执行、监控（监督）、收尾。这是管理工作的逻辑分组。

**为什么不选其他：**

- 不选 A：这是在套通用项目生命周期（开始、组织与准备、执行、结束），不是过程组，而且漏了监控。
- 不选 B：「前期准备」不是过程组名称。
- 不选 D：商业论证是启动前的商业文件，不是过程组；这里还漏了监控。

### 034. 项目集管理是通过以下哪个具体管理措施来关注项目间的依赖关系的？（）

**选项：**

- A. 审查项目并确定项目之间的资源分配优先顺序
- B. 特别关注各项目所共享的资源
- C. 处理同一个治理结构内的相关问题和变更管理
- D. 及时剔除不合理的项目，并补充新项目

**正确答案：C. 处理同一个治理结构内的相关问题和变更管理**

**为什么选这个：**

项目集管理关注项目之间的依赖关系，具体做法包括：在**同一个治理结构**内处理相关问题与变更，协调资源冲突，对齐战略方向，以便实现单独管理得不到的收益。C 是教材里的原句级表述。

- 不选 D：剔除不合理项目、补充新项目，是组合按战略调整成分，不是项目集管依赖的做法。

### 035. 为了使项目符合组织的战略业务目标，而对项目组合、项目集和项目进行系统化管理，你将考虑应用（）。

- B. 项目集管理
- C. 项目组合管理
- D. 组织级项目管理

组织级项目管理（OPM）是把**项目组合、项目集、项目**与组织战略系统整合在一起的框架，目的是让项目工作符合战略业务目标。题目同时提到对这三层进行系统化管理，对应的就是 OPM。

- 不选 B：独特的成果多指文件、知识、研究结果等（如研究课题发现的新知识）。举办培训不是这个。
- 不选 C：独特的产品多指有形物或可组件（如新设备、课程教材若作为开发对象）。这里是办课，不是造出一个有形产品。
- 不选 D：项目产出必须独特，不是「相似的结果」。相似、重复是运营。

**记忆要点：** 举办培训/会议 → 服务；开发出新课/新知识 → 成果或产品；造东西 → 产品。

### 037. 之所以需要对项目管理过程进行裁剪，主要是因为（）。

**选项：**

- A. 项目具有临时性
- B. 项目具有目标性
- C. 项目具有独特性
- D. 项目具有渐进明细性

**正确答案：C. 项目具有独特性**

**为什么选这个：**

没有两个项目完全一样。规模、行业、复杂度、组织环境都不同，所以《指南》里的过程、工具必须按具体项目**裁剪**。裁剪的根本原因是独特性。

**为什么不选其他：**

- 不选 A：临时性说明项目有开始和结束，解释不了为什么过程要裁剪。
- 不选 B：有目标是项目的普遍属性，不是裁剪的原因。
- 不选 D：渐进明细是计划由粗到细，属于规划方式，不是「要对过程裁剪」的主因。

**记忆要点：** 独特 → 必须裁剪（接 001 题）。

### 038. 项目完成后的维护和连续运营对项目至关重要，因此应该（）。

**选项：**

- A. 作为项目的一部分
- B. 由项目团队来开展
- C. 在项目执行过程中加以考虑，但不应看作项目的一部分
- D. 作为整个项目中的一个子项目

**正确答案：C. 在项目执行过程中加以考虑，但不应看作项目的一部分**

**为什么选这个：**

项目结束后的维护和连续运营是**运营**工作，不是项目的组成部分。但它们很重要，所以要在项目期间就考虑（可维护性、培训、移交、文档），以便收尾时顺利交接。

**为什么不选其他：**

- 不选 A：维护和连续运营不是项目的一部分，项目有明确结束点。
- 不选 B：通常由运营团队接手，不是项目团队长期去做。
- 不选 D：也不是子项目。子项目仍在项目范围内，而维护运营在项目结束之后。

**记忆要点：** 运营要考虑，但不纳入项目范围（接 012、020 题）。

<p style="color:#e74c3c">### 039. 要建立一个新的通信卫星系统，其中包括卫星与地面站的设计与建造、卫星发射以及系统整合，适合选择以下哪一项来进行管理？（）

**选项：**

- A. 项目
- B. 大型项目
- C. 项目集
- D. 项目组合

**正确答案：C. 项目集**

**为什么选这个：**

卫星与地面站的设计建造、发射、系统整合是一组**相互配套**的组成部分，必须协调管理才能得到「可用的通信卫星系统」这个单独做各块得不到的收益。这是《指南》里项目集的典型例子。

**为什么不选其他：**

- 不选 A：当成一个项目会忽略各组成部分之间必须协调的依赖和整体收益。
- 不选 B：「大型项目」不是 PMI 的标准管理层级。
- 不选 D：组合是按战略归堆、成员不必相互配套。这里各部分必须配套才能形成系统。

**记忆要点：** 配套组件共同形成更大收益 → 项目集（接 018、028 题）。

### 040. 项目效益管理计划应该包括以下哪项内容？（）

**选项：**

- A. 目标效益、项目经理、效益实现时间
- B. 目标效益、战略一致性、效益责任人
- C. 目标效益、业务需求、实现效益的风险
- D. 目标效益、战略一致性、项目可交付成果

**正确答案：B. 目标效益、战略一致性、效益责任人**

**为什么选这个：**

效益管理计划说明效益是什么、何时如何实现、如何测量。常见内容包括：目标效益、战略一致性、效益实现时间、**效益责任人**、测量指标、假设和风险。B 三项都在这个清单里。

**为什么不选其他：**

- 不选 A：效益实现时间对，但效益责任人通常是业务负责人/发起人，不是项目经理。项目结束后效益还要继续追踪。
- 不选 C：业务需求写在商业论证里，不是效益管理计划的核心条目。
- 不选 D：战略一致性对，但可交付成果属于项目产出；效益是使用成果之后得到的好处，两者不是一回事。

**记忆要点：** 效益计划 = 目标效益 + 对齐战略 + 责任人（通常不是项目经理）。

### 041. 以下哪项最好地描述了项目与运营的区别？（）

**选项：**

- A. 项目受制于有限的资源，而用于运营的资源通常不受限制
- B. 项目不会因目标的实现而结束，而运营会因目标的实现而结束
- C. 项目的持续时间很短，而运营的持续时间很长
- D. 项目生产独特的结果，而运营生产重复性结果

**正确答案：D. 项目生产独特的结果，而运营生产重复性结果**

**为什么选这个：**

项目与运营最核心的差别之一是：项目产出**独特**成果，运营产出**重复性**成果。再加上项目临时、运营持续。

**为什么不选其他：**

- 不选 A：运营同样受资源限制，不是无限资源。
- 不选 B：说反了。项目会因目标达成（或无法达成、需求消失）而结束；运营是持续的，不会因某次目标达成就停。
- 不选 C：临时 ≠ 时间短。有的项目可以做很多年，只是有明确结束点。

**记忆要点：** 项目 = 临时 + 独特；运营 = 持续 + 重复。临时不是「短」。

### 042. 在项目的阶段关口，可能做出相关决定，以便（）。

**选项：**

- A. 撤换项目经理
- B. 启动一个新项目
- C. 停留在当前阶段
- D. 表彰项目团队

**正确答案：C. 停留在当前阶段**

**为什么选这个：**

阶段关口（阶段审查、杀点）把阶段绩效和商业文件对比后，可能决定：进入下一阶段、修改后继续、结束项目、**停留在当前阶段**，或重复本阶段部分工作。C 是标准决定之一。

**为什么不选其他：**

- 不选 A：撤换项目经理不是阶段关口的标准决定。
- 不选 B：关口管的是本项目本阶段走不走下去，不是用来启动另一个新项目。
- 不选 D：表彰团队可以做，但不是阶段关口要做的治理决定。

**记忆要点：** 关口决定 = 继续 / 修改后继续 / 终止 / 留下重做（接 014 题）。

### 043. 任何一个项目都肯定（）。

**选项：**

- A. 隶属于某个项目集
- B. 隶属于某个项目组合
- C. 隶属于项目管理办公室
- D. 以上说法都不对

**正确答案：D. 以上说法都不对**

**为什么选这个：**

「任何一个」「肯定」太绝对。项目可以独立存在，不必属于项目集、项目组合或 PMO。

**为什么不选其他：**

- 不选 A：只有相互配套、要拿更大收益时才进项目集。
- 不选 B：组合按战略归堆，不是每个项目都必须纳入（见 026 题）。
- 不选 C：不是每个组织都有 PMO，有 PMO 也不等于每个项目都归它管。

**记忆要点：** 看到「任何/必须/肯定」隶属组合、项目集、PMO，多半不对。

### 044. 某组织实施一个运用精益六西格玛价值流图的项目，这属于以下哪项的示例？（）

**选项：**

- A. 符合法规、法律或社会要素
- B. 满足相关方的要求或需求
- C. 创造、改进或修复产品、过程或服务
- D. 执行、变更业务或技术战略

立项常见驱动有四类：合规、满足相关方需求、改进产品/过程/服务、执行或变更战略。精益六西格玛、价值流图是典型的**过程改进**手段，对应「创造、改进或修复产品、过程或服务」。

- 不选 D：没有说要改变业务或技术战略，只是改进现有过程。

**记忆要点：** 精益 / 六西格玛 / 价值流图 → 改进过程。

### 045. 以下哪个项目管理知识领域有来自全部五大过程组的过程？（）

**选项：**

- A. 项目采购管理
- B. 项目相关方管理
- C. 项目进度管理
- D. 项目整合管理

**正确答案：D. 项目整合管理**

**为什么选这个：**

只有**整合管理**在五大过程组都有过程：启动（制定章程）、规划（制定项目管理计划）、执行（指导与管理项目工作、管理项目知识）、监控（监控项目工作、实施整体变更控制）、收尾（结束项目或阶段）。

**为什么不选其他：**

- 不选 A：采购没有启动过程；收尾也并入监控/项目收尾，不覆盖全部五个过程组。
- 不选 B：相关方管理有启动、规划、执行、监控，没有收尾过程。
- 不选 C：进度管理主要在规划和监控，没有启动、执行、收尾过程。

**记忆要点：** 五大过程组都有过程的知识领域 = 整合。

### 046. 《指南》收录了项目管理知识体系中普遍认为“良好实践的那一部分”。关于普遍认可，下列哪项是正确的？（）

**选项：**

- A. 这些知识和做法适用于所有项目，并且其价值和有效性已获得一致认可
- B. 这些知识和做法适用于大多数项目，并且其价值和有效性已获得大多数人认可
- C. 这些知识和做法适用于所有项目，并且其价值和有效性已获得大多数人认可
- D. 这些知识和做法适用于大多数项目，并且其价值和有效性已获得一致认可

**正确答案：D. 这些知识和做法适用于大多数项目，并且其价值和有效性已获得一致认可**

**为什么选这个：**

「普遍认可」有两层：① 适用于**大多数项目**在大多数时候；② 对其价值和有效性存在**共识（一致认可）**。D 正好是这两层。

**为什么不选其他：**

- 不选 A：不是所有项目。适用大多数，还要裁剪（见 001 题）。
- 不选 B：大多数项目对了，但「大多数人认可」不够。「普遍认可」强调的是共识/一致认可，不是简单多数。
- 不选 C：所有项目错，大多数人认可也弱了。

**记忆要点：** 普遍认可 = 大多数项目 + 一致认可；不是全部项目，也不是简单多数人同意。

### 047. 关于项目成功评价标准的说法，以下哪项是正确的？（）

**选项：**

- A. 只要在规定的范围、时间、成本和质量之下完成，项目就是成功的
- B. 应该由高级管理层确定项目的成功评价标准
- C. 主要相关方和项目经理应就如何评价项目成功达成一致
- D. 应该由项目经理确定项目的成功评价标准

**正确答案：C. 主要相关方和项目经理应就如何评价项目成功达成一致**

**为什么选这个：**

项目成功标准要在项目早期由**项目经理和主要相关方共同商定并记录**。只完成范围、时间、成本、质量不一定算成功，还要看目标、效益和相关方满意度。

**为什么不选其他：**

- 不选 A：铁三角达标只是传统标准，现在还要看商业目标、效益是否实现。按时按预算做完一个没人用的产品，不算成功。
- 不选 B：不是高管单方面确定，要相关方和项目经理一起达成一致。
- 不选 D：也不是项目经理一个人定。

**记忆要点：** 成功标准 = 相关方 + 项目经理事先达成一致；不只看铁三角。

### 048. 以下哪项是应该在项目的前期准备阶段编制的文件？（）

**选项：**

- A. 项目章程
- B. 项目管理计划
- C. 需求文件
- D. 需要评估

**正确答案：D. 需要评估**

**为什么选这个：**

前期准备发生在项目正式启动之前。先做**需要评估**，再据此形成商业论证和效益管理计划，然后才批准章程、正式启动项目。

**为什么不选其他：**

- 不选 A：项目章程是启动过程组的输出，标志着项目正式授权，不是启动前的前期文件。
- 不选 B：项目管理计划是规划过程组的输出。
- 不选 C：需求文件是规划过程（收集需求）的输出。

**记忆要点：** 启动前 = 需要评估 / 商业论证 / 效益管理计划；启动 = 章程；规划 = 管理计划、需求文件。

### 049. 全球项目管理业界定义的最重要的价值是（）。

**选项：**

- A. 责任、尊重、公正和诚实
- B. 责任、尊重、公平和公正
- C. 善良、公平、公开和公正
- D. 善良、公平、公正和诚实

**正确答案：A. 责任、尊重、公正和诚实**

**为什么选这个：**

PMI《道德与专业行为准则》的四大价值是：**责任、尊重、公正、诚实**（Responsibility, Respect, Fairness, Honesty）。

**为什么不选其他：**

- 不选 B：公平和公正对应的是同一个价值（Fairness），这里缺了诚实。
- 不选 C：善良、公开都不是 PMI 四大价值。
- 不选 D：善良不是四大价值之一。

### 050. 项目的商业价值是指（）。

**选项：**

- A. 项目成果能够为相关方带来的有形效益
- B. 项目成果能够为相关方带来的效益
- C. 项目成果能够为相关方带来的无形效益
- D. 项目成果能够为项目所在组织带来的效益

**正确答案：B. 项目成果能够为相关方带来的效益**

**为什么选这个：**

商业价值是项目成果给相关方带来的净效益，可以是有形的（货币、股份、设备），也可以是无形的（品牌、商誉、公益），或两者都有。B 没有把效益类型和组织范围说窄。

**为什么不选其他：**

- 不选 A：只写有形效益，漏了无形效益。
- 不选 C：只写无形效益，漏了有形效益。
- 不选 D：对象不限于项目所在组织，还包括客户等其他相关方。

**记忆要点：** 商业价值 = 给相关方的效益（有形 + 无形）。

<p style="color:#e74c3c">### 051. 工作绩效报告的主要用途是（）。

**选项：**

- A. 实时显示项目的进展情况
- B. 显示项目已经发生的绩效偏差
- C. 显示项目工作安排
- D. 便于制定决策和采取行动

**正确答案：D. 便于制定决策和采取行动**

**为什么选这个：**

工作绩效报告是工作绩效信息的汇编（状态报告、备忘录、仪表盘等），目的是让相关方**了解情况、做出决策、采取行动**。

**为什么不选其他：**

- 不选 A：实时显示进展更接近工作绩效数据/仪表盘原始刷新，不是报告的主要定义。
- 不选 B：显示已发生偏差属于工作绩效信息（分析后的结果），还没到「报告用来决策」这一层。
- 不选 C：工作安排是计划（进度计划等），不是绩效报告。

**记忆要点：** 数据（原始）→ 信息（分析/偏差）→ 报告（决策和行动）。接 032 题。

### 052. 项目的独特性表现在（）。

**选项：**

- A. 独特的产品
- B. 独特的服务
- C. 独特的成果
- D. 独特的可交付成果

**正确答案：D. 独特的可交付成果**

**为什么选这个：**

项目创造独特的产品、服务或成果，这三者都属于**可交付成果**。独特性表现在交付物是独特的，选涵盖最全的 D。

**为什么不选其他：**

- 不选 A：只覆盖产品，漏了服务和成果。
- 不选 B：只覆盖服务（如 036 举办培训）。
- 不选 C：只覆盖成果（如研究得到的新知识）。

**记忆要点：** 可交付成果 = 产品 + 服务 + 成果；项目独特就独特在交付物。

### 054. 适应型项目经常开展回顾性审查，以下哪个是开展回顾性审查的目的？（）

**选项：**

- A. 为团队成员安排下个迭代期的工作
- B. 审查团队是否需要辅导
- C. 向产品负责人（客户）展示产品原型
- D. 布置今天的工作任务

**正确答案：B. 审查团队是否需要辅导**

**为什么选这个：**

回顾性审查（retrospective）看的是过程、协作和团队怎么改进：哪些做得好、哪些要改、下一轮承诺改什么，以及**团队是否需要辅导**。

**为什么不选其他：**

- 不选 A：给下一迭代安排工作是迭代规划（sprint planning），见 011 题。
- 不选 C：向产品负责人/客户展示原型是迭代评审（sprint review / demo）。
- 不选 D：布置当天任务是每日站会。

**记忆要点：** 规划排活、评审给客户看、回顾改过程、站会同步当天。

### 055. 项目的临时性是指（）。

**选项：**

- A. 项目的持续时间短
- B. 项目的重要性低
- C. 项目成果的寿命期短
- D. 项目有明确的开始和结束时间

**正确答案：D. 项目有明确的开始和结束时间**

**为什么选这个：**

临时性的含义是项目有明确的开始和结束，不是一项持续不断的工作。结束可以是目标达成、无法达成或需求消失。

**为什么不选其他：**

- 不选 A：临时 ≠ 短。大项目可以持续很多年。见 041 题。
- 不选 B：临时性和重要性无关。战略项目也可以是临时的。
- 不选 C：项目结束不等于成果寿命短。大桥、系统可以运营几十年。

**记忆要点：** 临时 = 有始有终；不是短、不是不重要、不是成果不长久。

### 043. 任何一个项目都肯定（）。

**选项：**

- A. 隶属于某个项目集
- B. 隶属于某个项目组合
- C. 隶属于项目管理办公室
- D. 以上说法都不对

**正确答案：D. 以上说法都不对**

**为什么选这个：**

项目可以独立存在。不是每个项目都必须进项目集、进组合，或归 PMO 管。A、B、C 都把「可以」说成了「肯定」，所以都不对。

**为什么不选其他：**

- 不选 A：只有相互配套、要协调拿更大收益时才进项目集。很多项目是独立项目。
- 不选 B：组织可以有未纳入组合的项目，并非每个项目都必须属于某个组合（见 026 题）。
- 不选 C：不是每个组织都有 PMO，有 PMO 也不是所有项目都归它管。

**记忆要点：** 项目可以独立；组合/项目集/PMO 都不是必选项。
