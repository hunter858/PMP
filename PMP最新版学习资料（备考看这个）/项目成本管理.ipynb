## 001. 以下哪项是执行组织可以使用的完工估算预测技术？（）

**选项**
- A. EAC=AC+(BAC-EV)/CPI
- B. EAC=AC+BAC-EV
- C. EAC=BAC/CPI
- D. 以上都是

**正确答案：D**

**为什么选这个**
常用 EAC 预测方法都可用：非典型偏差用 B；典型偏差用 A（与 C 数学等价）。

**为什么不选其他**
- 不选只选 A/B/C 之一：三种都是教材认可的预测技术，应选“以上都是”。

**出处**
PMBOK® Guide 第6版 7.4.2.2 挣值分析—完工估算。


<span style="color:red">## 017. 如果净值是300000美元，实际成本是355000美元，计划价值是370000美元，项目的进度绩效指数是（）。

**选项**
- A. 0.85
- B. 1.25
- C. 1.16
- D. 0.81

**正确答案：D. 0.81**

**为什么选这个**
SPI = EV / PV = 300000 / 370000 ≈ 0.81；SPI<1 表示进度落后。

**为什么不选其他**
- 不选 A：那是 CPI=EV/AC=300000/355000≈0.85，成本绩效指数干扰项。
- 不选 B/C：计算错误。

**出处**
PMBOK® Guide 第6版：SPI=EV/PV，CPI=EV/AC。


## 035. 在敏捷型项目中，最适合的资源供应方法通常是（）。

**选项**
- A. 准时制供应
- B. 尽量提前供应
- C. 库存一些资源
- D. 准备一些替代资源

**正确答案：A**

**为什么选这个**
敏捷需求变化快，资源通常按当前迭代需要**准时制（JIT）**获取，避免过早占用或囤积。

**为什么不选其他**
- 不选 B/C：提前供应或囤库存易造成浪费。
- 不选 D：准备替代资源偏风险应对，不是敏捷最常用的资源供应方式。

**出处**
PMBOK® Guide 第6版 / 敏捷环境：资源常准时制获取。


<span style="color:red"> ## 036. 项目经理基于以往类似项目经验，对项目成本开展估算，准确度是-25%～+75%。他正在使用下列哪种成本估算方法？（）

**选项**
- A. 确定性估算
- B. 三点估算
- C. 类比估算
- D. 参数估算

**正确答案：C**

**为什么选这个**
基于以往类似项目经验 → 类比估算；准确度 -25%～+75% 属粗数量级（ROM），常见于早期、信息不足时。

**为什么不选其他**
- 不选 A：确定性估算准确度约 -5%～+10%。
- 不选 B：三点估算用乐观/最可能/悲观三个值。
- 不选 D：参数估算用统计关系/参数模型。

**出处**
PMBOK® Guide 第6版 7.2 估算成本—类比估算与准确度区间。


## 037. 在制定成本管理计划的过程中，项目团队不需要（）。

**选项**
- A. 征求成本管理专家的意见
- B. 考察不同的项目融资方式
- C. 召开成本规划会议
- D. 考虑花多少钱去做质量管理

**正确答案：D**

**为什么选这个**
规划成本管理工具与技术：专家判断、数据分析（含融资备选方案）、会议。

**为什么不选其他**
- 不选 A/B/C：这些属于制定成本管理计划时要做的。
- 不选 D（本题选它）：质量成本是规划质量管理或估算成本时考虑的，不是规划成本管理必须项。

**出处**
PMBOK® Guide 第6版 7.1 规划成本管理。


## 038. 在编制项目预算时，可以把各时点的累计成本基准确定为略高于预计的累计资金支出数，以便（）。

**选项**
- A. 尽量拖延资金支出
- B. 尽早取得资金投入
- C. 更灵活地安排资金的使用
- D. 考虑预计的债务

**正确答案：D**

**为什么选这个**
成本基准按应计成本累计，资金支出按实际付款累计；成本常先入账、钱后付，差额即预计债务（如应付账款）。

**为什么不选其他**
- 不选 A/B：不是拖延支出或尽早投入的目的。
- 不选 C：灵活安排资金不是该做法的标准解释。

**出处**
PMBOK® Guide 第6版 7.3 制定预算—成本基准与资金需求。


## 039. 经过分析，采用非节能材料可降低项目成本，但是会导致项目投入运营后成本非常高。项目团队应该（）。

**选项**
- A. 只关注完成项目成本最低，不考虑项目成果的运营成本
- B. 考虑项目决策对产品运营成本的影响后，再做出相应决策
- C. 请运营部门决策是否使用节能材料
- D. 按设计图纸执行

**正确答案：B**

**为什么选这个**
全生命周期成本：不能只看建设期，还要评估投产后运营成本，再决策。

**为什么不选其他**
- 不选 A：只盯项目成本、不管运营成本，正是要避免的。
- 不选 C：推给运营部门，项目团队仍要对生命周期成本负责。
- 不选 D：回避了决策本身。

**出处**
PMBOK® Guide 第6版：全生命周期成本（Life Cycle Costing）。


## 040. 项目成本控制的重点是（）。

**选项**
- A. 分析资金支出与已完工作是否匹配
- B. 监督资金的支出
- C. 确保成本管理不超出批准的资金限额
- D. 把成本超支控制在可接受范围内

**正确答案：A**

**为什么选这个**
成本控制重点是分析**花出去的钱与已完成工作是否匹配**（挣值思想），而不只是盯支出金额。

**为什么不选其他**
- 不选 B：只监督支出、不对照完成工作，无法判断是否合理。
- 不选 C/D：相关但不是“重点”表述。

**出处**
PMBOK® Guide 第6版 7.4 控制成本。

<span style="color:red">## 042. 良好的成本管理可以确保（）。

## 043. 项目最可能的成本是1000万元，最乐观的成本是800万元，最悲观的成本是1400万元。那么基于贝塔分布的预期成本是（）。

**选项**
- A. 1033万元
- B. 1067万元
- C. 1000万元
- D. 1100万元

**正确答案：A. 1033万元**

**为什么选这个**
贝塔分布（PERT）：E=(O+4M+P)/6=(800+4×1000+1400)/6=6200/6≈1033万元。

**为什么不选其他**
- 不选 B：那是三角分布 (O+M+P)/3≈1067。
- 不选 C/D：计算错误。

**出处**
PMBOK® Guide 第6版：三点估算—贝塔分布与三角分布。

## 045. 下列哪项是资金限制平衡的作用？（）

**选项**
- A. 按已知资金限制调整资金支出计划
- B. 在资源需求与资源供给之间取得平衡
- C. 增加资金投入，满足项目的资金支出需要
- D. 增加投入的资源以达到压缩工期的目的

**正确答案：A**

**为什么选这个**
资金限制平衡：对照已知资金限额，调整支出计划（必要时调整进度），使各阶段支出不突破限制。

**为什么不选其他**
- 不选 B：那是资源平衡。
- 不选 C：增加资金不是“平衡”。
- 不选 D：那是赶工。

**出处**
PMBOK® Guide 第6版 7.3.2.5 资金限制平衡。


## 046. 项目工作绩效报告中包括如下数字：PV=110, EV=100, AC=120, 总预算为150。成本偏差是多少？（）

**选项**
- A. -20
- B. -10
- C. +10
- D. +20

**正确答案：A. -20**

**为什么选这个**
CV = EV - AC = 100 - 120 = -20；负值表示成本超支。总预算本题用不上。

**为什么不选其他**
- 不选 B：那是 SV=EV-PV=100-110=-10。
- 不选 C/D：符号或计算错误。

**出处**
PMBOK® Guide 第6版：CV=EV-AC，SV=EV-PV。


<span style="color:red">## 047. 关于项目资金需求的说法，正确的是（）。

**选项**
- A. 项目总资金需求就是项目的成本基准
- B. 通常以连续且均衡的方式投入资金
- C. 项目资金需求是一种项目文件
- D. 项目总资金需求就是项目的成本预算

**正确答案：C**

- 不选 D：成本预算/成本基准一般不含管理储备；总资金需求还要加管理储备。

PMBOK® Guide 第6版 7.3.3 制定预算：输出。


## 050. 你的朋友告诉你，他所管理的项目已经花了5000万美元，而按计划本应花费4000万美元。该项目的当前状态是（）。

**选项**
- A. 进度超前，成本超支
- B. 进度落后，成本节余
- C. 进度超前，成本节余
- D. 无法判断

**正确答案：D**

**为什么选这个**
只给了 AC 与 PV，没有 **EV**。成本看 EV-AC，进度看 EV-PV，缺 EV 无法判断。

**为什么不选其他**
- 不选 A/B/C：仅知花钱比计划多，不能断定超支还是进度超前。

**出处**
PMBOK® Guide 第6版：挣值三要素 EV/AC/PV。
