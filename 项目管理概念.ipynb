# 001. 项目生命周期

通用项目生命周期有 4 个阶段：

1. 开始项目
2. 组织与准备
3. 执行项目工作
4. 结束项目


# 002. 相互竞争的制约因素

管理项目时，需要平衡这些相互竞争的制约因素：

1. 范围
2. 进度
3. 预算
4. 质量
5. 资源
6. 风险

质量和范围、资源和风险、进度和预算都在其中，所以答案是「以上都是」。


# 003. 项目、项目集、项目组合与组织级项目管理

| | 管什么 |
| --- | --- |
| 项目管理 | 用正确的方式完成单个项目，实现项目目标 |
| 项目集管理 | 完成一系列相互关联的项目，获得单独管理得不到的更大收益 |
| 项目组合管理 | 选择最有利于实现战略目标的项目集和项目，也就是做正确的项目 |
| 组织级项目管理 | 为使项目符合组织的战略业务目标，对项目组合、项目集和项目做系统化管理 |

项目之间如果只共享顾主、供应商、技术或资源，没有必须的内在联系，当作项目组合来协调管理，不当作项目集。


# 004. 五大过程组

项目管理的五大过程组：

1. 启动
2. 规划
3. 执行
4. 监控
5. 收尾

开始项目、组织与准备、执行项目工作、结束项目是项目生命周期，不是过程组。


# 005. 工作绩效数据、信息、报告

三者是一条加工链：边干边记原始数 → 和计划对比得出结论 → 汇编成给相关方看的报告。

| | 是什么 | 用在什么场景 |
| --- | --- | --- |
| 工作绩效数据 | 执行中收集的原始资料，尚未分析 | 指导与管理项目工作时记录。如实际工时、实际成本、完成百分比、缺陷个数 |
| 工作绩效信息 | 把数据与计划对比后整理出的结论 | 控制范围、进度、成本等监控过程的输出。如进度落后 2 天、成本超支 10% |
| 工作绩效报告 | 把信息汇编成的状态报告 | 监控项目工作的输出，用于沟通和决策。如周报、仪表盘、给发起人的状态报告 |


# 006. 项目集管理关注项目间依赖

项目集管理通过处理同一个治理结构内的相关问题和变更管理，来关注项目之间的依赖关系。

| 做法 | 属于 |
| --- | --- |
| 处理同一治理结构内的问题和变更 | 项目集管理 |
| 审查项目，确定资源分配优先顺序；剔除不合理项目并补充新项目 | 项目组合管理 |
| 特别关注各项目共享的资源 | 项目管理办公室 |
