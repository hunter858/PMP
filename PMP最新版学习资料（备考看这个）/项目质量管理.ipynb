# 第八章：项目质量管理

每道题包含：题目、选项、正确答案、为什么选这个、为什么不选其他。
本节先收录已做过的题目，后续继续追加。


## 004. 在敏捷项目中，整个项目期间的质量管理由谁来执行？（）

**选项**
- A. 项目经理
- B. 质量保证部门
- C. 项目团队
- D. 质量管理专家

**正确答案：C**

**为什么选这个**
敏捷中质量由**整个项目团队**在全生命周期共同负责，贯穿开发、测试、评审与回顾。

**为什么不选其他**
- 不选 A/B/D：不是只靠项目经理、质量部门或个别专家。

**出处**
PMBOK® Guide / 敏捷实践指南：团队共同对质量负责。


## 047. 基于二八定律的帕累托图属于以下哪一种图形？（）

**选项**
- A. 因果图
- B. 散点图
- C. 直方图
- D. 矩阵图

**正确答案：C**

**为什么选这个**
帕累托图按频率从高到低排列，找造成多数问题的少数关键原因；本质是特殊的**直方图**。

**为什么不选其他**
- 不选 A：因果图（鱼骨图）分析原因类别。
- 不选 B：散点图看两个变量相关。
- 不选 D：矩阵图展示因素关系。

**出处**
PMBOK® Guide 第6版：数据表现—帕累托图。


## 048. 关于面向X的设计，以下描述不正确的是（）。

**选项**
- A. 使用这一技术旨在优化设计，以控制或提高产品的最终特性
- B. X可以是产品的某种特性，比如可靠性、可用性
- C. 使用这一技术可以提高客户满意度
- D. 使用这一技术可以降低质量成本

**正确答案：D**

**为什么选这个**
DfX 优化设计某特定方面；可得降低成本、改进质量、提高绩效与客户满意度。指南说的是降低**成本**，未把“降低质量成本”列为必然结果。

**为什么不选其他**
- 不选 A/B/C：这些符合 DfX 的正确描述。
- 本题选 D：不正确的是“可以降低质量成本”这一表述。

**出处**
PMBOK® Guide 第6版：面向 X 的设计（DfX）。


## 049. 下列哪项属于一致性质量成本的例子？（）

**选项**
- A. 返工
- B. 质量培训
- C. 废品
- D. 保修费用

**正确答案：B**

**为什么选这个**
一致性成本=预防+评估（如质量培训、测试检查）；为避免失败而花的钱。

**为什么不选其他**
- 不选 A/C：返工、废品是内部失败成本（非一致性）。
- 不选 D：保修是外部失败成本（非一致性）。

**出处**
PMBOK® Guide 第6版：质量成本（CoQ）。


## 050. 流程图中可以不包括（）。

**选项**
- A. 需要开展的活动
- B. 每个活动的责任人
- C. 相关的决策点
- D. 相关活动的处理顺序

**正确答案：B**

**为什么选这个**
流程图通常含活动、决策点、处理顺序；责任人不是必要内容（更接近 RACI/泳道图）。

**为什么不选其他**
- 不选 A/C/D：这些是流程图常见要素。

**出处**
PMBOK® Guide 第6版：流程图。

## 053. 以下哪项不是质量测量指标的例子？（）

**选项**
- A. 识别的缺陷数量，每月总停机时间
- B. 客户满意度分数，每个代码行的错误
- C. 测试覆盖度，故障率
- D. 统计抽样的频率

**正确答案：D**

**为什么选这个**
质量测量指标是要测的对象（缺陷数、停机、满意度、测试覆盖等）。抽样频率是如何做质量控制的安排，属质量管理计划，不是指标本身。

**为什么不选其他**
- 不选 A/B/C：这些都是质量测量指标例子。

**出处**
PMBOK® Guide 第6版：质量测量指标 vs 质量管理计划。


## 054. 下列关于过程分析的说法，不正确的是（）。

**选项**
- A. 过程分析可以识别出过程改进的机会
- B. 过程分析在管理质量过程中使用
- C. 需要分析在过程运行期间遇到的问题和制约因素，以及发现的非增值活动
- D. 过程分析旨在实现跨越式的改进，把产品质量做得超过要求

**正确答案：D**

**为什么选这个**
过程分析用于管理质量，找改进机会并分析问题、制约与非增值活动。质量是满足要求，不是超额；也不专门追求跨越式改进。

**为什么不选其他**
- 不选 A/B/C：这些是过程分析的正确说法。

**出处**
PMBOK® Guide 第6版 8.2 管理质量—过程分析。


## 055. 敏捷方法关注小批量试开发工作，其主要目的是（）。

**选项**
- A. 在项目早期发现和解决质量问题
- B. 降低项目开发成本
- C. 防止后续在大批量开发中出现变更
- D. 给团队成员提供锻炼的机会

**正确答案：A**

**为什么选这个**
小批量、短迭代是为了尽早交付与反馈，在早期发现并解决质量问题。

**为什么不选其他**
- 不选 B：降成本不是主要目的表述。
- 不选 C：敏捷并不以“防止变更”为目标。
- 不选 D：不是主要目的。

**出处**
PMBOK® Guide / 敏捷：小批量与早期质量反馈。


## 056. 产品交给客户使用后，团队对客户开展满意度问卷调查。问卷调查中识别的缺陷相关成本将会（）。

**选项**
- A. 被视为外部失败成本
- B. 被视为内部失败成本
- C. 对组织没有太大影响
- D. 是沉没成本

**正确答案：A**

**为什么选这个**
交付后由客户侧发现的缺陷相关成本，属**外部失败成本**。

**为什么不选其他**
- 不选 B：内部失败是交付前在组织内发现（返工、废品）。
- 不选 C/D：不符合质量成本分类。

**出处**
PMBOK® Guide 第6版：质量成本—外部失败成本。


## 057. 保证在生产过程中不出现错误，这是对下列哪个概念的解释？（）

**选项**
- A. 检查
- B. 预防
- C. 评估
- D. 补救

**正确答案：B**

**为什么选这个**
预防：保证过程中不出错，把错误挡在过程之外。检查：保证错误不交给客户。

**为什么不选其他**
- 不选 A：检查是发现错误、防止交给客户。
- 不选 C/D：不是该经典对比中的对应概念。

**出处**
PMBOK® Guide：预防优于检查。


## 058. 质量是指（）。

**选项**
- A. 符合三重制约
- B. 符合要求
- C. 优于同类产品
- D. 产品的功能众多

**正确答案：B**

**为什么选这个**
质量的核心定义是**符合要求**（克劳斯比）。

**为什么不选其他**
- 不选 A：三重制约不是质量定义。
- 不选 C/D：不是越优/功能越多越好。

**出处**
质量管理：质量=符合要求。


## 059. 以下关于项目质量管理的说法错误的是（）。

**选项**
- A. 项目质量管理既要考虑项目管理的质量，又要考虑项目产品的质量
- B. 项目质量管理同时兼顾项目过程的质量和项目结果的质量
- C. 项目质量管理方法适用于所有的项目，无论项目会产出怎样的结果
- D. 项目质量管理是为了保证项目结果达到尽可能高的质量标准

**正确答案：D**

**为什么选这个**
质量管理要满足既定要求；不是追求尽可能高的标准（易镀金浪费）。A/B/C 正确。

**为什么不选其他**
- 本题选 D：错误说法是追求尽可能高的质量标准。

**出处**
PMBOK® Guide 第6版第8章：满足要求，而非镀金。


## 060. 项目团队想通过一种更加客观和结构化的方法，找出产品存在的缺陷、漏洞和其他不合规问题，可以使用以下哪项技术？（）

**选项**
- A. 检查
- B. 审计
- C. 测试或产品评估
- D. 统计抽样

**正确答案：C**

**为什么选这个**
测试/产品评估是客观、结构化方法，用于发现产品缺陷、漏洞与不合规问题。

**为什么不选其他**
- 不选 A：检查核对是否符合标准。
- 不选 B：审计独立审查过程是否符合政策程序。
- 不选 D：统计抽样是从总体抽取部分检查。

**出处**
PMBOK® Guide 第6版 8.3 控制质量—测试与产品评估。
