# 第五章：项目范围管理

每道题包含：题目、选项、正确答案、为什么选这个、为什么不选其他。

<span style="color:red">### 001. 以下是访谈中记录的客户要求，哪一项可以作为项目的需求？（）

**选项：**

- A. 办公场所装修要高档
- B. 软件操作要便捷
- C. 成本控制在合同额的±10%范围内
- D. 购买足够客户使用的车辆

**正确答案：C. 成本控制在合同额的±10%范围内**

**为什么选这个：** 需求应可测量、可验证；±10% 可量化验收。

**为什么不选其他：**
- 不选 A：「高档」主观，不可测量。
- 不选 B：「便捷」主观，不可测量。
- 不选 D：「足够」模糊，不可测量。

**出处：** PMBOK 第六版第 5 章，需求应可测量。

### 002. 项目范围管理旨在确保（）。

**选项：**

- A. 做尽可能多的工作，使项目尽可能完美
- B. 只做为成功完成项目所需要的工作
- C. 为客户提供额外的产品功能
- D. 确定不包括在项目范围内的工作

**正确答案：B. 只做为成功完成项目所需要的工作**

**为什么选这个：** 范围管理确保只做成功完成项目所需的全部工作（不多也不少）。

**为什么不选其他：**
- 不选 A：做尽可能多会导致镀金、范围蔓延。
- 不选 C：额外功能是镀金，应避免。
- 不选 D：只说排除项不完整；核心是只做所需工作。

**出处：** PMBOK 第六版第 5 章，项目范围管理概述。

### 003. 一共9人参与需求排序的决策，最后结果显示4人同意目前的排序，5人反对。项目经理决定重新开会讨论需求的排序。这是采用了哪种决策技术？（）

**选项：**

- A. 投票
- B. 大多数同意
- C. 相对多数同意
- D. 独裁型决策制定

**正确答案：B. 大多数同意**

**为什么选这个：** 9人中需过半（至少5人）同意才通过；仅4人同意未达大多数，故重开讨论。

**为什么不选其他：**
- 不选 A：投票是大类，本题具体用的是大多数同意规则。
- 不选 C：相对多数是多方中得票最多者胜（可不过半）；本题是同意/反对过半规则。
- 不选 D：独裁是一人做决定，本题按表决结果处理。

**出处：** PMBOK 第六版第 5 章，收集需求—决策技术。

### 004. 工作分解结构底层的要素是（）。

**选项：**

- A. 控制账户
- B. 规划包
- C. 规划要素
- D. 工作包

**正确答案：D. 工作包**

**为什么选这个：** WBS 最底层是工作包。

**为什么不选其他：**
- 不选 A：控制账户在工作包之上，是管理控制点。
- 不选 B：规划包在控制账户下、工作包之上（细节未定时）。
- 不选 C：「规划要素」不是 WBS 底层标准术语。

**出处：** PMBOK 第六版第 5 章，创建 WBS。

### 005. 在适应型生命周期的项目中，用什么表示项目的当前需求？（）

**选项：**

- A. 原型
- B. 用户故事
- C. 产品未完项
- D. 产品需求文件

**正确答案：C. 产品未完项**

**为什么选这个：** 适应型项目用产品未完项（产品待办列表）表示不断演进的当前需求。

**为什么不选其他：**
- 不选 A：原型是收集/验证需求的技术。
- 不选 B：用户故事是未完项中的条目形式。
- 不选 D：产品需求文件偏预测型文档化方式。

**出处：** PMBOK 第六版第 5 章，适应型环境中的范围管理。

### 007. 在创建WBS时，为WBS组件制定和分配标识编码后的下一步工作是（）。

### 008. 下列哪项属于规划范围管理的主要作用？（）

**选项：**

- A. 为定义和管理产品范围与项目范围奠定基础
- B. 在整个项目中对如何管理范围提供指南和方向
- C. 明确项目、服务或成果的边界
- D. 在整个项目期间保持对范围基准的维护

**正确答案：B. 在整个项目中对如何管理范围提供指南和方向**

**为什么选这个：** 规划范围管理的主要作用（关键收益）是在整个项目中对如何管理范围提供指南和方向。

**为什么不选其他：**
- 不选 A：更接近收集需求——为定义范围奠定基础。
- 不选 C：更接近定义范围——明确边界。
- 不选 D：更接近控制范围——维护范围基准。

**出处：** PMBOK 第六版第 5 章，5.1 规划范围管理—主要作用。

### 009. 通过下列哪个过程监督项目和产品的范围状态，管理范围基准变更？（）

**选项：**

- A. 规划范围管理
- B. 定义范围
- C. 确认范围
- D. 控制范围

**正确答案：D. 控制范围**

**为什么选这个：** 控制范围的定义就是监督项目和产品的范围状态，并管理范围基准的变更。

**为什么不选其他：**
- 不选 A：规划范围管理——制定如何定义、确认和控制范围的计划。
- 不选 B：定义范围——制定详细的项目/产品描述。
- 不选 C：确认范围——正式验收已完成的可交付成果，不是管理基准变更。

**出处：** PMBOK 第六版第 5 章，5.6 控制范围—过程定义。

### 010. 编制工作分解结构，必须遵守下列哪项规则？（）

**选项：**

- A. 100%规则
- B. 对称规则
- C. 编号规则
- D. 一次分解到位规则

**正确答案：A. 100%规则**

**为什么选这个：** WBS 必须遵守 100% 规则——包含项目范围所定义工作的 100%，不多也不少（含项目管理）。

**为什么不选其他：**
- 不选 B：各分支不必对称到同一层，无对称规则强制要求。
- 不选 C：标识编码常用，但不是核心强制规则。
- 不选 D：应渐进明细，不必一次分解到位。

**出处：** PMBOK 第六版第 5 章 / WBS 实践标准—100% 规则。

### 011. 以下哪项不是需求管理计划中的内容？（）

**选项：**

- A. 如何规划、跟踪和报告各种需求活动
- B. 如何进行需求优先级排序
- C. 如何制定项目范围说明书
- D. 测量指标及使用这些指标的理由

**正确答案：C. 如何制定项目范围说明书**

**为什么选这个：** 制定项目范围说明书属于范围管理计划的内容，不属于需求管理计划。

**为什么不选其他：**
- 不选 A：需求活动的规划、跟踪与报告——需求管理计划内容。
- 不选 B：需求优先级排序——需求管理计划内容。
- 不选 D：测量指标及使用理由——需求管理计划内容。

**出处：** PMBOK 第六版第 5 章，5.1.3.1 范围管理计划 / 5.1.3.2 需求管理计划。

### 012. 应该在哪个过程组确认范围，以保证在可交付成果完成时及时进行验收？（）

**选项：**

- A. 执行过程组
- B. 监控过程组
- C. 收尾过程组
- D. 验收过程组

**正确答案：B. 监控过程组**

**为什么选这个：** 确认范围属于监控过程组，在可交付成果完成时及时验收，而非等到项目收尾。

**为什么不选其他：**
- 不选 A：执行过程组产出可交付成果，但不负责正式验收范围。
- 不选 C：收尾是结束项目/阶段；确认范围贯穿项目，不在收尾过程组。
- 不选 D：无“验收过程组”这一标准过程组。

**出处：** PMBOK 第六版第 5 章，确认范围—过程组归属（监控）。

### 013. 下列哪个文件用于详细描述WBS的每个组件，包括其工作内容、进度信息、成本信息等？（）

**选项：**

- A. 项目范围说明书
- B. 工作分解结构
- C. 工作分解结构词典
- D. 项目范围管理计划

**正确答案：C. 工作分解结构词典**

**为什么选这个：** WBS 词典对每个 WBS 组件做详细说明，可含工作内容、进度、成本、责任人等。

**为什么不选其他：**
- 不选 A：范围说明书描述范围、边界、验收标准等，不是逐组件明细。
- 不选 B：WBS 是分解后的层级结构，本身不含详细说明。
- 不选 D：范围管理计划说明如何管理范围，不是组件明细。

**出处：** PMBOK 第六版第 5 章，5.4.3.1—WBS 词典。

### 014. 在收集需求过程中，需要逐一、深入地了解每一个相关方的具体需求。最好使用下列哪个技术？（）

**选项：**

- A. 访谈
- B. 问卷和调查
- C. 焦点小组
- D. 引导

**正确答案：A. 访谈**

**为什么选这个：** “逐一”“深入”对应访谈——与每位相关方直接、深入了解具体需求。

**为什么不选其他：**
- 不选 B：问卷/调查适合人多、快速收集，通常不够深入。
- 不选 C：焦点小组是群体讨论，不是逐一。
- 不选 D：引导（研讨会）是多方一起达成共识，不是逐一深入。

**出处：** PMBOK 第六版第 5 章，5.2.2.2 数据收集—访谈。

### 015. 确认范围过程旨在（）。

**选项：**

- A. 审核项目范围基准的正确性
- B. 审核工作分解结构和工作分解结构词典的正确性
- C. 审核项目范围说明书的正确性
- D. 验收已经完成的可交付成果

**正确答案：D. 验收已经完成的可交付成果**

**为什么选这个：** 确认范围是正式验收已完成的可交付成果。

**为什么不选其他：**
- 不选 A/B/C：审核范围基准、WBS/词典、范围说明书是否正确，不是确认范围的目的；确认范围关注验收，不是审文档对不对。

**出处：** PMBOK 第六版第 5 章，5.5 确认范围—过程定义。

### 016. 确认范围过程与控制质量过程的关系是（）。

**选项：**

- A. 经质量检查合格的可交付成果才能进行范围确认
- B. 经确认范围过程符合要求的可交付成果才能进行质量检查
- C. 应根据具体需要，先做质量检查或先做范围确认
- D. 由客户开展质量控制，由项目团队开展范围确认

**正确答案：A. 经质量检查合格的可交付成果才能进行范围确认**

**为什么选这个：** 一般先控制质量再确认范围；核实的可交付成果是控制质量的输出、确认范围的输入。

**为什么不选其他：**
- 不选 B：顺序反了——不是先确认范围再做质量检查。
- 不选 C：通常固定为先质量后验收，不是随意谁先谁后。
- 不选 D：说反了——质量控制多由项目团队做，范围确认由客户/发起人正式验收。

**出处：** PMBOK 第六版第 5/8 章—核实的可交付成果 → 确认范围。

### 017. 某公司的项目管理信息系统升级项目，需要把旧版系统中的所有数据迁移到新版系统中。这是哪种需求的例子？（）

**选项：**

- A. 业务需求
- B. 相关方需求
- C. 过渡和就绪需求
- D. 项目需求

**正确答案：C. 过渡和就绪需求**

**为什么选这个：** 数据迁移是从现状过渡到未来状态所需的临时能力，典型过渡和就绪需求（含数据转换、培训等）。

**为什么不选其他：**
- 不选 A：业务需求偏高层为何做、业务目标/问题。
- 不选 B：相关方需求是个人/群体的需要。
- 不选 D：项目需求偏进度、成本等项目管理类要求。

**出处：** PMBOK 第六版第 5 章，5.2 收集需求—需求分类（过渡和就绪需求）。

### 018. 控制账户是工作分解结构某个层次上的要素，以便（）。

**选项：**

- A. 指定该要素的责任人或责任小组
- B. 进行挣值管理
- C. 与工作包一一对应
- D. 汇总项目的成本数据

**正确答案：B. 进行挣值管理**

**为什么选这个：** 控制账户是把范围、预算、实际成本和进度整合起来、用于挣值管理（绩效测量）的管理控制点。

**为什么不选其他：**
- 不选 A：指定责任人更接近 OBS/RAM，不是设控制账户的目的。
- 不选 C：一个控制账户通常包含多个工作包，并非一一对应。
- 不选 D：成本可在多层汇总；控制账户的核心是绩效测量/挣值，不是单纯汇总成本。

**出处：** PMBOK 第六版第 5 章，控制账户—挣值管理的控制点。

### 019. 应该如何处理已经完成但未通过验收的可交付成果？（）

**选项：**

- A. 把未通过验收的原因记录在案
- B. 提出变更请求
- C. 实施缺陷补救
- D. 说服客户改变主意，接受该可交付成果

**正确答案：A. 把未通过验收的原因记录在案**

**为什么选这个：** 确认范围时，未通过验收的可交付成果要把原因记录在案；之后可能再提变更请求做缺陷补救。

**为什么不选其他：**
- 不选 B：“可能需要”提变更请求，不是一律必须；题干对应的首要动作是记录原因。
- 不选 C：不能直接实施缺陷补救，须经变更流程批准后再实施。
- 不选 D：不能靠说服客户凑合验收。

**出处：** PMBOK 第六版第 5 章，5.5 确认范围—未验收可交付成果。

### 020. 使用引导技术召集各个部门的相关方一起定义产品需求，不利于（）。

- B. 挖掘隐藏的需求

- D. 协调需求矛盾

### 021. 关于WBS的说法，以下哪项是正确的？（）

### 023. 以下哪项是收集需求过程的输出？（）

**选项：**

- A. 范围管理计划、需求管理计划
- B. 需求管理计划、需求识别文件
- C. 需求管理计划、需求文件
- D. 需求文件、需求跟踪矩阵

**正确答案：D. 需求文件、需求跟踪矩阵**

**为什么选这个：** 收集需求的输出是需求文件和需求跟踪矩阵。

**为什么不选其他：**
- 不选 A：范围管理计划、需求管理计划是规划范围管理的输出。

### 055. 范围蔓延是指（）。
