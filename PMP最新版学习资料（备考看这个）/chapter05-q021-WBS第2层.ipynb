# 第五章：项目范围管理 · 021

关于WBS的说法，正确的是()。

- A. WBS的第2层可以是可交付成果或者项目阶段
- B. WBS的第2层可以是可交付成果或者项目部门
- C. WBS的第2层可以是可交付成果或者项目活动
- D. WBS的第2层可以是项目部门或者项目活动

## 答案

**A**

## 解析

- **A 正确**：WBS 第2层可按**可交付成果**或**项目阶段**分解。
- **B 错**：部门不属于 WBS，属于 **OBS**（组织分解结构）。
- **C 错**：活动不属于 WBS 第2层；活动在工作包之后，来自「定义活动」/活动清单。
- **D 错**：部门与活动都不作为 WBS 第2层。

**考点：** 工作分解结构的分解标准；对比 WBS vs OBS vs 活动清单。

材料常注《指南》约第 159 页。

## 对比口诀

**WBS 管事，OBS 管人，活动清单管做。**

| 结构 | 管什么 | 第2层/内容要点 |
| --- | --- | --- |
| WBS | 事（可交付成果/工作） | 第2层可为可交付成果或项目阶段 |
| OBS | 人（组织/部门） | 按组织单元分解，不是 WBS |
| 活动清单 | 做（具体活动） | 工作包之后，来自「定义活动」 |
