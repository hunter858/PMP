# 第十章：项目沟通管理

每道题包含：题目、选项、正确答案、为什么选这个、为什么不选其他。
本节先收录已做过的题目，后续继续追加。


## 001. 以下哪项是书面沟通5C原则中的内容?( )

**选项**
- A. 正确的读者(Correct Readers)
- B. 清晰的目的和表述(Clear Purpose and Expression)
- C. 正确的时间(Correct Time)
- D. 受控的篇幅(Controlled Pages)

**正确答案：B**

**为什么选这个**
书面沟通**5C**包括：正确的语法与拼写、简洁的表达、**清晰的目的与表述**、连贯的逻辑流、受控的思路流。B 属于其中之一。

**为什么不选其他**
- 不选 A：没有“正确的读者”这一条。
- 不选 C：没有“正确的时间”这一条。
- 不选 D：是 Controlling flow of words and ideas（受控的思路流），不是 Controlled Pages（受控的篇幅）。

**出处**
PMBOK® Guide 第6版 10.2 管理沟通：书面沟通的5C原则。


## 002. 在敏捷项目的每日站会上，项目团队沟通的主要内容是( )。

**选项**
- A. 前一天完成的工作和遇到的问题、当天的工作计划
- B. 为待完成的产品功能排列优先级
- C. 对当前迭代期的经验教训总结
- D. 解决当前遇到的项目问题

**正确答案：A**

**为什么选这个**
每日站会聚焦三件事：昨天完成了什么、今天计划做什么、有何障碍/问题——对应 **A**。

**为什么不选其他**
- 不选 B：产品功能排优先级属待办列表梳理/计划会，不是站会内容。
- 不选 C：经验教训总结属回顾会。
- 不选 D：站会只识别障碍，深入解决问题通常会后另开。

**出处**
敏捷实践指南 / Scrum：每日站会（Daily Standup）。


## 003. 信息发送者对下列哪一项负责?( )

**选项**
- A. 确保信息被正确接收和理解
- B. 促使信息接收者赞同信息的内容
- C. 尽量减少沟通中的噪声
- D. 确保信息清晰和完整以便被正确理解

**正确答案：D**

**为什么选这个**
发送者负责把信息**编码得清晰、完整**，便于接收者正确理解；题干问发送者职责，对应 **D**。

**为什么不选其他**
- 不选 A：正确接收并理解更偏接收者职责（接收者解码并确认理解）；发送者侧重把信息说清楚，并通过反馈确认，但本题标准表述选清晰完整。
- 不选 B：接收者不必赞同内容，沟通成功不等于认同。
- 不选 C：减少噪声是双方/环境因素，不是发送者的核心专属职责表述。

**出处**
PMBOK® Guide 第6版 10.1 / 沟通模型：发送者确保信息清晰完整。


## 004. 沟通风格评估有助于( )。

**选项**
- A. 针对不同相关方的偏好，选择合适的沟通风格
- B. 了解项目和组织的政治氛围，根据政治环境来规划沟通
- C. 了解相关方的文化差异和文化需求，根据文化差异来规划沟通
- D. 了解支持项目的相关方的沟通需求

**正确答案：A**

**为什么选这个**
**沟通风格评估**用于识别相关方偏好的沟通方式/风格，以便选择合适的沟通风格与方法。

**为什么不选其他**
- 不选 B：属政治意识（political awareness），不是沟通风格评估。
- 不选 C：属文化意识（cultural awareness）。
- 不选 D：沟通规划要覆盖各类相关方，不限于支持者；且“风格评估”侧重偏好风格，不是笼统的沟通需求清单。

**出处**
PMBOK® Guide 第6版 10.1 规划沟通管理：沟通风格评估。


## 005. 沟通管理计划通常不包括( )。

**选项**
- A. 项目相关方的沟通需求
- B. 项目主要里程碑和目标日期
- C. 接收信息的人或组织
- D. 信息分发的时限和频率

**正确答案：B**

**为什么选这个**
主要里程碑与目标日期属**进度计划/进度管理计划**内容，不是沟通管理计划的典型组成部分。

**为什么不选其他**
- 不选 A：相关方沟通需求是沟通管理计划的核心内容。
- 不选 C：信息接收人或组织属于计划中“谁接收什么信息”。
- 不选 D：分发时限与频率是沟通管理计划的常见内容。

**出处**
PMBOK® Guide 第6版 10.1.3.1 沟通管理计划。


## 006. 以下哪项是管理沟通过程的输入？( )

- A. 工作绩效数据
- B. 工作绩效信息
- C. 工作绩效报告

**正确答案：C**

- 不选 D：项目章程是**规划沟通管理**的输入。

**出处**
PMBOK® Guide 第6版 10.2 管理沟通：输入含工作绩效报告。


## 007. 了解组织中谁能对项目施加较大影响，并重点培养与这些相关方的沟通能力，这属于哪种人际关系与团队技能？( )

**选项**
- A. 沟通风格评估
- B. 政治意识
- C. 文化意识
- D. 人际交往

**正确答案：B**

**为什么选这个**
识别谁有较大影响力并针对性加强沟通，属于**政治意识（Political awareness）**——认清权力结构与政治环境。

**为什么不选其他**
- 不选 A：沟通风格评估关注偏好的沟通风格/方式。
- 不选 C：文化意识关注文化差异与需求。
- 不选 D：人际交往更泛指建立关系网络，题干强调“谁有影响力”的政治判断。

**出处**
PMBOK® Guide 第6版 10.1 规划沟通管理：人际关系与团队技能——政治意识。


## 008. 在项目沟通中，谁负责确保信息的清楚、明确和完整？( )

**选项**
- A. 项目经理
- B. 信息发送者
- C. 信息接收者
- D. 沟通双方

**正确答案：B**

**为什么选这个**
沟通模型中，**信息发送者**负责把信息编码得清楚、明确、完整，便于正确理解。

**为什么不选其他**
- 不选 A：项目经理不一定是每条信息的发送者。
- 不选 C：接收者负责正确接收、理解并确认。
- 不选 D：清晰完整主要是发送者编码职责，不是双方共同编码。

**出处**
PMBOK® Guide 第6版 10.1 / 沟通模型：发送者职责。


## 009. 使用以下哪种工具与技术，既可以有效收集和发布信息，又有助于监控项目的沟通情况？( )

**选项**
- A. 专家判断
- B. 项目报告发布
- C. 会议
- D. 项目管理信息系统

**正确答案：D**

**为什么选这个**
**项目管理信息系统（PMIS）**既能收集与发布项目信息，也可用于跟踪、存储与检索沟通记录，从而支持对沟通情况的监控。

**为什么不选其他**
- 不选 A：专家判断用于规划或决策建议，不兼具系统化收集发布与监控沟通。
- 不选 B：项目报告发布侧重分发报告，本身不是监控沟通的综合工具。
- 不选 C：会议可交流信息，但不等同于可持续收集发布并监控沟通状况的工具。

**出处**
PMBOK® Guide 第6版 10.2 / 10.3：管理沟通与监督沟通中的项目管理信息系统。


## 010. 某一团队成员用难懂的方言和其他人进行工作交谈，其他人表示无法和他进行有效沟通，项目经理要求该成员使用普通话交谈。项目经理这样做的原因是( )。

**选项**
- A. 对该团队成员有偏见
- B. 为了团队成员彼此团结
- C. 语言差异是沟通中的噪声，会影响沟通质量
- D. 沟通渠道太少

**正确答案：C**

**为什么选这个**
方言造成理解障碍，属于沟通模型中的**噪声**，干扰信息传递与理解；要求使用普通话是为减少噪声、提高沟通质量。

**为什么不选其他**
- 不选 A：要求统一工作语言是为有效沟通，不是偏见。
- 不选 B：目的不是笼统的“团结”，而是消除语言噪声。
- 不选 D：问题是语言障碍（噪声），不是沟通渠道数量不足。

**出处**
PMBOK® Guide 第6版 沟通模型：噪声（语言差异等）。

## 013. 进行项目沟通需求分析，旨在确定( )。

**选项**
- A. 能用于沟通的时间和资金多少
- B. 所需信息的类型和格式，以及信息对相关方的价值
- C. 可使用的沟通技术
- D. 沟通渠道的多少

**正确答案：B**

**为什么选这个**
**沟通需求分析**旨在确定相关方所需信息的**类型与格式**，以及该信息对相关方的**价值**。

**为什么不选其他**
- 不选 A：时间与资金约束会影响沟通规划，但不是需求分析的核心目的表述。
- 不选 C：沟通技术是另一项工具，在确定需求后选择。
- 不选 D：渠道数量可用公式计算，属于规划时考虑因素，不是需求分析的定义目标。

**出处**
PMBOK® Guide 第6版 10.1.2.3 沟通需求分析。


## 014. 你管理一个大型的跨国项目，项目团队有35名成员，后来又增加了6名新成员。这个情况引起了你的高度关注，因为( )。

**选项**
- A. 沟通管理的难度会增加
- B. 文化融合会变得更难
- C. 沟通阻碍者会增加
- D. 导致冲突的增加

**正确答案：A**

**为什么选这个**
成员从35增至41，沟通渠道数按 $n(n-1)/2$ 大幅增加（由595增至820），**沟通管理难度显著上升**，这是最直接、最标准的关注点。

**为什么不选其他**
- 不选 B：跨国项目文化问题本就存在，增员不一定是“高度关注”的核心考点。
- 不选 C：“沟通阻碍者”不是因人数增加的必然结论。
- 不选 D：冲突可能增加，但本题对应的是沟通渠道/复杂度考点。

**出处**
PMBOK® Guide 第6版 10.1：沟通渠道公式 $n(n-1)/2$。


## 015. 项目出现例外情况，项目经理应该使用哪个技术来编制和发布临时报告，以便及时与相关方沟通？( )

**选项**
- A. 项目管理信息系统
- B. 项目报告发布
- C. 会议
- D. 沟通技术

**正确答案：B**

**为什么选这个**
**项目报告发布（Project reporting）**负责收集并分发项目信息，报告可定期编制，也可在**例外情况**下临时编制发布。

**为什么不选其他**
- 不选 A：PMIS 是支撑收集/分发的系统平台，题干问的是编制发布临时报告的技术本身。
- 不选 C：会议可用于沟通，但不是“编制和发布临时报告”的对应技术。
- 不选 D：沟通技术指媒介/手段选择，范围更宽，不特指例外临时报告。

**出处**
PMBOK® Guide 第6版 10.2.2.5 项目报告发布。


## 016. 与相关方有关的敏感性和保密性信息，在沟通管理中应该如何处理？( )

**选项**
- A. 列入沟通管理的相关文件中
- B. 采取安全措施并选择合适的沟通技术
- C. 向发起人请示处理方法
- D. 与客户沟通处理方法

**正确答案：B**

**为什么选这个**
敏感/保密信息应**采取安全措施**，并据此**选择合适的沟通技术**（如加密、受限访问渠道等）。

**为什么不选其他**
- 不选 A：敏感信息不宜随意写入可广泛分发的沟通管理文件。
- 不选 C/D：不一定都要先请示发起人或与客户商量处理方法；标准做法是按安全要求选技术并管控。

**出处**
PMBOK® Guide 第6版 10.1.2.4 沟通技术：信息敏感性与保密性。


## 017. 在规划沟通管理中，分析相关方参与度评估矩阵的目的是( )。

**选项**
- A. 分析相关方参与程度的不足，识别额外的沟通需求
- B. 分析相关方参与程度的变化，对沟通进行必要调整
- C. 制定合适的相关方管理策略

**正确答案：A**

规划沟通管理中使用**相关方参与度评估矩阵**，对比当前与期望参与度，分析差距（不足），从而**识别额外的沟通需求**。

**为什么不选其他**
- 不选 B：根据参与度变化调整沟通更偏监督沟通/管理相关方参与。
- 不选 C：制定相关方管理策略主要属规划相关方参与。
- 不选 D：沟通活动效果信息更偏监督沟通的输出/作用。

**出处**
PMBOK® Guide 第6版 10.1.2.5 数据表现：相关方参与度评估矩阵。


## 018. 在沟通管理中，既强调向正确的人提供正确的信息，也强调只提供所需要的信息。前者和后者分别是指( )。

**选项**
- A. 及时的沟通，充分的沟通
- B. 有效率的沟通，有效果的沟通
- C. 充分的沟通，及时的沟通
- D. 有效果的沟通，有效率的沟通

**正确答案：D**

**为什么选这个**
向正确的人提供正确信息 = **有效果的沟通（effective）**；只提供所需信息 = **有效率的沟通（efficient）**。

**为什么不选其他**
- 不选 B：把有效率与有效果顺序颠倒了。
- 不选 A/C：及时、充分不是本题这对概念的标准对应。

**出处**
PMBOK® Guide 第6版第10章引言：有效果 vs 有效率的沟通。


## 019. 在两方或多方之间进行的实时多向信息交换，属于以下哪种沟通方法？( )

**选项**
- A. 交互式沟通
- B. 推式沟通
- C. 拉式沟通
- D. 电话沟通

**正确答案：A**

**为什么选这个**
两方或多方之间的**实时多向信息交换**，定义即**交互式沟通**（会议、电话、即时消息等）。

**为什么不选其他**
- 不选 B：推式是把信息推送给接收方（邮件、报告），不保证理解。
- 不选 C：拉式是接收方自行获取（门户、知识库）。
- 不选 D：电话只是交互式沟通的一种具体形式，不是方法分类名称。

**出处**
PMBOK® Guide 第6版 10.1.2.5 沟通方法。


## 020. 无论哪种形式的沟通都需要遵守5C原则。以下哪项不属于5C原则?( )

**选项**
- A. 目的明确(Clear Purpose)
- B. 表达正确(Correct Expression)
- C. 逻辑连贯(Coherent Logic)
- D. 关心对方(Care for Each Other)

**正确答案：D**

**为什么选这个**
书面沟通5C为：正确、简洁、清晰（目的明确）、连贯、受控的思路流。**关心对方**不属于5C原则。

**为什么不选其他**
- 不选 A：清晰的目的和表述属于5C。
- 不选 B：正确的语法与拼写/表达正确属于5C。
- 不选 C：连贯的逻辑流属于5C。

**出处**
PMBOK® Guide 第6版 10.2：书面沟通的5C原则。


## 021. 以下哪个是沟通工件？( )

**选项**
- A. 沟通模型
- B. 推式沟通
- C. 电子邮件
- D. 沟通渠道

**为什么不选其他**
- 不选 A：沟通模型是工具概念，不是工件。
- 不选 B：推式沟通是沟通方法。
- 不选 D：沟通渠道是路径/数量概念，不是工件。

**出处**
PMBOK® Guide / 沟通管理：沟通工件 vs 方法、模型、渠道。


## 022. 以下哪项最好地描述了事业环境因素对规划沟通管理过程的影响？( )

**选项**
- A. 项目执行组织的组织文化对项目的沟通需求有重要影响
- B. 项目执行组织的标准化工作流程对项目的沟通需求有重要影响

PMBOK® Guide 第6版 10.1：沟通渠道 $n(n-1)/2$。

- 不选 C：题干写一切井然有序，与“冲突太多”矛盾。

**出处**
